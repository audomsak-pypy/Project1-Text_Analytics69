## เทคนิคที่อาจใช้ — เลือกเท่าที่จำเป็น

ไม่ต้องใช้ทั้งหมด โดยทั่วไป **2–4 เทคนิคก็เพียงพอ** ให้เลือกตามคำถามของทีม

| เทคนิค | เหมาะเมื่ออยากรู้ |
|---|---|
| **Word / n-gram frequency** | คนพูดถึงอะไรบ่อย |
| **TF-IDF / distinctive words** | สองกลุ่มพูดต่างกันอย่างไร |

| **Trend / segmentation** | ปัญหาเปลี่ยนตามเวลา หรือหนักในกลุ่มไหน |

> เทคนิคขั้นสูง เช่น embedding, clustering, regression, classification หรือ anomaly detection ทำได้ แต่ **ไม่บังคับ** และเหมาะสำหรับ Bonus เมื่อใช้ถูกและอธิบายได้

## แบบฟอร์ม Insight Card

ทุก insight เขียนในรูปแบบนี้ 5 บรรทัด ไม่ต้องยาว

> **Insight:** ประโยคเดียวที่บอกสิ่งที่พบ ต้องมีตัวเลขอยู่ในประโยค \
> **หลักฐาน:** ตัวเลขพร้อม n และชี้ว่าดูได้จากกราฟหรือตารางไหน \
> **So what:** กระทบธุรกิจอย่างไร ถ้าประมาณขนาดผลกระทบเป็นตัวเลขได้ยิ่งดี \
> **Action:** ทีมไหนควรทำอะไร และจะรู้ได้อย่างไรว่าได้ผล \
> **ความมั่นใจ:** สูง กลาง หรือ ต่ำ พร้อมเหตุผลสั้น ๆ

**ตัวอย่างที่ยังไม่ใช่ insight**
> ลูกค้าพูดถึงเรื่องราคาบ่อยที่สุด คิดเป็น 32% ของรีวิว

**ตัวอย่างที่ใช้ได้**
> **Insight:** รีวิวที่พูดถึงบรรจุภัณฑ์รั่วมีเพียง 6.2% ของทั้งหมด (n = 3,100) แต่ได้ดาวเฉลี่ย 2.1 เทียบกับค่าเฉลี่ยรวม 4.0 และกระจุกอยู่ในสินค้า 12 ชิ้นที่ใช้ขวดแบบปั๊ม \
> **หลักฐาน:** ตารางค่าเฉลี่ยดาวแยกตามด้าน และกราฟแท่งสินค้า 12 ชิ้นในผลวิเคราะห์ \
> **So what:** สินค้า 12 ชิ้นนี้คิดเป็น 18% ของยอดรีวิวทั้งหมวด ถ้าแก้ปัญหาฝาขวดได้ คาดว่าคะแนนเฉลี่ยของหมวดจะขยับขึ้นราว 0.15 ดาว \
> **Action:** ทีม Product ตรวจสอบซัพพลายเออร์ฝาขวดของสินค้า 12 รายการนี้ภายในไตรมาสหน้า แล้วติดตามสัดส่วนรีวิวที่มีคำว่า leak, spill, broken cap ซึ่งตอนนี้อยู่ที่ 14% ต่อเดือน \
> **ความมั่นใจ:** กลาง ยืนยันได้จากข้อความ แต่ยังไม่มีข้อมูลการคืนสินค้ามาช่วยยืนยัน

````python
# ===== Hacker News API (JSON,) =====
# ได้หัวข้อข่าวพร้อม score และจำนวนคอมเมนต์ เหมาะกับการดูว่าเรื่องแบบไหนคนสนใจ
import urllib.request, json, time
import pandas as pd

def hn_get(url, timeout=30):
    with urllib.request.urlopen(url, timeout=timeout) as r:
        return json.loads(r.read().decode("utf-8"))

def fetch_hackernews(n=250, sleep=0.1):
    base = "https://hacker-news.firebaseio.com/v0"

    ids = hn_get(f"{base}/topstories.json")[:n]
    rows = []
    for i, item_id in enumerate(ids):
        try:
            it = hn_get(f"{base}/item/{item_id}.json")
        except Exception as e:
            print("ข้าม", item_id, e)
            continue
        if not it or it.get("type") != "story" or not it.get("title"):
            continue
        rows.append({
            "doc_id": str(it["id"]),
            "text": it["title"] + (" " + it.get("text", "") if it.get("text") else ""),
            "score": it.get("score", 0),
            "n_comments": it.get("descendants", 0),
            "author": it.get("by"),
            "published_at": pd.to_datetime(it.get("time", 0), unit="s").isoformat(),
            "url": f"https://news.ycombinator.com/item?id={it['id']}",
            "source": "hackernews",
        })
        time.sleep(sleep)
        if (i + 1) % 100 == 0:
            print(f"ดึงแล้ว {i+1} รายการ")
    return rows

# rows = fetch_hackernews(50)
# print(len(rows)); rows[0]
````

````python
import urllib.request
import json
import pandas as pd

def fetch_political_news(query="politics", hits_per_page=250):
    # Endpoint สำหรับค้นหาข่าวการเมืองโดยเฉพาะ (ไม่จำต้องใช้ API Key)
    url = f"https://hn.algolia.com/api/v1/search?query={query}&tags=story&hitsPerPage={hits_per_page}"

    req = urllib.request.Request(url, headers={'User-Agent': 'Mozilla/5.0'})
    with urllib.request.urlopen(req) as response:
        data = json.loads(response.read().decode("utf-8"))

    rows = []
    for item in data.get("hits", []):
        rows.append({
            "doc_id": str(item.get("objectID")),
            "text": item.get("title", ""),
            "score": item.get("points", 0),
            "n_comments": item.get("num_comments", 0),
            "author": item.get("author"),
            "published_at": item.get("created_at"),
            "url": item.get("url"),
            "source": "algolia_hn"
        })
    return rows

# 1. นำเข้าข้อมูลผ่าน API
raw_data = fetch_political_news(query="politics", hits_per_page=250)
df = pd.DataFrame(raw_data)

print(f"จำนวน Valid Records : {len(df)} รายการ")
````

  * **ข้อจำกัด (Limitation):** ข้อมูลมาจาก Hacker Newsซึ่งสะท้อนมุมมองของคนสาย Tech ฝั่งตะวันตกเป็นหลัก ไม่สามารถนำไปสรุปอ้างอิงถึงประชากรทั่วไปได้ (ไม่ Overclaim)



````python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.linear_model import LinearRegression

# ==============================================================================
# BONUS PART 1: Creativity / Interesting Question (+2 คะแนน)
# การวิเคราะห์ Discussion Ratio และจำแนกหมวดหมู่ข่าว (Tech Policy vs General)
# ==============================================================================

# 1. คำนวณ Discussion Ratio = n_comments / (score + 1)
# (อัตราส่วนการคอมเมนต์ต่อโหวต: ค่าน้อย = แค่กดโหวต, ค่ามาก = เกิดการถกเถียงดุเดือด)
df['discussion_ratio'] = (df['n_comments'] / (df['score'] + 1)).round(3)

# 2. จำแนกหมวดหมู่ข่าวด้วย Keyword Pattern (Tech Policy vs General Politics)
tech_keywords = r'ai|tech|crypto|privacy|regulation|sec|monopoly|copyright|law|policy|court'
df['category'] = np.where(
    df['text'].str.contains(tech_keywords, case=False, na=False),
    'Tech Policy & Regulation',
    'General Politics'
)

# เปรียบเทียบค่าเฉลี่ยตามเกณฑ์ Creativity
cat_summary = df.groupby('category').agg(
    total_posts=('doc_id', 'count'),
    avg_score=('score', 'mean'),
    avg_comments=('n_comments', 'mean'),
    avg_discussion_ratio=('discussion_ratio', 'mean')
).reset_index()

print("=== 1. CREATIVITY: COMPARATIVE ANALYSIS ===")
print(cat_summary.to_string(index=False))
print("\n" + "="*60 + "\n")


# ==============================================================================
# BONUS PART 2: Advanced Technique (+2 คะแนน)
# Topic Clustering (KMeans) + Regression Analysis (Linear Regression)
# ==============================================================================

# 2.1 Advanced NLP Clustering: จัดกลุ่มหัวข้อข่าวด้วย TF-IDF + KMeans (k=3)
tfidf = TfidfVectorizer(stop_words='english', max_features=100)
X_tfidf = tfidf.fit_transform(df['text'])

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['topic_cluster'] = kmeans.fit_predict(X_tfidf)

# 2.2 Regression Analysis: หาความสัมพันธ์ว่า score ส่งผลต่อ n_comments อย่างไร
X_reg = df[['score']]
y_reg = df['n_comments']

model = LinearRegression()
model.fit(X_reg, y_reg)

r2_score = model.score(X_reg, y_reg)
coef = model.coef_[0]

print("=== 2. ADVANCED TECHNIQUE: CLUSTERING & REGRESSION ===")
print(f"R-squared Score (การอธิบายความแปรปรวน): {r2_score:.4f}")
print(f"Regression Coefficient (ทุกๆ 1 Upvote คาดว่าจะเพิ่มคอมเมนต์): {coef:.2f} คอมเมนต์")
print("\n" + "="*60 + "\n")


# ==============================================================================
# BONUS PART 3: Exceptional Business Value (+1 คะแนน)
# การสร้าง Engagement Priority Score (KPI) สำหรับดักจับกระแสข่าวร้อน
# ==============================================================================

# 1. Normalize ค่า score และ n_comments ให้อยู่ในช่วง 0 - 1 (Min-Max Scaling)
min_s, max_s = df['score'].min(), df['score'].max()
min_c, max_c = df['n_comments'].min(), df['n_comments'].max()

norm_score = (df['score'] - min_s) / (max_s - min_s + 1e-6)
norm_comments = (df['n_comments'] - min_c) / (max_c - min_c + 1e-6)

# 2. คำนวณ Priority Score (0 - 100) โดยให้น้ำหนักการคอมเมนต์ (Discussion) 60% และโหวต 40%
df['priority_score'] = ((0.4 * norm_score) + (0.6 * norm_comments)) * 100
df['priority_score'] = df['priority_score'].round(2)

# 3. จัดลำดับ Top 5 ข่าวที่ต้องโฟกัสก่อน (Actionable Output สำหรับบรรณาธิการ/ทีมนโยบาย)
top_actionable_news = df[[
    'doc_id', 'text', 'category', 'score', 'n_comments',
    'discussion_ratio', 'priority_score'
]].sort_values(by='priority_score', ascending=False).head(5)

print("=== 3. BUSINESS VALUE: TOP 5 PRIORITY NEWS DASHBOARD ===")
print(top_actionable_news.to_string(index=False))
````

````python
import pandas as pd
import json
# --- 1. เริ่มกระบวนการทำความสะอาดและนับจำนวน ---
initial_count = len(df)

# ตัด Missing Values (ไม่มีหัวข้อข่าว)
df_clean = df.dropna(subset=['text'])
missing_removed = initial_count - len(df_clean)

# ตัด Duplicates (ข้อมูลซ้ำเช็คจาก doc_id)
temp_count = len(df_clean)
df_clean = df_clean.drop_duplicates(subset=['doc_id'])
duplicate_removed = temp_count - len(df_clean)

# ตัดข้อความที่สั้นผิดปกติ (น้อยกว่า 15 ตัวอักษร) หรือคะแนนติดลบ
temp_count = len(df_clean)
df_clean = df_clean[(df_clean['text'].str.len() >= 15) & (df_clean['score'] >= 0)]
invalid_removed = temp_count - len(df_clean)

final_count = len(df_clean)

# 1. แปลงข้อมูลใน DataFrame ให้กลับเป็น List of Dictionaries
cleaned_rows = df_clean.to_dict('records')

# 2. บันทึกไฟล์ (ใช้โค้ดตามที่ระบุ)
with open("corpus.jsonl", "w", encoding="utf-8") as f:
    for d in cleaned_rows:
        f.write(json.dumps(d, ensure_ascii=False) + "\n")
print("บันทึกไฟล์ 'corpus.jsonl' สำเร็จแล้ว")

# 3. อ่านกลับมาใช้
corp = pd.read_json("corpus.jsonl", lines=True)

# ทดสอบแสดงผลข้อมูลที่โหลดกลับมาเพื่อยืนยันความถูกต้อง
print("\n--- ข้อมูลที่อ่านกลับมาจาก corpus.jsonl ---")
print(f"จำนวนข้อมูลทั้งหมด: {len(corp)} records")
print(corp.head(3)) # ดูตัวอย่าง 3 แถวแรก
````

````python
print(f"\n--- สรุปคุณภาพข้อมูล ---")
print(f"พยายามเก็บทั้งหมด: {initial_count} records")
print(f"ตัด Missing Values (ไม่มีหัวข้อข่าว): {missing_removed} records")
print(f"ตัด Duplicates (จาก doc_id): {duplicate_removed} records")
print(f"ตัดข้อความที่สั้นผิดปกติ (น้อยกว่า 15 ตัวอักษร) หรือคะแนนติดลบ: {invalid_removed} records")
print(f"ใช้งานได้จริง (Final corpus) เหลือ: {final_count} records")

# หาช่วงเวลาที่ครอบคลุม
if not corp.empty:
    corp['published_at'] = pd.to_datetime(corp['published_at'])
    min_date = corp['published_at'].min().strftime('%Y-%m-%d')
    max_date = corp['published_at'].max().strftime('%Y-%m-%d')
    print(f"ครอบคลุมช่วงเวลา: ตั้งแต่ {min_date} ถึง {max_date}")
else:
    print("ไม่สามารถระบุช่วงเวลาได้เนื่องจากไม่มีข้อมูลใน Final corpus")

print(f"หมวดหมู่ที่ครอบคลุม: Political Story (ตาม query: 'politics')")
````

### 3.3 หา Insight

ใช้ **1–2 เทคนิค** ที่เหมาะกับคำถาม เช่น

* frequency / n-gram
* sentiment
* topic
* comparison ระหว่างกลุ่ม
* trend ตามเวลา

สร้างกราฟหรือตารางหลักอย่างน้อย **1 ชิ้น** แล้วตอบว่า

> “จากข้อมูลนี้ ผู้ใช้ผลวิเคราะห์ควรรู้อะไรหรือทำอะไรต่อ?”


````python
# ----------------- Your code here -----------------
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import CountVectorizer

# 1. โหลดข้อมูลจากไฟล์ที่บันทึกไว้
corp = pd.read_json("corpus.jsonl", lines=True)

# ---------------------------------------------------------
# เทคนิคที่ 1: Frequency Analysis (หาความถี่ของคำศัพท์ในหัวข้อข่าว)
# ใช้ CountVectorizer ตัดคำที่เป็นจุดเชื่อมภาษาอังกฤษ (stop words) เช่น the, in, on ออก
vectorizer = CountVectorizer(stop_words='english', ngram_range=(1, 2), max_features=20)
X = vectorizer.fit_transform(corp['text'])

# สร้าง DataFrame เพื่อเรียงลำดับคำที่เจอมากที่สุด
word_freq = pd.DataFrame({
    'Keyword': vectorizer.get_feature_names_out(),
    'Frequency': X.toarray().sum(axis=0)
}).sort_values(by='Frequency', ascending=False)
````

````python
# ----------------- Your code here -----------------
# กราฟสำหรับนำเสนอ
# --- สร้างกราฟ Bar Chart 10 อันดับแรก ---
plt.figure(figsize=(10, 6))
plt.bar(word_freq['Keyword'][:10], word_freq['Frequency'][:10], color='#4C72B0')
plt.title('Top 10 Keywords in Political News (Hacker News)')
plt.xlabel('Keywords')
plt.ylabel('Frequency (จำนวนหัวข้อข่าว)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('political_keywords.png') # เซฟเป็นรูปภาพไปใส่รายงาน
plt.show()

# ---------------------------------------------------------
# เทคนิคที่ 2: Comparison (เปรียบเทียบ Engagement หา 5 อันดับที่คนถกเถียงสูงสุด)
top_engaged = corp.sort_values(by='n_comments', ascending=False).head(5)
print("\n--- 5 อันดับหัวข้อข่าวที่มีการคอมเมนต์ (ถกเถียง) สูงสุด ---")
print(top_engaged[['text', 'n_comments', 'score']])
````
