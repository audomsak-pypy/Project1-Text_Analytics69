## เทคนิคที่อาจใช้ — เลือกเท่าที่จำเป็น

ไม่ต้องใช้ทั้งหมด โดยทั่วไป **2–4 เทคนิคก็เพียงพอ** ให้เลือกตามคำถามของทีม

| เทคนิค | เหมาะเมื่ออยากรู้ |
|---|---|
| **Word / n-gram frequency** | คนพูดถึงอะไรบ่อย |
| **TF-IDF / distinctive words** | สองกลุ่มพูดต่างกันอย่างไร |
| **Sentiment** | ข้อความเป็นบวก/ลบเพียงใด |
| **Aspect analysis** | คนพูดถึงราคา คุณภาพ การส่งของ ฯลฯ อย่างไร |
| **Topic modeling** | มีประเด็นอะไรที่เราอาจยังไม่รู้ |
| **Trend / segmentation** | ปัญหาเปลี่ยนตามเวลา หรือหนักในกลุ่มไหน |

> เทคนิคขั้นสูง เช่น embedding, clustering, regression, classification หรือ anomaly detection ทำได้ แต่ **ไม่บังคับ** และเหมาะสำหรับ Bonus เมื่อใช้ถูกและอธิบายได้

ถ้ายังไม่เคยใช้เทคนิคใด ให้ทดลองกับข้อมูลชุดเล็กก่อน แล้วตรวจด้วยตาว่าผลสมเหตุสมผลหรือไม่


## แบบฟอร์ม Insight Card

ทุก insight เขียนในรูปแบบนี้ 5 บรรทัด ไม่ต้องยาว

> **Insight:** ประโยคเดียวที่บอกสิ่งที่พบ ต้องมีตัวเลขอยู่ในประโยค \
> **หลักฐาน:** ตัวเลขพร้อม n และชี้ว่าดูได้จากกราฟหรือตารางไหน \
> **So what:** กระทบธุรกิจอย่างไร ถ้าประมาณขนาดผลกระทบเป็นตัวเลขได้ยิ่งดี \
> **Action:** ทีมไหนควรทำอะไร และจะรู้ได้อย่างไรว่าได้ผล \
> **ความมั่นใจ:** สูง กลาง หรือ ต่ำ พร้อมเหตุผลสั้น ๆ

**ตัวอย่างที่ยังไม่ใช่ insight**
> ลูกค้าพูดถึงเรื่องราคาบ่อยที่สุด คิดเป็น 32% ของรีวิว

เป็นเพียงข้อสังเกต ไม่ได้บอกว่าใครต้องทำอะไร และ 32% นั้นมากหรือน้อยก็ไม่รู้เพราะไม่มีอะไรให้เทียบ

**ตัวอย่างที่ใช้ได้**
> **Insight:** รีวิวที่พูดถึงบรรจุภัณฑ์รั่วมีเพียง 6.2% ของทั้งหมด (n = 3,100) แต่ได้ดาวเฉลี่ย 2.1 เทียบกับค่าเฉลี่ยรวม 4.0 และกระจุกอยู่ในสินค้า 12 ชิ้นที่ใช้ขวดแบบปั๊ม \
> **หลักฐาน:** ตารางค่าเฉลี่ยดาวแยกตามด้าน และกราฟแท่งสินค้า 12 ชิ้นในผลวิเคราะห์ \
> **So what:** สินค้า 12 ชิ้นนี้คิดเป็น 18% ของยอดรีวิวทั้งหมวด ถ้าแก้ปัญหาฝาขวดได้ คาดว่าคะแนนเฉลี่ยของหมวดจะขยับขึ้นราว 0.15 ดาว \
> **Action:** ทีม Product ตรวจสอบซัพพลายเออร์ฝาขวดของสินค้า 12 รายการนี้ภายในไตรมาสหน้า แล้วติดตามสัดส่วนรีวิวที่มีคำว่า leak, spill, broken cap ซึ่งตอนนี้อยู่ที่ 14% ต่อเดือน \
> **ความมั่นใจ:** กลาง ยืนยันได้จากข้อความ แต่ยังไม่มีข้อมูลการคืนสินค้ามาช่วยยืนยัน

### 2.2 เตรียมข้อมูลที่เชื่อถือได้

ทำเพียงสิ่งจำเป็นก่อนเริ่มวิเคราะห์

* สุ่มรีวิวอย่างน้อย **20,000 รายการ** และอย่าใช้เพียง N บรรทัดแรก
* เชื่อม review กับ product metadata ด้วย `parent_asin`
* รายงานสั้น ๆ ว่า join สำเร็จกี่เปอร์เซ็นต์ และมี missing/duplicate สำคัญหรือไม่
* เขียน 2–3 บรรทัดว่า **ข้อมูลนี้เป็นตัวแทนของใคร และไม่เป็นตัวแทนของใคร**

> จุดประสงค์คือให้รู้ว่าข้อมูลที่ใช้เชื่อถือได้พอสำหรับคำถามของทีม ไม่ใช่ทำ data cleaning ทุกอย่างที่เป็นไปได้


````python
def iter_jsonl(path):
    with open(path, encoding='utf-8') as f:
        for line in f:
            yield json.loads(line)
````

### Sampling Reviews from `review_path`

````python
# Sample N reviews from the review_path
print(f"Sampling {SAMPLE_N:,} reviews from {review_path}...")

sampled_reviews = []
for i, review in enumerate(iter_jsonl(review_path)):
    if len(sampled_reviews) < SAMPLE_N:
        sampled_reviews.append(review)
    else:
        # Reservoir sampling to get a random sample of N reviews
        r = random.randint(0, i)
        if r < SAMPLE_N:
            sampled_reviews[r] = review

df_reviews = pd.DataFrame(sampled_reviews)
print(f"Loaded {len(df_reviews):,} sampled reviews.")

# Load all meta data
print(f"Loading metadata from {meta_path}...")
df_meta = pd.DataFrame(iter_jsonl(meta_path))
print(f"Loaded {len(df_meta):,} metadata items.")

display(df_reviews.head())
````

### Joining Reviews with Product Metadata

````python
# Perform a left merge to enrich reviews with product metadata
initial_review_count = len(df_reviews)
df_joined = pd.merge(df_reviews, df_meta, on='parent_asin', how='left', suffixes=('_review', '_meta'))

# Report join success percentage
successful_join_count = len(df_joined.dropna(subset=['main_category'])) # Assuming main_category is a good indicator of a successful meta join
join_percentage = (successful_join_count / initial_review_count) * 100
print(f"Successfully joined {successful_join_count:,} reviews with metadata ({join_percentage:.2f}%).")

# Check for missing values in critical columns
print("\nMissing values after join:")
missing_values = df_joined[['text', 'title_review', 'main_category', 'price']].isnull().sum()
display(missing_values[missing_values > 0])

# Check for duplicate reviews (based on a combination of user_id, timestamp, and asin)
# Note: parent_asin is the product group, asin is the specific product variant.
# We are interested in unique reviews, which is usually identified by user_id and timestamp for a product.
df_joined_before_dedup = len(df_joined)
df_joined = df_joined.drop_duplicates(subset=['user_id', 'timestamp', 'asin'])
duplicate_removed_count = df_joined_before_dedup - len(df_joined)
print(f"\nRemoved {duplicate_removed_count:,} duplicate reviews.")

display(df_joined.head())
````

### ความเป็นตัวแทนของข้อมูล (Data Representativeness)
ชุดข้อมูลนี้เป็นตัวแทนของรีวิวจำนวน 20,000 รายการ ที่ถูกสุ่มเลือกมาจากสินค้าในหมวดหมู่ 'Digital_Music' บน Amazon โดยรีวิวเหล่านี้รวมถึงข้อมูลอธิบายเพิ่มเติม (Metadata) ที่เกี่ยวข้อง เช่น ชื่อสินค้า หมวดหมู่ และคะแนนดาวเฉลี่ย ข้อมูลตัวอย่างนี้ช่วยให้เห็นภาพรวมความรู้สึกของลูกค้า (Customer Sentiment) และลักษณะเฉพาะของสินค้าในหมวดหมู่นี้

อย่างไรก็ตาม ชุดข้อมูลนี้ไม่ได้เป็นตัวแทนของรีวิวทั้งหมดบน Amazon (เช่น สินค้าในหมวดหมู่อื่นๆ หรือรีวิวเก่าที่ไม่ได้รวมอยู่ในชุดข้อมูลปี 2023) และไม่ได้เป็นตัวแทนของประชากรรีวิวทั้งหมดในหมวด 'Digital_Music' (เนื่องจากเป็นเพียงกลุ่มตัวอย่าง ไม่ใช่ประชากรทั้งหมด) นอกจากนี้ อาจไม่ได้เป็นตัวแทนของสินค้ากลุ่มเฉพาะ (Niche) หรือสินค้าที่ได้รับความนิยมสูงเป็นพิเศษได้อย่างสมบูรณ์ หากการสุ่มตัวอย่างไม่ได้ครอบคลุมสินค้าเหล่านั้นในสัดส่วนที่เหมาะสม

### 2.3 สำรวจข้อมูลเท่าที่จำเป็น

เลือกดูเฉพาะสิ่งที่ช่วยให้เข้าใจ brief ของทีม เช่น

* การแจกแจงดาว
* verified purchase / helpful vote
* ราคา หรือกลุ่มสินค้า
* เวลา
* ความยาวรีวิว

ทำ **2–4 ตารางหรือกราฟ** ก็พอ แล้วเขียนสั้น ๆ ว่า

> “สิ่งใดจาก EDA ทำให้ทีมอยากวิเคราะห์ต่อ และเพราะอะไร”

ไม่ต้องทำ EDA ทุกตัวแปร


````python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Prepare data for plotting
# Calculate counts of ratings for verified and non-verified purchases
rating_counts = df_joined.groupby(['verified_purchase', 'rating']).size().unstack(fill_value=0)

# Calculate percentages for each group (row-wise)
rating_percentages = rating_counts.apply(lambda x: x / x.sum(), axis=1)

# Melt the dataframe for seaborn's barplot
df_plot = rating_percentages.stack().reset_index(name='percentage')
df_plot['rating'] = df_plot['rating'].astype(str) + ' Star'

# --- Plotting ---

# 1. Set up the plot with Spotify Light Mode style
plt.style.use('default')
fig, ax = plt.subplots(figsize=(10, 6), facecolor='#FFFFFF')
ax.set_facecolor('#FFFFFF')

# Spotify palette: Verified = Bright Green, Non-Verified = Dark Charcoal
spotify_colors = {
    True: '#1DB954',
    False: '#191414'
}

# Create the clustered bar chart
sns.barplot(
    data=df_plot,
    x='rating',
    y='percentage',
    hue='verified_purchase',
    palette=spotify_colors,
    ax=ax,
    edgecolor='none' # Remove bar borders
)

# 3. Apply Spotify Minimal Styling
ax.set_title('Rating Distribution by Verified Purchase Status', fontsize=18, fontweight='bold', color='#191414', pad=20)
ax.set_xlabel('') # Hide x-axis label
ax.set_ylabel('') # Hide y-axis label

# Adjust x-axis tick labels for clarity
ax.tick_params(axis='x', colors='#191414', labelsize=12, rotation=0)
ax.tick_params(axis='y', colors='#555555', labelsize=11)

# Remove all spines
for spine in ['top', 'right', 'left', 'bottom']:
    ax.spines[spine].set_visible(False)

# Add horizontal grid lines
ax.grid(axis='y', linestyle='--', alpha=0.4, color='#CCCCCC')

# Add annotations (percentages) on top of each bar
for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.annotate(f'{height:.1%}',
                    (p.get_x() + p.get_width() / 2., height),
                    ha='center', va='bottom', xytext=(0, 5),
                    textcoords='offset points', fontsize=10, color='#191414', fontweight='bold')

# Adjust legend
handles, _ = ax.get_legend_handles_labels()
ax.legend(handles=handles, labels=['Non-Verified Purchase', 'Verified Purchase'],
          title='Purchase Status', frameon=False)

plt.tight_layout()
plt.show()
````

รีวิวส่วนใหญ่เป็นรีวิวเชิงบวก โดย 5 ดาวเป็นคะแนนที่มีสัดส่วนสูงที่สุดทั้งในกลุ่ม Verified และ Non-Verified Purchase โดยกลุ่ม Verified Purchase มีรีวิว 5 ดาวสูงถึง 79.8% ขณะที่กลุ่ม Non-Verified Purchase มี 70.4% นอกจากนี้ กลุ่ม Non-Verified Purchase มีสัดส่วนรีวิว 4 ดาว 15.5%, 3 ดาว 6.1% และ 1 ดาว 5.8% ซึ่งแสดงให้เห็นว่าคะแนนของกลุ่ม Non-Verified มีการกระจายตัวในหลายระดับมากกว่ากลุ่ม Verified (ซึ่งคะแนนกระจุกตัวอยู่ที่ 5 ดาวเป็นหลัก)   

ผลจากกราฟนี้ช่วยกำหนดแนวทางการคัดเลือกข้อมูลสำหรับการวิเคราะห์ต่อ เนื่องจาก Brief B3 ต้องการค้นหาว่า ลูกค้าที่มีความประทับใจพูดถึงจุดเด่นและคุณค่าของสินค้าอย่างไร จึงเลือกใช้ Verified Purchase ร่วมกับ Rating 4–5 ดาว เป็นกลุ่มหลักในการวิเคราะห์ Positive Voice of Customer เพื่อให้ผลการวิเคราะห์สะท้อนความคิดเห็นจากกลุ่มที่ยืนยันการซื้อสินค้า

````python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. PREPARE DATA
# กรองเฉพาะรีวิวเชิงบวก (4-5 ดาว)
df_positive = df_joined[df_joined['rating'].isin([4.0, 5.0])].copy()

# 2. SET MINIMAL LIGHT THEME & COLOR PALETTE
plt.style.use('default')

BG_COLOR = '#FFFFFF'
TEXT_DARK = '#191414'
TEXT_MUTED = '#555555'
GRID_COLOR = '#CCCCCC'

# โทนสีเขียว 2 เฉดทึบแน่น (Non-Verified: เขียวเข้มหม่น / Verified: เขียวสว่าง Spotify)
SPOTIFY_PALETTE = ['#137938', '#1DB954']

# 3. PLOT BARPLOT
fig, ax = plt.subplots(figsize=(7, 5), dpi=120, facecolor=BG_COLOR)
ax.set_facecolor(BG_COLOR)

sns.barplot(
    data=df_positive,
    x='verified_purchase',
    y='helpful_vote',
    hue='verified_purchase',
    palette=SPOTIFY_PALETTE,
    legend=False,
    errorbar=None,
    ax=ax,
)

# 4. STYLING & ANNOTATIONS
ax.set_title(
    'Average Helpful Votes',
    fontsize=16,
    fontweight='bold',
    color=TEXT_DARK,
    pad=20,
)

# ซ่อนชื่อลาเบลแกน X และ Y
ax.set_xlabel('')
ax.set_ylabel('')

# กำหนดข้อความป้ายแกน X และสีตัวเลขแกน Y
ax.set_xticks([0, 1])
ax.set_xticklabels(
    ['Non-Verified', 'Verified Purchase'],
    color=TEXT_DARK,
    fontsize=11,
    fontweight='bold',
)
ax.tick_params(axis='y', colors=TEXT_MUTED)

# ซ่อนขอบกราฟทั้ง 4 ด้าน
for spine in ['top', 'right', 'left', 'bottom']:
  ax.spines[spine].set_visible(False)

# เปิด Grid เส้นประแนวนอน
ax.grid(axis='y', linestyle='--', alpha=0.4, color=GRID_COLOR)

# คำนวณขอบบนแกน Y (Headroom) ป้องกันตัวเลขถูกตัดขอบ
max_val = max([p.get_height() for p in ax.patches]) if ax.patches else 1.0
ax.set_ylim(0, max_val * 1.15)

# ตัวเลขแสดงค่าบนหัวแท่งกราฟ
for p in ax.patches:
  height = p.get_height()
  if height > 0:
    ax.annotate(
        f'{height:.2f}',
        (p.get_x() + p.get_width() / 2.0, height),
        ha='center',
        va='bottom',
        xytext=(0, 5),
        textcoords='offset points',
        fontsize=11,
        fontweight='bold',
        color=TEXT_DARK,
    )

plt.tight_layout()
plt.show()
````

### 2.4 เตรียมข้อความและสร้าง Text Features

ก่อนนำข้อความไปวิเคราะห์ จำเป็นต้องมีการ **preprocessing** เพื่อทำให้ข้อความอยู่ในรูปที่เหมาะสมกับเทคนิคที่จะใช้ อย่างไรก็ตาม ในงานนี้ไม่จำเป็นต้องทำ preprocessing ทุกขั้นตอน ให้เลือกทำ **เท่าที่จำเป็น** ตามวัตถุประสงค์ของการวิเคราะห์

ตัวอย่างเช่น การเปลี่ยนข้อความเป็นตัวพิมพ์เล็ก (lower case) การนำ URL หรือ HTML ออก การแบ่งข้อความเป็นคำ (tokenization) หรือการตัด stopwords เช่น `the`, `and`, `is` ออก ในกรณีที่คำเหล่านี้ไม่ได้ช่วยในการวิเคราะห์

**ไม่บังคับให้ใช้ stemming หรือ lemmatization** หากไม่มีเหตุผลว่าขั้นตอนดังกล่าวจะช่วยตอบคำถามของทีมได้ดีขึ้น

หลังจากเตรียมข้อความแล้ว ให้สร้าง **Text Features อย่างน้อย 2 ชนิด** เพื่อนำข้อความมาเปลี่ยนให้อยู่ในรูปข้อมูลที่สามารถนำไปสำรวจ เปรียบเทียบ หรือวิเคราะห์ต่อได้

---

#### Text Feature คืออะไร?

Text Feature คือ **ข้อมูลเชิงตัวเลขที่สกัดหรือสร้างขึ้นจากข้อความ** เพื่อแทนลักษณะหรือความหมายบางอย่างของข้อความนั้น

Text Feature อาจอยู่ในรูปของ

* **ค่าตัวเลขค่าเดียว** เช่น จำนวนคำ หรือ sentiment score
* **ตัวแปรหลายตัว** เช่น `price=1`, `quality=0`, `delivery=1`
* **เวกเตอร์หลายมิติ** เช่น TF-IDF vector หรือ Text Embedding

ตัวอย่างเช่น รีวิว

> `"The product is easy to use but the delivery was very slow."`

สามารถเปลี่ยนเป็น feature ได้หลายรูปแบบ

| Feature                    |                 ตัวอย่างค่าที่ได้ |
| -------------------------- | --------------------------------: |
| จำนวนคำในรีวิว             |                                12 |
| Sentiment score            |                             -0.15 |
| กล่าวถึง usability หรือไม่ |                                 1 |
| กล่าวถึง delivery หรือไม่  |                                 1 |
| กล่าวถึง price หรือไม่     |                                 0 |
| TF-IDF vector              |   `[0.00, 0.41, 0.00, 0.72, ...]` |
| Embedding vector           | `[0.21, -0.08, 0.63, ..., -0.14]` |

ดังนั้นข้อความหนึ่งข้อความสามารถถูกแทนได้ทั้งด้วย **feature ที่ตีความได้โดยตรง** และ **feature แบบเวกเตอร์ที่ใช้แทนเนื้อหาหรือความหมายของข้อความ**

---

## ตัวอย่าง Text Features ที่สามารถเลือกใช้

ผมจะแก้ตารางเดิมเป็นแบบนี้:

| Text Feature          | คืออะไร                                     | ตัวอย่าง                     | ใช้ตอบคำถามอะไรได้                                    |
| --------------------- | ------------------------------------------- | ---------------------------- | ----------------------------------------------------- |
| **Review Length**     | จำนวนคำหรือตัวอักษร                         | 120 words                    | รีวิวที่ยาวกว่าได้รับ helpful vote มากกว่าหรือไม่     |
| **Sentiment Score**   | คะแนนความเป็นบวก/ลบ                         | `-0.72`, `+0.85`             | ความรู้สึกลูกค้าสัมพันธ์กับ rating หรือไม่            |
| **Aspect Indicators** | ระบุว่าพูดถึงหัวข้อใด                       | `delivery=1`, `price=0`      | ลูกค้าบ่นเรื่องใดมากที่สุด                            |
| **Lexical Diversity** | ความหลากหลายของคำ                           | unique words / total words   | รีวิวที่เขียนละเอียดมีประโยชน์มากกว่าหรือไม่          |
| **TF-IDF Vector**     | เวกเตอร์ที่ให้น้ำหนักคำเด่นในเอกสาร         | `[0, .35, 0, .81, ...]`      | คำใดแยกกลุ่มรีวิวบวกกับลบ                             |
| **Text Embedding**    | dense vector ที่แทนความหมายโดยรวมของข้อความ | `[.12, -.44, .08, ...]`      | รีวิวใดมีความหมายคล้ายกัน หรือมี customer themes อะไร |
| **Custom Feature**    | feature ที่ทีมสร้างตามโจทย์                 | `refund=1`, `defect_count=3` | ตอบ business question เฉพาะของทีม                     |

### ตัวอย่าง Embedding แบบง่าย

ถ้าสอน `sentence-transformers` มาแล้ว อาจใส่ตัวอย่างสั้น ๆ:

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

texts = [

    "Shipping took too long.",
    "Excellent sound quality."

print(embeddings.shape)
```

**รายงานเพียง 3 อย่าง**
1. ทำ preprocessing อะไร และทำไม
2. สร้าง feature อะไรอย่างน้อย 2 ตัว
3. feature เหล่านั้นช่วยตอบคำถามทางธุรกิจอย่างไร

> ไม่ต้องรายงาน vocabulary ก่อน–หลัง เว้นแต่ประเด็นนั้นเกี่ยวข้องกับผลวิเคราะห์

````python
import ast
import html
import re
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns



# 1. TEXT PREPROCESSING
url_re = re.compile(r'https?://\S+|www\.\S+')
html_re = re.compile(r'<.*?>')
space_re = re.compile(r'\s+')

titles = df_joined['title_review'].fillna('').astype(str).tolist()
texts = df_joined['text'].fillna('').astype(str).tolist()


def clean_text_fast(title, text):
    combined = f'{title} {text}'
    combined = html.unescape(combined)
    combined = url_re.sub('', combined)
    combined = html_re.sub('', combined)
    combined = space_re.sub(' ', combined)
    return combined.strip().lower()


cleaned_list = [
    clean_text_fast(t, x)
    for t, x in zip(titles, texts)
]

df_joined['cleaned_text'] = cleaned_list

# 2. TEXT FEATURES
# ---------- 2.1 Review Length ----------
# นับจำนวนคำจาก Review Text โดยไม่รวม Title

review_only = [
    clean_text_fast('', x)
    for x in texts
]

df_joined['review_length'] = [
    len(x.split()) for x in review_only
]

# ---------- 2.2 Aspect Indicators ----------

# Sound Quality
ADJ_POS_NEG = (
    r'great|good|excellent|awesome|amazing|wonderful|superb|fantastic|'
    r'terrible|bad|poor|awful|clear|crisp|muddy|tinny|thin|flat'
)

sound_pattern = re.compile(
    r'\b(?:'
    r'sound quality|audio quality'
    rf'|(?:{ADJ_POS_NEG})\s+(?:sound|audio)'
    rf'|sound(?:s|ed)?\s+(?:is |are |was |so |very |really )?(?:{ADJ_POS_NEG})'
    r'|crystal clear|distortion|distorted|hi-?fi|fidelity|remaster(?:ed)?|mastering'
    r'|vocals? (?:are|is|sound)s?'
    r')\b',
    flags=re.IGNORECASE
)


# Nostalgia
nostalgia_pattern = re.compile(
    r'\b(?:'
    r'nostalgi(?:a|c)|childhood|brings? (?:me )?back|takes? me back'
    r'|back in the day|good old days|growing up|grew up|my youth'
    r'|reminds? me of|memories'
    r')\b',
    flags=re.IGNORECASE
)


# Value for Money
value_pattern = re.compile(
    r'\b(?:'
    r'worth (?:every|the|it)|good (?:price|value)|great (?:price|value)'
    r'|bargain|affordable|cheap|overpriced|rip-?off|value for money'
    r'|money well spent|waste of money|a steal'
    r')\b',
    flags=re.IGNORECASE
)


def flag(pattern, texts):
    return np.fromiter(
        (1 if pattern.search(x) else 0 for x in texts),
        dtype=np.int8
    )


df_joined['mention_sound'] = flag(
    sound_pattern, cleaned_list
)

df_joined['mention_nostalgia'] = flag(
    nostalgia_pattern, cleaned_list
)

df_joined['mention_value'] = flag(
    value_pattern, cleaned_list
)


# 3. INFERRED GENRE
#    Rule-based Regex Classification
def extract_clean_artist(store_val):
    if pd.isna(store_val):
        return 'Unknown'

    clean_val = re.sub(
        r'Format:.*$',
        '',
        str(store_val)
    )

    clean_val = re.sub(
        r'\(Artist\)',
        '',
        clean_val
    )

    return clean_val.strip()


df_joined['clean_artist'] = df_joined['store'].apply(
    extract_clean_artist
)


# กฎสำหรับจำแนกแนวเพลงด้วย Regular Expression
GENRE_RULES = [
    (
        'K-Pop / Asian Pop',
        r'\b(?:bts|bangtan|k-?pop|hanteo|photobook)\b'
    ),

    (
        'Classical / Jazz',
        r'\b(?:jazz|classical|symphony|orchestra|concerto|opera|yo-yo ma)\b'
    ),

    (
        'Classic Rock / Oldies',
        r"\b(?:beatles|elvis|pink floyd|grateful dead|rock|oldies|bbc sessions|60s|70s|80s)\b"
    ),

    (
        'Dance / R&B / Electronic',
        r'\b(?:r&b|house music|dj|remix|dance|electronic|edm)\b'
    ),

    (
        'Audiobook / Spoken Word',
        r'\b(?:unabridged|audiobook|spoken word)\b'
    ),
]

GENRE_RULES = [
    (g, re.compile(p, flags=re.IGNORECASE))
    for g, p in GENRE_RULES
]


def infer_genre(title, desc, artist):

    d = (
        ' '.join(map(str, desc))
        if isinstance(desc, list)
        else str(desc or '')
    )

    text = f'{title} {d} {artist}'.lower()

    for g, pat in GENRE_RULES:
        if pat.search(text):
            return g

    return 'General / Mainstream'


df_joined['inferred_genre'] = [
    infer_genre(t, d, a)
    for t, d, a in zip(
        df_joined['title_meta'],
        df_joined['description'],
        df_joined['clean_artist']
    )
]

# 4. แสดงผลลัพธ์ TEXT FEATURES
display(
    df_joined[
        [
            'title_review',
            'cleaned_text',
            'review_length',
            'mention_sound',
            'mention_nostalgia',
            'mention_value',
            'inferred_genre'
        ]
    ].head()
)

# 5. แสดง Distribution ของ Inferred Genre

print("\n=== Distribution of Inferred Genres ===")

print(
    df_joined['inferred_genre'].value_counts()
)

print("=" * 40)
````

# Text Preprocessing & Feature Engineering

## 1. Preprocessing

### ทำอะไรบ้าง
- **Combine Text:** รวม `Title` และ `Review Text` เพื่อใช้เป็นข้อความสำหรับการตรวจจับ Keyword/Aspect
- **Decode HTML:** แปลง HTML entities เช่น `&amp;` → `&`
- **Remove URLs & HTML Tags:** ลบ URL และ HTML tags ที่ไม่จำเป็นต่อการวิเคราะห์
- **Normalize Whitespace:** ลดช่องว่างซ้ำให้เป็นช่องว่างเดียว
- **Lowercasing:** แปลงข้อความเป็นตัวพิมพ์เล็ก เพื่อให้การค้นหา Keyword สม่ำเสมอ

### ทำไมต้องทำ
- ลด Noise จาก URL และ HTML
- ทำให้รูปแบบข้อความเป็นมาตรฐานเดียวกัน
- ช่วยให้การตรวจจับ Keyword และ Regex มีความสม่ำเสมอ

#### **Sentiment Contrast (เปรียบเทียบรีวิว 4–5 ดาว vs 1–2 ดาว)**

````python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# HELPER FUNCTIONS (DEFINED FOR THE COMPARISONS)
FEATS = ['mention_sound', 'mention_nostalgia', 'mention_value']

def compare_rates(a, b, name_a, name_b, feats=FEATS):
    print(f'n {name_a} = {len(a):,} | n {name_b} = {len(b):,}')
    rows = []
    for c in feats:
        xa, xb = int(a[c].sum()), int(b[c].sum())
        rows.append({
            'feature': c,
            f'{name_a} %': round(xa / len(a) * 100, 1) if len(a) > 0 else 0.0,
            f'{name_a} count': xa,
            f'{name_b} %': round(xb / len(b) * 100, 1) if len(b) > 0 else 0.0,
            f'{name_b} count': xb,
        })
    return pd.DataFrame(rows)

def plot_compare_rates(res_df, name_a, name_b, title='Aspect Mention Rate Comparison'):
    melted_df = res_df.melt(
        id_vars=['feature'],
        value_vars=[f'{name_a} %', f'{name_b} %'],
        var_name='Group',
        value_name='Mention Rate (%)',
    )
    melted_df['Group'] = melted_df['Group'].str.replace(' %', '')
    melted_df['Aspect'] = (
        melted_df['feature'].str.replace('mention_', '').str.title()
    )

    # Update Aspect Display Names
    melted_df['Aspect'] = melted_df['Aspect'].replace({
        'Sound': 'Sound Quality',
        'Value': 'Value for Money'
    })

    fig, ax = plt.subplots(figsize=(9, 5.5), dpi=120)

    sns.barplot(
        data=melted_df,
        x='Aspect',
        y='Mention Rate (%)',
        hue='Group',
        ax=ax,
        palette=['#1DB954', '#535353'],
        edgecolor='black',
    )

    for p in ax.patches:
        h = p.get_height()
        if h > 0:
            ax.annotate(
                f'{h:.1f}%',
                (p.get_x() + p.get_width() / 2.0, h),
                ha='center',
                va='bottom',
                color='black',
                fontweight='bold',
                fontsize=9.5,
                xytext=(0, 3),
                textcoords='offset points',
            )

    ax.set_title(title, fontsize=13, color='black', fontweight='bold', pad=15)
    ax.set_xlabel('Marketing Aspect', fontsize=11, color='black')
    ax.set_ylabel('Mention Rate (%)', fontsize=11, color='black')

    max_val = melted_df['Mention Rate (%)'].max() if not melted_df.empty else 10
    ax.set_ylim(0, max_val * 1.18)

    ax.legend(
        title='Group',
        facecolor='white',
        edgecolor='#CCCCCC',
        labelcolor='black',
    )
    ax.grid(True, axis='y', linestyle='--', alpha=0.6)

    plt.tight_layout()
    plt.show()

# 4. DEFINE B3 MARKETING SAMPLES
# Satisfied customers:
# Verified Purchase + Rating 4–5
marketing_df = df_joined[
    (df_joined['verified_purchase'] == True) &
    (df_joined['rating'] >= 4)
].copy()

# Negative comparison group:
# Verified Purchase + Rating 1–2
negative_df = df_joined[
    (df_joined['verified_purchase'] == True) &
    (df_joined['rating'] <= 2)
].copy()

print(f"Marketing sample (Verified + 4–5 stars): {len(marketing_df):,}")
print(f"Negative sample (Verified + 1–2 stars): {len(negative_df):,}")


# 5. EXECUTION
# 1) Positive vs Negative
pos_df = marketing_df
neg_df = negative_df

res_pos_neg = compare_rates(
    pos_df,
    neg_df,
    'Positive',
    'Negative'
)

display(res_pos_neg)

plot_compare_rates(
    res_pos_neg,
    'Positive',
    'Negative',
    title='Sentiment Contrast: Aspect Mention Rate (Verified Positive vs. Negative Reviews)',
)

# 2) High Impact vs General
high_df = marketing_df[
    marketing_df['helpful_vote'] >= 5
].copy()

gen_df = marketing_df[
    marketing_df['helpful_vote'] < 5
].copy()

res_high_gen = compare_rates(
    high_df,
    gen_df,
    'HighImpact',
    'General'
)

display(res_high_gen)

plot_compare_rates(
    res_high_gen,
    'HighImpact',
    'General',
    title='Impact Contrast: Aspect Mention Rate (Verified Positive Reviews)',
)
````

จากกราฟ Positive vs Negative Reviews พบว่า Sound Quality ถูกกล่าวถึงใน Positive Reviews 4.3% และ Negative Reviews 6.3% ส่วน Value for Money อยู่ที่ 3.8% และ 5.8% ตามลำดับ แสดงว่า Sound และ Value เป็นประเด็นที่ลูกค้าให้ความสำคัญ แต่สามารถเป็นได้ทั้ง จุดขายและจุดที่ทำให้เกิดความไม่พอใจ จึงควรสื่อสารโดยเน้นคุณภาพและความคุ้มค่าอย่างระมัดระวัง

````python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# CONSTANTS & THEME SETUP
BG_COLOR = "#FFFFFF"
DARK_GRAY = "#505050"
SPOTIFY_GREEN = "#2B9F5E"
MID_GRAY = "#8C8C8C"
ALERT_RED = "#D32F2F"
SPINE_COLOR = "#CCCCCC"
GRID_COLOR = "#E0E0E0"
BAR_EDGE = "#1A1A1A"

# PREPARE DATA FOR REALITY CHECK
# Satisfied Customers
# Definition:
# Verified Purchase + Rating 4–5
satisfied_df = df_joined[
    (df_joined["verified_purchase"] == True) &
    (df_joined["rating"] >= 4)
].copy()

# High Impact = Helpful Vote >= 5
high_helpful = satisfied_df[
    satisfied_df["helpful_vote"] >= 5
].copy()

# General = Helpful Vote < 5
general_helpful = satisfied_df[
    satisfied_df["helpful_vote"] < 5
].copy()


# REALITY CHECK 1
# High Impact vs General Satisfied Reviews
# Calculate aspect mention rate
reality_check_df = pd.DataFrame({
    "High Impact": [
        high_helpful["mention_sound"].mean() * 100,
        high_helpful["mention_nostalgia"].mean() * 100,
        high_helpful["mention_value"].mean() * 100
    ],

    "General": [
        general_helpful["mention_sound"].mean() * 100,
        general_helpful["mention_nostalgia"].mean() * 100,
        general_helpful["mention_value"].mean() * 100
    ]
}, index=[
    "Sound Quality",
    "Nostalgia",
    "Value for Money"
])

# GRAPH 1
fig, ax1 = plt.subplots(
    figsize=(8, 5),
    dpi=120,
    facecolor=BG_COLOR
)

ax1.set_facecolor(BG_COLOR)

colors_g1 = [
    SPOTIFY_GREEN,
    DARK_GRAY
]

reality_check_df.plot(
    kind="bar",
    ax=ax1,
    width=0.65,
    color=colors_g1,
    edgecolor=BAR_EDGE,
    linewidth=0.8
)

# Title
ax1.set_title(
    "Reality Check 1: Aspect Mention Rate\nHigh Impact vs General Reviews",
    fontsize=13,
    fontweight="bold",
    color="#000000",
    pad=15
)

# Axis labels
ax1.set_xlabel(
    "Customer Voice Aspects",
    fontsize=11,
    fontweight="bold",
    color="#000000"
)

ax1.set_ylabel(
    "Mention Rate (%)",
    fontsize=11,
    fontweight="bold",
    color="#000000"
)

ax1.tick_params(
    axis="x",
    rotation=0,
    labelsize=10,
    colors="#000000"
)

ax1.tick_params(
    axis="y",
    labelsize=10,
    colors="#000000"
)

# Spines
ax1.spines["top"].set_visible(False)
ax1.spines["right"].set_visible(False)
ax1.spines["left"].set_color(SPINE_COLOR)
ax1.spines["bottom"].set_color(SPINE_COLOR)

# Grid
ax1.grid(
    axis="y",
    linestyle="--",
    alpha=0.7,
    color=GRID_COLOR
)

# Y limit
max_value = reality_check_df.values.max()

ax1.set_ylim(
    0,
    max(max_value * 1.30, 20)
)

# Legend
ax1.legend(
    title="Review Group",
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    frameon=True,
    facecolor="white",
    edgecolor=SPINE_COLOR,
    fontsize=9.5
)

# Data labels
for container in ax1.containers:
    ax1.bar_label(
        container,
        fmt="%.1f%%",
        padding=3,
        fontsize=9.5,
        fontweight="bold",
        color="#000000"
    )

plt.tight_layout()
plt.show()

# REALITY CHECK 2
# GENRE SAMPLE SIZE CHECK
# Count reviews in each inferred genre
genre_counts = df_joined["inferred_genre"].value_counts()

df_genre = genre_counts.reset_index()
df_genre.columns = ["genre", "count"]

# GRAPH 2
# Highlight small samples
colors_g2 = [
    ALERT_RED if count < 100 else SPOTIFY_GREEN
    for count in df_genre["count"]
]

fig, ax2 = plt.subplots(
    figsize=(10, 6),
    dpi=120,
    facecolor=BG_COLOR
)

ax2.set_facecolor(BG_COLOR)

bars = ax2.barh(
    df_genre["genre"],
    df_genre["count"],
    color=colors_g2,
    height=0.6,
    edgecolor=BAR_EDGE,
    linewidth=0.8
)

# Threshold line
ax2.axvline(
    x=100,
    color=ALERT_RED,
    linestyle="--",
    linewidth=1.5,
    label="Warning Threshold (n = 100)"
)

# Title
ax2.set_title(
    "Reality Check 2: Sample Size by Inferred Genre",
    fontsize=13,
    fontweight="bold",
    color="#000000",
    pad=15
)

# Axis labels
ax2.set_xlabel(
    "Sample Size (n)",
    fontsize=11,
    fontweight="bold",
    color="#000000"
)

ax2.set_ylabel(
    "Inferred Genre",
    fontsize=11,
    fontweight="bold",
    color="#000000"
)

ax2.tick_params(
    axis="x",
    labelsize=10,
    colors="#000000"
)

ax2.tick_params(
    axis="y",
    labelsize=10,
    colors="#000000"
)

# Spines
ax2.spines["top"].set_visible(False)
ax2.spines["right"].set_visible(False)
ax2.spines["left"].set_color(SPINE_COLOR)
ax2.spines["bottom"].set_color(SPINE_COLOR)

# Grid
ax2.grid(
    axis="x",
    linestyle="--",
    alpha=0.7,
    color=GRID_COLOR
)

# Reverse order
ax2.invert_yaxis()

# Legend
ax2.legend(
    loc="lower right",
    frameon=True,
    facecolor="white",
    edgecolor=SPINE_COLOR,
    fontsize=9.5
)

# Data labels
max_c = max(df_genre["count"]) if not df_genre.empty else 100

for bar in bars:

    width = bar.get_width()

    is_small = width < 100

    ax2.text(
        width + (max_c * 0.01),
        bar.get_y() + bar.get_height() / 2,
        f"n={int(width)}",
        va="center",
        ha="left",
        fontsize=9.5,
        color=ALERT_RED if is_small else "#000000",
        fontweight="bold"
    )

ax2.set_xlim(
    0,
    max_c * 1.15
)

plt.tight_layout()
plt.show()


# PRINT REALITY CHECK SUMMARY
print("=" * 60)
print("REALITY CHECK SUMMARY")
print("=" * 60)

print(f"Satisfied Customers: n = {len(satisfied_df):,}")
print(f"High Impact Reviews: n = {len(high_helpful):,}")
print(f"General Reviews:     n = {len(general_helpful):,}")

print("\nAspect Mention Rate:")
print(reality_check_df.round(1))

print("\nGenre Sample Sizes:")
print(df_genre.to_string(index=False))
````

## 2.7 Insight Cards

เขียน **2 ใบ** ตามแบบฟอร์มด้านบนของ notebook

เลือก 2 insight ที่คิดว่า **มีประโยชน์ต่อการตัดสินใจมากที่สุด** ไม่ใช่ 2 insight ที่ใช้เทคนิคยากที่สุด


#### **Insight Card 1 — Voice of Customer**

* **Insight:** “sound quality” และ “great sound” พบใน Positive Reviews **1.07%** และ **0.50%**
* **หลักฐาน:** n = 13,073 ; พบ 140 และ 66 รีวิวตามลำดับ
* **So What:** นำภาษาของลูกค้าไปพัฒนา Marketing Message ได้
* **Action:** Copywriter ใช้ “Sound Quality / Great Sound” ทำ Ad Headline และ A/B Test CTR ภายใน 2 สัปดาห์
* **ความมั่นใจ:** **สูง** สำหรับการระบุ VoC แต่ไม่ได้ยืนยันว่า Sound Quality เป็นสาเหตุหลักของการซื้อ

````python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
import numpy as np
import pandas as pd
import seaborn as sns
import re

# GRAPH 1: INSIGHT CARD 1 PROOF
# Selected VoC Phrases from Satisfied Customers

# 1. FILTER SATISFIED CUSTOMERS
# B3 definition:
# Satisfied Customers = Verified Purchase + Rating 4–5

df_pos = df_joined[
    (df_joined['verified_purchase'] == True) &
    (df_joined['rating'] >= 4)
].copy()

TOTAL_POSITIVE_REVIEWS = len(df_pos)

print("=" * 60)
print("VOC PHRASE ANALYSIS")
print("=" * 60)
print(f"Verified Positive Reviews: {TOTAL_POSITIVE_REVIEWS:,}")

# 2. DEFINE SELECTED VOC PHRASES

voc_phrases = {
    "sound quality": (
        "sound quality",
        "Sound Quality"
    ),

    "great sound": (
        "great sound",
        "Sound Quality"
    ),

    "worth every penny": (
        "worth (?:every penny|every cent|the money)",
        "Value for Money"
    ),

    "good value": (
        "good value",
        "Value for Money"
    ),

    "brings back memories": (
        "brings? back (?:old )?memories|takes? me back",
        "Nostalgia"
    ),

    "reminds me of": (
        "reminds? me of",
        "Nostalgia"
    )
}


# 3. COUNT VOC PHRASES

voc_data = []

for phrase, (pattern, category) in voc_phrases.items():

    compiled = re.compile(
        r'\b' + pattern + r'\b',
        flags=re.IGNORECASE
    )

    count = (
        df_pos['cleaned_text']
        .fillna('')
        .apply(
            lambda x: 1 if compiled.search(str(x)) else 0
        )
        .sum()
    )

    mention_rate = (
        count / TOTAL_POSITIVE_REVIEWS * 100
        if TOTAL_POSITIVE_REVIEWS > 0
        else 0
    )

    voc_data.append({
        "Phrase": phrase,
        "Category": category,
        "Count": count,
        "Mention_Rate": mention_rate
    })


# Convert to DataFrame
df_voc = pd.DataFrame(voc_data)


# 4. SORT FOR PLOTTING
plot_voc = (
    df_voc
    .copy()
    .sort_values(
        by='Mention_Rate',
        ascending=True
    )
    .reset_index(drop=True)
)



# 5. DISPLAY RESULT TABLE
display(
    df_voc
    .sort_values(
        by='Mention_Rate',
        ascending=False
    )
    .round(2)
)


# 6. SET COLOR PALETTE
palette_map = {
    "Sound Quality": "#27ae60",
    "Value for Money": "#505050",
    "Nostalgia": "#a6a6a6",
}


colors_voc = [
    palette_map.get(
        category,
        "#808080"
    )
    for category in plot_voc["Category"]
]


# 7. CREATE PLOT
fig, ax = plt.subplots(
    figsize=(9, 5.5)
)


bars = ax.barh(
    plot_voc["Phrase"],
    plot_voc["Mention_Rate"],
    color=colors_voc,
    height=0.65,
    edgecolor="black",
    linewidth=0.8,
)


# 8. TITLE
plt.title(
    f"Selected VoC Phrases (% of Verified Positive Reviews | "
    f"Rating ≥ 4, Total n = {TOTAL_POSITIVE_REVIEWS:,})",
    fontsize=11.5,
    fontweight="bold",
    pad=14,
)


# 9. AXIS LABELS
plt.xlabel(
    "Mention Rate in Verified Positive Reviews (%)",
    fontsize=9.5,
    fontweight="bold"
)

plt.ylabel(
    "VoC Phrase / Copywriting Keyword",
    fontsize=9.5,
    fontweight="bold"
)

plt.xticks(fontsize=9)
plt.yticks(fontsize=9)

# 10. GRID + BORDER
sns.despine()

ax.grid(
    axis="x",
    linestyle="--",
    alpha=0.4,
    color="#cccccc"
)

ax.set_axisbelow(True)

# 11. DISPLAY PERCENTAGE + COUNT
for bar, percentage, count in zip(
    bars,
    plot_voc["Mention_Rate"],
    plot_voc["Count"]
):

    ax.annotate(
        f"{percentage:.2f}% (n={int(count):,})",

        xy=(
            bar.get_width(),
            bar.get_y() + bar.get_height() / 2
        ),

        xytext=(5, 0),

        textcoords="offset points",

        va="center",

        fontsize=8.5,

        fontweight="bold",
    )

# 12. LEGEND
legend_elements = [

    Patch(
        facecolor="#27ae60",
        edgecolor="black",
        label="Sound Quality"
    ),

    Patch(
        facecolor="#505050",
        edgecolor="black",
        label="Value for Money"
    ),

    Patch(
        facecolor="#a6a6a6",
        edgecolor="black",
        label="Nostalgia"
    ),

]


ax.legend(
    handles=legend_elements,
    title="Category",
    loc="lower right",
    frameon=True
)


# 13. FINALIZE
plt.tight_layout()
plt.show()
````

#### **Insight Card 2 — Category-to-Message Fit**

* **Insight:** Classic Rock / Oldies เด่นด้าน **Sound Quality 7.2%** และ ****Nostalgia 7.0%****

* **หลักฐาน:** Classic Rock / Oldies พบ Sound Quality 7.2%, Nostalgia 7.0% และ Value for Money 4.5%

* **So What:** ใช้ **Sound Quality + Nostalgia** เป็น Marketing Message สำหรับกลุ่ม Classic Rock / Oldies

* **Action:** ทำ Ad Copy แยก Segment และ A/B Test กับ Generic Message วัด CTR / Conversion

* **ความมั่นใจ:** **ปานกลาง** เป็น Pattern เชิงพรรณนา และ Genre เป็นการจัดกลุ่มจาก Metadata


````python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# GRAPH 2: CATEGORY-TO-MESSAGE FIT BY GENRE
# 1. FILTER SATISFIED CUSTOMERS
# B3 definition:
# Satisfied Customers = Verified Purchase + Rating 4–5

df_satisfied = df_joined[
    (df_joined['verified_purchase'] == True) &
    (df_joined['rating'] >= 4)
].copy()

print("=" * 60)
print("CATEGORY-TO-MESSAGE FIT BY INFERRED GENRE")
print("=" * 60)

print(f"Satisfied Customers: {len(df_satisfied):,}")


# 2. CREATE GENRE MATRIX
genre_matrix = (
    df_satisfied
    .groupby('inferred_genre')[
        [
            'mention_nostalgia',
            'mention_sound',
            'mention_value'
        ]
    ]
    .mean()
    * 100
)

genre_matrix = genre_matrix.reset_index()

genre_matrix.columns = [
    'inferred_genre',
    'Nostalgia',
    'Sound_Quality',
    'Value_for_Money'
]

# 3. DISPLAY RESULT TABLE
display(
    genre_matrix.round(2)
)


# 4. CONVERT WIDE → LONG
genre_melted = genre_matrix.melt(
    id_vars="inferred_genre",

    value_vars=[
        "Nostalgia",
        "Sound_Quality",
        "Value_for_Money"
    ],

    var_name="Marketing Aspect",

    value_name="Mention Rate (%)"
)


# Rename for display
genre_melted["Marketing Aspect"] = genre_melted[
    "Marketing Aspect"
].replace({

    "Sound_Quality": "Sound Quality",

    "Value_for_Money": "Value for Money",

    "Nostalgia": "Nostalgia"
})


# 5. PLOT
fig, ax = plt.subplots(
    figsize=(13, 6),
    dpi=120
)


sns.barplot(
    data=genre_melted,

    x="inferred_genre",

    y="Mention Rate (%)",

    hue="Marketing Aspect",

    ax=ax,

    palette=[
        "#a6a6a6",   # Nostalgia
        "#27ae60",   # Sound Quality
        "#505050"    # Value for Money
    ],

    edgecolor="black"
)


# 6. DISPLAY VALUES ON BARS

for container in ax.containers:

    labels = []

    for bar in container:

        height = bar.get_height()

        if pd.notna(height) and height > 0:

            labels.append(
                f"{height:.1f}%"
            )

        else:

            labels.append("")


    ax.bar_label(
        container,

        labels=labels,

        padding=3,

        fontsize=8.5,

        fontweight="bold"
    )

# 7. TITLE
ax.set_title(
    "Category-to-Message Fit by Inferred Genre",
    fontsize=13,
    fontweight="bold",
    pad=15
)


# 8. AXIS
ax.set_xlabel(
    "Inferred Genre",
    fontsize=11
)

ax.set_ylabel(
    "Mention Rate (%)",
    fontsize=11
)

# 9. X-AXIS
plt.xticks(
    rotation=20,
    ha="right",
    fontsize=9
)

# 10. GRID
ax.grid(
    True,
    axis="y",
    linestyle="--",
    alpha=0.4
)

ax.set_axisbelow(True)


# 11. LEGEND
ax.legend(
    title="Marketing Aspect",
    loc="upper right"
)



# 12. FINALIZE
plt.tight_layout()
plt.show()
````
