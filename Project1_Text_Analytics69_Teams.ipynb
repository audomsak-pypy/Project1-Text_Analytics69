## เทคนิคที่อาจใช้ — เลือกเท่าที่จำเป็น

ไม่ต้องใช้ทั้งหมด โดยทั่วไป **2–4 เทคนิคก็เพียงพอ** ให้เลือกตามคำถามของทีม

| เทคนิค | เหมาะเมื่ออยากรู้ |
|---|---|
| **Word / n-gram frequency** | คนพูดถึงอะไรบ่อย |
| **TF-IDF / distinctive words** | สองกลุ่มพูดต่างกันอย่างไร |
| **Sentiment** | ข้อความเป็นบวก/ลบเพียงใด |
| **Aspect analysis** | คนพูดถึงราคา คุณภาพ การส่งของ ฯลฯ อย่างไร |
| **Topic modeling** | มีประเด็นอะไรที่เราอาจยังไม่รู้ |
| **Trend / segmentation** | ปัญหาเปลี่ยนตามเวลา หรือหนักในกลุ่มไหน |

> เทคนิคขั้นสูง เช่น embedding, clustering, regression, classification หรือ anomaly detection ทำได้ แต่ **ไม่บังคับ** และเหมาะสำหรับ Bonus เมื่อใช้ถูกและอธิบายได้

ถ้ายังไม่เคยใช้เทคนิคใด ให้ทดลองกับข้อมูลชุดเล็กก่อน แล้วตรวจด้วยตาว่าผลสมเหตุสมผลหรือไม่


## แบบฟอร์ม Insight Card

ทุก insight เขียนในรูปแบบนี้ 5 บรรทัด ไม่ต้องยาว

> **Insight:** ประโยคเดียวที่บอกสิ่งที่พบ ต้องมีตัวเลขอยู่ในประโยค \
> **หลักฐาน:** ตัวเลขพร้อม n และชี้ว่าดูได้จากกราฟหรือตารางไหน \
> **So what:** กระทบธุรกิจอย่างไร ถ้าประมาณขนาดผลกระทบเป็นตัวเลขได้ยิ่งดี \
> **Action:** ทีมไหนควรทำอะไร และจะรู้ได้อย่างไรว่าได้ผล \
> **ความมั่นใจ:** สูง กลาง หรือ ต่ำ พร้อมเหตุผลสั้น ๆ

**ตัวอย่างที่ยังไม่ใช่ insight**
> ลูกค้าพูดถึงเรื่องราคาบ่อยที่สุด คิดเป็น 32% ของรีวิว

**ตัวอย่างที่ใช้ได้**
> **Insight:** รีวิวที่พูดถึงบรรจุภัณฑ์รั่วมีเพียง 6.2% ของทั้งหมด (n = 3,100) แต่ได้ดาวเฉลี่ย 2.1 เทียบกับค่าเฉลี่ยรวม 4.0 และกระจุกอยู่ในสินค้า 12 ชิ้นที่ใช้ขวดแบบปั๊ม \
> **หลักฐาน:** ตารางค่าเฉลี่ยดาวแยกตามด้าน และกราฟแท่งสินค้า 12 ชิ้นในผลวิเคราะห์ \
> **So what:** สินค้า 12 ชิ้นนี้คิดเป็น 18% ของยอดรีวิวทั้งหมวด ถ้าแก้ปัญหาฝาขวดได้ คาดว่าคะแนนเฉลี่ยของหมวดจะขยับขึ้นราว 0.15 ดาว \
> **Action:** ทีม Product ตรวจสอบซัพพลายเออร์ฝาขวดของสินค้า 12 รายการนี้ภายในไตรมาสหน้า แล้วติดตามสัดส่วนรีวิวที่มีคำว่า leak, spill, broken cap ซึ่งตอนนี้อยู่ที่ 14% ต่อเดือน \

### 2.2 เตรียมข้อมูลที่เชื่อถือได้

ทำเพียงสิ่งจำเป็นก่อนเริ่มวิเคราะห์

* สุ่มรีวิวอย่างน้อย **20,000 รายการ** และอย่าใช้เพียง N บรรทัดแรก
* เชื่อม review กับ product metadata ด้วย `parent_asin`
* รายงานสั้น ๆ ว่า join สำเร็จกี่เปอร์เซ็นต์ และมี missing/duplicate สำคัญหรือไม่
* เขียน 2–3 บรรทัดว่า **ข้อมูลนี้เป็นตัวแทนของใคร และไม่เป็นตัวแทนของใคร**

> จุดประสงค์คือให้รู้ว่าข้อมูลที่ใช้เชื่อถือได้พอสำหรับคำถามของทีม ไม่ใช่ทำ data cleaning ทุกอย่างที่เป็นไปได้


````python
def iter_jsonl(path):
    with open(path, encoding='utf-8') as f:
        for line in f:
            yield json.loads(line)
````

### Sampling Reviews from `review_path`

````python
# Sample N reviews from the review_path
print(f"Sampling {SAMPLE_N:,} reviews from {review_path}...")

sampled_reviews = []
for i, review in enumerate(iter_jsonl(review_path)):
    if len(sampled_reviews) < SAMPLE_N:
        sampled_reviews.append(review)
    else:
        # Reservoir sampling to get a random sample of N reviews
        r = random.randint(0, i)
        if r < SAMPLE_N:
            sampled_reviews[r] = review

df_reviews = pd.DataFrame(sampled_reviews)
print(f"Loaded {len(df_reviews):,} sampled reviews.")

# Load all meta data
print(f"Loading metadata from {meta_path}...")
df_meta = pd.DataFrame(iter_jsonl(meta_path))
print(f"Loaded {len(df_meta):,} metadata items.")

display(df_reviews.head())
````

### Joining Reviews with Product Metadata

````python
# Perform a left merge to enrich reviews with product metadata
initial_review_count = len(df_reviews)
df_joined = pd.merge(df_reviews, df_meta, on='parent_asin', how='left', suffixes=('_review', '_meta'))

# Report join success percentage
successful_join_count = len(df_joined.dropna(subset=['main_category'])) # Assuming main_category is a good indicator of a successful meta join
join_percentage = (successful_join_count / initial_review_count) * 100
print(f"Successfully joined {successful_join_count:,} reviews with metadata ({join_percentage:.2f}%).")

# Check for missing values in critical columns
print("\nMissing values after join:")
missing_values = df_joined[['text', 'title_review', 'main_category', 'price']].isnull().sum()
display(missing_values[missing_values > 0])

# Check for duplicate reviews (based on a combination of user_id, timestamp, and asin)
# Note: parent_asin is the product group, asin is the specific product variant.
# We are interested in unique reviews, which is usually identified by user_id and timestamp for a product.
df_joined_before_dedup = len(df_joined)
df_joined = df_joined.drop_duplicates(subset=['user_id', 'timestamp', 'asin'])
duplicate_removed_count = df_joined_before_dedup - len(df_joined)
print(f"\nRemoved {duplicate_removed_count:,} duplicate reviews.")

display(df_joined.head())
````

### 2.3 สำรวจข้อมูลเท่าที่จำเป็น

เลือกดูเฉพาะสิ่งที่ช่วยให้เข้าใจ brief ของทีม เช่น

* การแจกแจงดาว
* verified purchase / helpful vote
* ราคา หรือกลุ่มสินค้า

ทำ **2–4 ตารางหรือกราฟ** ก็พอ แล้วเขียนสั้น ๆ ว่า

> “สิ่งใดจาก EDA ทำให้ทีมอยากวิเคราะห์ต่อ และเพราะอะไร”

ไม่ต้องทำ EDA ทุกตัวแปร


````python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Prepare data for plotting
# Calculate counts of ratings for verified and non-verified purchases
rating_counts = df_joined.groupby(['verified_purchase', 'rating']).size().unstack(fill_value=0)

# Calculate percentages for each group (row-wise)
rating_percentages = rating_counts.apply(lambda x: x / x.sum(), axis=1)

# Melt the dataframe for seaborn's barplot
df_plot = rating_percentages.stack().reset_index(name='percentage')
df_plot['rating'] = df_plot['rating'].astype(str) + ' Star'

# --- Plotting ---

# 1. Set up the plot with Spotify Light Mode style
plt.style.use('default')
fig, ax = plt.subplots(figsize=(10, 6), facecolor='#FFFFFF')
ax.set_facecolor('#FFFFFF')

# Spotify palette: Verified = Bright Green, Non-Verified = Dark Charcoal
spotify_colors = {
    True: '#1DB954',
    False: '#191414'
}

# Create the clustered bar chart
sns.barplot(
    data=df_plot,
    x='rating',
    y='percentage',
    hue='verified_purchase',
    palette=spotify_colors,
    ax=ax,
    edgecolor='none' # Remove bar borders
)

# 3. Apply Spotify Minimal Styling
ax.set_title('Rating Distribution by Verified Purchase Status', fontsize=18, fontweight='bold', color='#191414', pad=20)
ax.set_xlabel('') # Hide x-axis label
ax.set_ylabel('') # Hide y-axis label

# Adjust x-axis tick labels for clarity
ax.tick_params(axis='x', colors='#191414', labelsize=12, rotation=0)
ax.tick_params(axis='y', colors='#555555', labelsize=11)

# Remove all spines
for spine in ['top', 'right', 'left', 'bottom']:
    ax.spines[spine].set_visible(False)

# Add horizontal grid lines
ax.grid(axis='y', linestyle='--', alpha=0.4, color='#CCCCCC')

# Add annotations (percentages) on top of each bar
for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.annotate(f'{height:.1%}',
                    (p.get_x() + p.get_width() / 2., height),
                    ha='center', va='bottom', xytext=(0, 5),
                    textcoords='offset points', fontsize=10, color='#191414', fontweight='bold')

# Adjust legend
handles, _ = ax.get_legend_handles_labels()
ax.legend(handles=handles, labels=['Non-Verified Purchase', 'Verified Purchase'],
          title='Purchase Status', frameon=False)

plt.tight_layout()
plt.show()
````

รีวิวส่วนใหญ่เป็นรีวิวเชิงบวก โดย 5 ดาวเป็นคะแนนที่มีสัดส่วนสูงที่สุดทั้งในกลุ่ม Verified และ Non-Verified Purchase โดยกลุ่ม Verified Purchase มีรีวิว 5 ดาวสูงถึง 79.8% ขณะที่กลุ่ม Non-Verified Purchase มี 70.4% นอกจากนี้ กลุ่ม Non-Verified Purchase มีสัดส่วนรีวิว 4 ดาว 15.5%, 3 ดาว 6.1% และ 1 ดาว 5.8% ซึ่งแสดงให้เห็นว่าคะแนนของกลุ่ม Non-Verified มีการกระจายตัวในหลายระดับมากกว่ากลุ่ม Verified (ซึ่งคะแนนกระจุกตัวอยู่ที่ 5 ดาวเป็นหลัก)   

ผลจากกราฟนี้ช่วยกำหนดแนวทางการคัดเลือกข้อมูลสำหรับการวิเคราะห์ต่อ เนื่องจาก Brief B3 ต้องการค้นหาว่า ลูกค้าที่มีความประทับใจพูดถึงจุดเด่นและคุณค่าของสินค้าอย่างไร จึงเลือกใช้ Verified Purchase ร่วมกับ Rating 4–5 ดาว เป็นกลุ่มหลักในการวิเคราะห์ Positive Voice of Customer เพื่อให้ผลการวิเคราะห์สะท้อนความคิดเห็นจากกลุ่มที่ยืนยันการซื้อสินค้า

### 2.4 เตรียมข้อความและสร้าง Text Features

ก่อนนำข้อความไปวิเคราะห์ จำเป็นต้องมีการ **preprocessing** เพื่อทำให้ข้อความอยู่ในรูปที่เหมาะสมกับเทคนิคที่จะใช้ อย่างไรก็ตาม ในงานนี้ไม่จำเป็นต้องทำ preprocessing ทุกขั้นตอน ให้เลือกทำ **เท่าที่จำเป็น** ตามวัตถุประสงค์ของการวิเคราะห์

ตัวอย่างเช่น การเปลี่ยนข้อความเป็นตัวพิมพ์เล็ก (lower case) การนำ URL หรือ HTML ออก การแบ่งข้อความเป็นคำ (tokenization) หรือการตัด stopwords เช่น `the`, `and`, `is` ออก ในกรณีที่คำเหล่านี้ไม่ได้ช่วยในการวิเคราะห์

**ไม่บังคับให้ใช้ stemming หรือ lemmatization** หากไม่มีเหตุผลว่าขั้นตอนดังกล่าวจะช่วยตอบคำถามของทีมได้ดีขึ้น

หลังจากเตรียมข้อความแล้ว ให้สร้าง **Text Features อย่างน้อย 2 ชนิด** เพื่อนำข้อความมาเปลี่ยนให้อยู่ในรูปข้อมูลที่สามารถนำไปสำรวจ เปรียบเทียบ หรือวิเคราะห์ต่อได้

---

#### Text Feature คืออะไร?

Text Feature คือ **ข้อมูลเชิงตัวเลขที่สกัดหรือสร้างขึ้นจากข้อความ** เพื่อแทนลักษณะหรือความหมายบางอย่างของข้อความนั้น

Text Feature อาจอยู่ในรูปของ

* **ค่าตัวเลขค่าเดียว** เช่น จำนวนคำ หรือ sentiment score
* **ตัวแปรหลายตัว** เช่น `price=1`, `quality=0`, `delivery=1`
* **เวกเตอร์หลายมิติ** เช่น TF-IDF vector หรือ Text Embedding

ตัวอย่างเช่น รีวิว

> `"The product is easy to use but the delivery was very slow."`

สามารถเปลี่ยนเป็น feature ได้หลายรูปแบบ

| Feature                    |                 ตัวอย่างค่าที่ได้ |
| -------------------------- | --------------------------------: |
| จำนวนคำในรีวิว             |                                12 |
| Sentiment score            |                             -0.15 |
| กล่าวถึง usability หรือไม่ |                                 1 |
| กล่าวถึง delivery หรือไม่  |                                 1 |
| กล่าวถึง price หรือไม่     |                                 0 |
| TF-IDF vector              |   `[0.00, 0.41, 0.00, 0.72, ...]` |
| Embedding vector           | `[0.21, -0.08, 0.63, ..., -0.14]` |

ดังนั้นข้อความหนึ่งข้อความสามารถถูกแทนได้ทั้งด้วย **feature ที่ตีความได้โดยตรง** และ **feature แบบเวกเตอร์ที่ใช้แทนเนื้อหาหรือความหมายของข้อความ**

---

## ตัวอย่าง Text Features ที่สามารถเลือกใช้

ผมจะแก้ตารางเดิมเป็นแบบนี้:

| Text Feature          | คืออะไร                                     | ตัวอย่าง                     | ใช้ตอบคำถามอะไรได้                                    |
| --------------------- | ------------------------------------------- | ---------------------------- | ----------------------------------------------------- |
| **Review Length**     | จำนวนคำหรือตัวอักษร                         | 120 words                    | รีวิวที่ยาวกว่าได้รับ helpful vote มากกว่าหรือไม่     |
| **Sentiment Score**   | คะแนนความเป็นบวก/ลบ                         | `-0.72`, `+0.85`             | ความรู้สึกลูกค้าสัมพันธ์กับ rating หรือไม่            |
| **Aspect Indicators** | ระบุว่าพูดถึงหัวข้อใด                       | `delivery=1`, `price=0`      | ลูกค้าบ่นเรื่องใดมากที่สุด                            |
| **Lexical Diversity** | ความหลากหลายของคำ                           | unique words / total words   | รีวิวที่เขียนละเอียดมีประโยชน์มากกว่าหรือไม่          |
| **TF-IDF Vector**     | เวกเตอร์ที่ให้น้ำหนักคำเด่นในเอกสาร         | `[0, .35, 0, .81, ...]`      | คำใดแยกกลุ่มรีวิวบวกกับลบ                             |
| **Text Embedding**    | dense vector ที่แทนความหมายโดยรวมของข้อความ | `[.12, -.44, .08, ...]`      | รีวิวใดมีความหมายคล้ายกัน หรือมี customer themes อะไร |
| **Custom Feature**    | feature ที่ทีมสร้างตามโจทย์                 | `refund=1`, `defect_count=3` | ตอบ business question เฉพาะของทีม                     |

### ตัวอย่าง Embedding แบบง่าย

ถ้าสอน `sentence-transformers` มาแล้ว อาจใส่ตัวอย่างสั้น ๆ:

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

texts = [
    "The delivery was very slow.",
    "Shipping took too long.",
    "Excellent sound quality."
]

embeddings = model.encode(texts)

print(embeddings.shape)
```

**รายงานเพียง 3 อย่าง**
1. ทำ preprocessing อะไร และทำไม
2. สร้าง feature อะไรอย่างน้อย 2 ตัว
3. feature เหล่านั้นช่วยตอบคำถามทางธุรกิจอย่างไร

> ไม่ต้องรายงาน vocabulary ก่อน–หลัง เว้นแต่ประเด็นนั้นเกี่ยวข้องกับผลวิเคราะห์

````python
import ast
import html
import re
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns


# ==========================================
# 1. TEXT PREPROCESSING
# ==========================================

url_re = re.compile(r'https?://\S+|www\.\S+')
html_re = re.compile(r'<.*?>')
space_re = re.compile(r'\s+')

titles = df_joined['title_review'].fillna('').astype(str).tolist()
texts = df_joined['text'].fillna('').astype(str).tolist()


def clean_text_fast(title, text):
    combined = f'{title} {text}'
    combined = html.unescape(combined)
    combined = url_re.sub('', combined)
    combined = html_re.sub('', combined)
    combined = space_re.sub(' ', combined)
    return combined.strip().lower()


cleaned_list = [
    clean_text_fast(t, x)
    for t, x in zip(titles, texts)
]

df_joined['cleaned_text'] = cleaned_list


# ==========================================
# 2. TEXT FEATURES
# ==========================================

# ---------- 2.1 Review Length ----------
# นับจำนวนคำจาก Review Text โดยไม่รวม Title

review_only = [
    clean_text_fast('', x)
    for x in texts
]

df_joined['review_length'] = [
    len(x.split()) for x in review_only
]

# ---------- 2.2 Aspect Indicators ----------

# Sound Quality
ADJ_POS_NEG = (
    r'great|good|excellent|awesome|amazing|wonderful|superb|fantastic|'
    r'terrible|bad|poor|awful|clear|crisp|muddy|tinny|thin|flat'
)

sound_pattern = re.compile(
    r'\b(?:'
    r'sound quality|audio quality'
    rf'|(?:{ADJ_POS_NEG})\s+(?:sound|audio)'
    rf'|sound(?:s|ed)?\s+(?:is |are |was |so |very |really )?(?:{ADJ_POS_NEG})'
    r'|crystal clear|distortion|distorted|hi-?fi|fidelity|remaster(?:ed)?|mastering'
    r'|vocals? (?:are|is|sound)s?'
    r')\b',
    flags=re.IGNORECASE
)


# Nostalgia
nostalgia_pattern = re.compile(
    r'\b(?:'
    r'nostalgi(?:a|c)|childhood|brings? (?:me )?back|takes? me back'
    r'|back in the day|good old days|growing up|grew up|my youth'
    r'|reminds? me of|memories'
    r')\b',
    flags=re.IGNORECASE
)


# Value for Money
value_pattern = re.compile(
    r'\b(?:'
    r'worth (?:every|the|it)|good (?:price|value)|great (?:price|value)'
    r'|bargain|affordable|cheap|overpriced|rip-?off|value for money'
    r'|money well spent|waste of money|a steal'
    r')\b',
    flags=re.IGNORECASE
)


def flag(pattern, texts):
    return np.fromiter(
        (1 if pattern.search(x) else 0 for x in texts),
        dtype=np.int8
    )


df_joined['mention_sound'] = flag(
    sound_pattern, cleaned_list
)

df_joined['mention_nostalgia'] = flag(
    nostalgia_pattern, cleaned_list
)

df_joined['mention_value'] = flag(
    value_pattern, cleaned_list
)


# ==========================================
# 3. INFERRED GENRE
#    Rule-based Regex Classification
# ==========================================

def extract_clean_artist(store_val):
    if pd.isna(store_val):
        return 'Unknown'

    clean_val = re.sub(
        r'Format:.*$',
        '',
        str(store_val)
    )

    clean_val = re.sub(
        r'\(Artist\)',
        '',
        clean_val
    )

    return clean_val.strip()


df_joined['clean_artist'] = df_joined['store'].apply(
    extract_clean_artist
)


# กฎสำหรับจำแนกแนวเพลงด้วย Regular Expression
GENRE_RULES = [
    (
        'K-Pop / Asian Pop',
        r'\b(?:bts|bangtan|k-?pop|hanteo|photobook)\b'
    ),

    (
        'Classical / Jazz',
        r'\b(?:jazz|classical|symphony|orchestra|concerto|opera|yo-yo ma)\b'
    ),

    (
        'Classic Rock / Oldies',
        r"\b(?:beatles|elvis|pink floyd|grateful dead|rock|oldies|bbc sessions|60s|70s|80s)\b"
    ),

    (
        'Dance / R&B / Electronic',
        r'\b(?:r&b|house music|dj|remix|dance|electronic|edm)\b'
    ),

    (
        'Audiobook / Spoken Word',
        r'\b(?:unabridged|audiobook|spoken word)\b'
    ),
]

GENRE_RULES = [
    (g, re.compile(p, flags=re.IGNORECASE))
    for g, p in GENRE_RULES
]


def infer_genre(title, desc, artist):

    d = (
        ' '.join(map(str, desc))
        if isinstance(desc, list)
        else str(desc or '')
    )

    text = f'{title} {d} {artist}'.lower()

    for g, pat in GENRE_RULES:
        if pat.search(text):
            return g

    return 'General / Mainstream'


df_joined['inferred_genre'] = [
    infer_genre(t, d, a)
    for t, d, a in zip(
        df_joined['title_meta'],
        df_joined['description'],
        df_joined['clean_artist']
    )
]


# ==========================================
# 4. แสดงผลลัพธ์ TEXT FEATURES
# ==========================================

display(
    df_joined[
        [
            'title_review',
            'cleaned_text',
            'review_length',
            'mention_sound',
            'mention_nostalgia',
            'mention_value',
            'inferred_genre'
        ]
    ].head()
)


# ==========================================
# 5. แสดง Distribution ของ Inferred Genre
# ==========================================

print("\n=== Distribution of Inferred Genres ===")

print(
    df_joined['inferred_genre'].value_counts()
)

print("=" * 40)
````

````python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import re

# ==========================================
# 0. SAFETY CHECK: Ensure features exist
# ==========================================
FEATS = ['mention_sound', 'mention_nostalgia', 'mention_value']

# If columns are missing, re-generate them quickly
missing_feats = [col for col in FEATS if col not in df_joined.columns]
if missing_feats:
    print(f"Re-generating missing features in df_joined: {missing_feats}")

    # Ensure we have cleaned_text
    if 'cleaned_text' not in df_joined.columns:
        url_re = re.compile(r'https?://\S+|www\.\S+')
        html_re = re.compile(r'<.*?>')
        space_re = re.compile(r'\s+')
        import html

        titles = df_joined['title_review'].fillna('').astype(str).tolist()
        texts = df_joined['text'].fillna('').astype(str).tolist()
        cleaned_list = []
        for t, x in zip(titles, texts):
            combined = f'{t} {x}'
            combined = html.unescape(combined)
            combined = url_re.sub('', combined)
            combined = html_re.sub('', combined)
            combined = space_re.sub(' ', combined)
            cleaned_list.append(combined.strip().lower())
        df_joined['cleaned_text'] = cleaned_list

    cleaned_list = df_joined['cleaned_text'].tolist()

    # Compile patterns and flag
    ADJ_POS_NEG = r'great|good|excellent|awesome|amazing|wonderful|superb|fantastic|terrible|bad|poor|awful|clear|crisp|muddy|tinny|thin|flat'

    sound_pattern = re.compile(
        r'\b(?:sound quality|audio quality|(?:' + ADJ_POS_NEG + r')\s+(?:sound|audio)|sound(?:s|ed)?\s+(?:is |are |was |so |very |really )?(?:' + ADJ_POS_NEG + r')|crystal clear|distortion|distorted|hi-?fi|fidelity|remaster(?:ed)?|mastering|vocals? (?:are|is|sound)s?)\b',
        flags=re.IGNORECASE
    )
    nostalgia_pattern = re.compile(
        r'\b(?:nostalgi(?:a|c)|childhood|brings? (?:me )?back|takes? me back|back in the day|good old days|growing up|grew up|my youth|reminds? me of|memories)\b',
        flags=re.IGNORECASE
    )
    value_pattern = re.compile(
        r'\b(?:worth (?:every|the|it)|good (?:price|value)|great (?:price|value)|bargain|affordable|cheap|overpriced|rip-?off|value for money|money well spent|waste of money|a steal)\b',
        flags=re.IGNORECASE
    )

    df_joined['mention_sound'] = np.fromiter((1 if sound_pattern.search(x) else 0 for x in cleaned_list), dtype=np.int8)
    df_joined['mention_nostalgia'] = np.fromiter((1 if nostalgia_pattern.search(x) else 0 for x in cleaned_list), dtype=np.int8)
    df_joined['mention_value'] = np.fromiter((1 if value_pattern.search(x) else 0 for x in cleaned_list), dtype=np.int8)

# ==========================================
# 1. SET SPOTIFY THEME & PALETTE
# ==========================================
SPOTIFY_GREEN = '#1DB954'  # สีเขียวหลักสำหรับ Positive / HighImpact
NEGATIVE_RED = '#E24A4A'  # สีแดงสำหรับ Negative / General
BG_COLOR = '#FFFFFF'
TEXT_COLOR = '#000000'
GRID_COLOR = '#CCCCCC'

plt.rcParams.update({
    'figure.facecolor': BG_COLOR,
    'axes.facecolor': BG_COLOR,
    'axes.edgecolor': GRID_COLOR,
    'axes.linewidth': 1.2,
    'axes.labelcolor': TEXT_COLOR,
    'axes.titlecolor': TEXT_COLOR,
    'xtick.color': TEXT_COLOR,
    'ytick.color': TEXT_COLOR,
    'grid.color': GRID_COLOR,
    'grid.linestyle': '--',
    'grid.alpha': 0.6,
    'font.size': 10,
})

# ==========================================
# 2. DATA CALCULATION FUNCTION (NO P-VALUE)
# ==========================================
def compare_rates(a, b, name_a, name_b, feats=FEATS):
  print(f'n {name_a} = {len(a):,} | n {name_b} = {len(b):,}')
  rows = []
  for c in feats:
    xa, xb = int(a[c].sum()), int(b[c].sum())
    rows.append({
        'feature': c,
        f'{name_a} %': round(xa / len(a) * 100, 1),
        f'{name_a} count': xa,
        f'{name_b} %': round(xb / len(b) * 100, 1),
        f'{name_b} count': xb,
    })
  return pd.DataFrame(rows)

# ==========================================
# 3. PLOTTING FUNCTION
# ==========================================
def plot_compare_rates(
    res_df, name_a, name_b, title='Aspect Mention Rate Comparison'
):
  # แปลงข้อมูลเป็น Long Format โดยดึงทั้ง 2 กลุ่มมาแสดงผล
  melted_df = res_df.melt(
      id_vars=['feature'],
      value_vars=[f'{name_a} %', f'{name_b} %'],
      var_name='Group',
      value_name='Mention Rate (%)',
  )
  melted_df['Group'] = melted_df['Group'].str.replace(' %', '')
  melted_df['Aspect'] = (
      melted_df['feature'].str.replace('mention_', '').str.title()
  )

  fig, ax = plt.subplots(figsize=(9, 5.5), dpi=120)

  sns.barplot(
      data=melted_df,
      x='Aspect',
      y='Mention Rate (%)',
      hue='Group',
      ax=ax,
      palette=[SPOTIFY_GREEN, NEGATIVE_RED],
      edgecolor=TEXT_COLOR,
  )

  # ตัวเลขเปอร์เซ็นต์บนหัวแท่งกราฟ
  for p in ax.patches:
    h = p.get_height()
    if h > 0:
      ax.annotate(
          f'{h:.1f}%',
          (p.get_x() + p.get_width() / 2.0, h),
          ha='center',
          va='bottom',
          color=TEXT_COLOR,
          fontweight='bold',
          fontsize=9.5,
          xytext=(0, 3),
          textcoords='offset points',
      )

  # ตกแต่งรายละเอียดกราฟ
  ax.set_title(title, fontsize=13, color=TEXT_COLOR, fontweight='bold', pad=15)
  ax.set_xlabel('Marketing Aspect', fontsize=11, color=TEXT_COLOR)
  ax.set_ylabel('Mention Rate (%)', fontsize=11, color=TEXT_COLOR)

  # ปรับระยะขอบบนแกน Y ไม่ให้ตัวเลขติดขอบ
  max_val = melted_df['Mention Rate (%)'].max() if not melted_df.empty else 10
  ax.set_ylim(0, max_val * 1.18)

  ax.legend(
      title='Group',
      facecolor=BG_COLOR,
      edgecolor=GRID_COLOR,
      labelcolor=TEXT_COLOR,
  )
  ax.grid(True, axis='y')

  plt.tight_layout()
  plt.show()

# ==========================================
# 4. EXECUTION
# ==========================================
# 1) เปรียบเทียบ Positive vs Negative
pos_df = df_joined[df_joined['rating'] >= 4]
neg_df = df_joined[df_joined['rating'] <= 2]
res_pos_neg = compare_rates(pos_df, neg_df, 'Positive', 'Negative')
display(res_pos_neg)

plot_compare_rates(
    res_pos_neg,
    'Positive',
    'Negative',
    title='Sentiment Contrast: Aspect Mention Rate (Positive vs. Negative Reviews)',
)

# 2) เปรียบเทียบ HighImpact vs General
high_df = df_joined[df_joined['helpful_vote'] >= 5]
gen_df = df_joined[df_joined['helpful_vote'] < 5]
res_high_gen = compare_rates(high_df, gen_df, 'HighImpact', 'General')
display(res_high_gen)

plot_compare_rates(
    res_high_gen,
    'HighImpact',
    'General',
    title='Impact Contrast: Aspect Mention Rate (HighImpact vs. General Reviews)',
)
````

#### **High Helpful Vote Analysis (วิเคราะห์รีวิวอิทธิพลสูง)**

````python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# ==========================================
# 0. SAFETY CHECK: Ensure review_length exists
# ==========================================
if 'review_length' not in df_joined.columns:
    print("Re-generating missing 'review_length' column...")
    df_joined['word_count'] = df_joined['text'].fillna('').apply(lambda x: len(str(x).split()))
    df_joined['review_length'] = df_joined['word_count']

# ==========================================
# 1. SET SPOTIFY THEME & PALETTE
# ==========================================
SPOTIFY_GREEN = '#1DB954'  # High Impact Reviews
GENERAL_GRAY = '#535353'  # General Reviews
BG_COLOR = '#FFFFFF'
TEXT_COLOR = '#000000'
GRID_COLOR = '#CCCCCC'

plt.rcParams.update({
    'figure.facecolor': BG_COLOR,
    'axes.facecolor': BG_COLOR,
    'axes.edgecolor': GRID_COLOR,
    'axes.linewidth': 1.2,
    'axes.labelcolor': TEXT_COLOR,
    'axes.titlecolor': TEXT_COLOR,
    'xtick.color': TEXT_COLOR,
    'ytick.color': TEXT_COLOR,
    'grid.color': GRID_COLOR,
    'grid.linestyle': '--',
    'grid.alpha': 0.6,
    'font.size': 10,
})

# ==========================================
# 2. PREPARE DATA FOR PLOTTING (รวม Value for Money)
# ==========================================
# Define high_helpful and general_helpful based on helpful_vote
high_helpful = df_joined[df_joined['helpful_vote'] >= 5].copy()
general_helpful = df_joined[df_joined['helpful_vote'] < 5].copy()

# Data 1: Aspect Mention Rates (%) รวม Sound, Nostalgia, Value
aspect_data = pd.DataFrame([
    {
        'Aspect': 'Sound Quality',
        'General Reviews (< 5)': general_helpful['mention_sound'].mean() * 100,
        'High Impact Reviews (≥ 5)': high_helpful['mention_sound'].mean() * 100,
    },
    {
        'Aspect': 'Nostalgia',
        'General Reviews (< 5)': (
            general_helpful['mention_nostalgia'].mean() * 100
        ),
        'High Impact Reviews (≥ 5)': (
            high_helpful['mention_nostalgia'].mean() * 100
        ),
    },
    {
        'Aspect': 'Value for Money',
        'General Reviews (< 5)': general_helpful['mention_value'].mean() * 100,
        'High Impact Reviews (≥ 5)': high_helpful['mention_value'].mean() * 100,
    },
])
aspect_melted = aspect_data.melt(
    id_vars=['Aspect'],
    var_name='Review Group',
    value_name='Mention Rate (%)',
)

# Data 2: Avg Review Length (Words)
length_data = pd.DataFrame([{
    'General Reviews (< 5)': general_helpful['review_length'].mean(),
    'High Impact Reviews (≥ 5)': high_helpful['review_length'].mean(),
}])
length_melted = length_data.melt(
    var_name='Review Group', value_name='Avg Words'
)

# ==========================================
# 3. PLOT SIDE-BY-SIDE SUBPLOTS (1x2)
# ==========================================
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5.5), dpi=120)

# --- Subplot 1: Aspect Mention Rate (%) ---
sns.barplot(
    data=aspect_melted,
    x='Aspect',
    y='Mention Rate (%)',
    hue='Review Group',
    ax=ax1,
    palette=[GENERAL_GRAY, SPOTIFY_GREEN],
    edgecolor=TEXT_COLOR,
)

for p in ax1.patches:
  h = p.get_height()
  if h > 0:
    ax1.annotate(
        f'{h:.1f}%',
        (p.get_x() + p.get_width() / 2.0, h),
        ha='center',
        va='bottom',
        color=TEXT_COLOR,
        fontweight='bold',
        fontsize=9.5,
    )

ax1.set_title(
    'Aspect Mention Rate Comparison (%)',
    fontsize=12,
    color=TEXT_COLOR,
    fontweight='bold',
    pad=12,
)
ax1.set_xlabel('Marketing Aspect', fontsize=11, color=TEXT_COLOR)
ax1.set_ylabel('Mention Rate (%)', fontsize=11, color=TEXT_COLOR)
ax1.legend(
    title='Group',
    facecolor=BG_COLOR,
    edgecolor=GRID_COLOR,
    labelcolor=TEXT_COLOR,
)
ax1.grid(True)

# --- Subplot 2: Avg Review Length (Words) ---
sns.barplot(
    data=length_melted,
    x='Review Group',
    y='Avg Words',
    hue='Review Group',
    ax=ax2,
    palette=[GENERAL_GRAY, SPOTIFY_GREEN],
    edgecolor=TEXT_COLOR,
    legend=False,
)

for p in ax2.patches:
  h = p.get_height()
  if h > 0:
    ax2.annotate(
        f'{h:.1f} words',
        (p.get_x() + p.get_width() / 2.0, h),
        ha='center',
        va='bottom',
        color=TEXT_COLOR,
        fontweight='bold',
        fontsize=9.5,
    )

ax2.set_title(
    'Average Review Length Comparison (Words)',
    fontsize=12,
    color=TEXT_COLOR,
    fontweight='bold',
    pad=12,
)
ax2.set_xlabel('Review Group', fontsize=11, color=TEXT_COLOR)
ax2.set_ylabel('Avg Review Length (Words)', fontsize=11, color=TEXT_COLOR)
ax2.grid(True)

fig.suptitle(
    'High Impact vs. General Reviews Analysis',
    fontsize=15,
    fontweight='bold',
    color=TEXT_COLOR,
    y=1.02,
)
plt.tight_layout()
plt.show()
````

ผลการวิเคราะห์พบว่า **High Impact Reviews** มีการกล่าวถึงทั้ง 3 Marketing Aspects มากกว่า General Reviews โดย **Sound Quality** มีความแตกต่างชัดเจนที่สุด (15.2% เทียบกับ 5.0%) รองลงมาคือ **Value for Money (9.3%)** และ **Nostalgia (6.5%)**

**Marketing Insight:** จึงสามารถใช้ **Sound Quality เป็นจุดขายหลัก** ในการสื่อสารการตลาด และใช้ Value for Money หรือ Nostalgia เป็นข้อความเสริมตามกลุ่มเป้าหมาย

**Marketing Action:** นำคำและประเด็นที่ลูกค้าใช้จริงไปพัฒนาเป็น **Ad Headline / Ad Copy** แล้วทำ **A/B Test** เพื่อวัด CTR หรือ Conversion

**ข้อจำกัด:** ผลนี้แสดงเพียงว่า High Impact Reviews **พูดถึง Aspect เหล่านี้บ่อยกว่า** ยังไม่สามารถสรุปได้ว่า Aspect เหล่านี้เป็นสาเหตุที่ทำให้รีวิวได้รับ Helpful Vote สูง





## 2.6 Reality Check

ก่อนเอา insight ไปบอกลูกค้า ให้ลองหักล้างมันด้วยตัวเองก่อน อย่างน้อย 1 insight ต้องผ่านขั้นนี้

คำอธิบายทางเลือกที่พบบ่อย
* **ผลของขนาด:** กลุ่มที่ดูโดดเด่นอาจมีตัวอย่างน้อยมาก ตรวจ n ของทุกกลุ่ม
* **ผลของส่วนผสม:** ความต่างอาจมาจากสัดส่วนสินค้าในแต่ละกลุ่มไม่เหมือนกัน ไม่ใช่จากพฤติกรรมลูกค้า ลองคุมให้เทียบกันเฉพาะสินค้าประเภทเดียวกัน
* **ผลของเวลา:** ของเก่ามีเวลาสะสม vote และรีวิวมากกว่า
* **ผลของความยาว:** ข้อความยาวมีโอกาสมีคำที่เราค้นหามากกว่าโดยอัตโนมัติ
* **ผลของคนไม่กี่คน:** ผู้ใช้หรือสินค้าไม่กี่รายอาจครองสัดส่วนใหญ่ของกลุ่ม

เลือกอย่างน้อยหนึ่งคำอธิบาย แล้ว **ทดสอบด้วยตัวเลขจริง** สรุปว่า insight ยังยืนอยู่ได้ หรือต้องปรับถ้อยคำให้แคบลง

#### **1. ผลของความยาวรีวิว (Review Length Effect) ที่ทำให้ % Mention Rate ของกลุ่ม Non-Verified สูงกว่า Verified**

#### **2. ผลของขนาดตัวอย่าง (Sample Size Effect) มีจำนวนตัวอย่างน้อยเกินไป ($n < 100$) จนทำให้อัตรา Nostalgia ผันผวนหรือไม่**

````python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# ==========================================
# CONSTANTS & THEME SETUP (MATCHING IMAGE STYLE)
# ==========================================
BG_COLOR = "#FFFFFF"
DARK_GRAY = "#505050"
SPOTIFY_GREEN = "#2B9F5E"
MID_GRAY = "#8C8C8C"
ALERT_RED = "#D32F2F"
SPINE_COLOR = "#CCCCCC"
GRID_COLOR = "#E0E0E0"
BAR_EDGE = "#1A1A1A"

# ==========================================
# เตรียมข้อมูลสำหรับ Reality Check
# ==========================================
# 1) คำนวณ verified_aspects จากข้อมูลจริง
verified_aspects = (
    df_joined.groupby("verified_purchase")[
        ["mention_sound", "mention_nostalgia", "mention_value"]
    ].mean()
    * 100
)
verified_aspects.columns = ["Sound Quality", "Nostalgia", "Value for Money"]

# 2) คำนวณ genre_counts จากข้อมูลจริง
genre_counts = df_joined["inferred_genre"].value_counts()


# ==========================================
# GRAPH 1: Verified vs Non-Verified Purchase
# ==========================================
plot_df1 = verified_aspects.copy()
plot_df1.index = ["Non-Verified", "Verified"]

fig, ax1 = plt.subplots(figsize=(8, 5), dpi=120, facecolor=BG_COLOR)
ax1.set_facecolor(BG_COLOR)

# กำหนด Palette 3 สีให้เข้ากับโทนในรูป
colors_g1 = [SPOTIFY_GREEN, DARK_GRAY, MID_GRAY]

plot_df1.plot(
    kind="bar",
    ax=ax1,
    width=0.6,
    color=colors_g1,
    edgecolor=BAR_EDGE,
    linewidth=0.8,
)

# ตกแต่งชื่อและป้ายแกน
ax1.set_title(
    "Reality Check 1: Aspect Mention Rate by Purchase Status",
    fontsize=13,
    fontweight="bold",
    color="#000000",
    pad=15,
)
ax1.set_xlabel("Purchase Status", fontsize=11, fontweight="bold", color="#000000")
ax1.set_ylabel("Mention Rate (%)", fontsize=11, fontweight="bold", color="#000000")
ax1.tick_params(axis="x", rotation=0, labelsize=10, colors="#000000")
ax1.tick_params(axis="y", labelsize=10, colors="#000000")

# ปรับ Spines & Grid ตามแบบในรูป (เปิดเฉพาะซ้าย-ล่าง สีเทาอ่อน)
ax1.spines["top"].set_visible(False)
ax1.spines["right"].set_visible(False)
ax1.spines["left"].set_color(SPINE_COLOR)
ax1.spines["bottom"].set_color(SPINE_COLOR)
ax1.grid(axis="y", linestyle="--", alpha=0.7, color=GRID_COLOR)

# ตั้งค่า Y-lim และ Legend
ax1.set_ylim(0, max(plot_df1.values.max() * 1.25, 20))
ax1.legend(
    title="Aspects",
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    frameon=True,
    facecolor="white",
    edgecolor=SPINE_COLOR,
    fontsize=9.5,
)

# ใส่ตัวเลขเปอร์เซ็นต์กำกับบนแท่งกราฟ (ตัวหนา)
for container in ax1.containers:
  ax1.bar_label(
      container,
      fmt="%.1f%%",
      padding=3,
      fontsize=9.5,
      fontweight="bold",
      color="#000000",
  )

plt.tight_layout()
plt.show()


# ==========================================
# GRAPH 2: Genre Sample Size Constraint Check
# ==========================================
df_genre = genre_counts.reset_index()
df_genre.columns = ["genre", "count"]

# กำหนดสีแท่งกราฟ (เขียว = ผ่าน threshold, แดง = ไม่ถึง threshold)
colors_g2 = [
    ALERT_RED if count < 100 else SPOTIFY_GREEN for count in df_genre["count"]
]

fig, ax2 = plt.subplots(figsize=(10, 6), dpi=120, facecolor=BG_COLOR)
ax2.set_facecolor(BG_COLOR)

bars = ax2.barh(
    df_genre["genre"],
    df_genre["count"],
    color=colors_g2,
    height=0.6,
    edgecolor=BAR_EDGE,
    linewidth=0.8,
)

# เส้น Threshold ที่ n = 100
ax2.axvline(
    x=100,
    color=ALERT_RED,
    linestyle="--",
    linewidth=1.5,
    label="Min Threshold (n = 100)",
)

# ตกแต่งชื่อและป้ายแกน
ax2.set_title(
    "Reality Check 2: Sample Size by Genre (Highlighting Unreliable Samples)",
    fontsize=13,
    fontweight="bold",
    color="#000000",
    pad=15,
)
ax2.set_xlabel("Sample Size (n)", fontsize=11, fontweight="bold", color="#000000")
ax2.set_ylabel("Genre", fontsize=11, fontweight="bold", color="#000000")
ax2.tick_params(axis="x", labelsize=10, colors="#000000")
ax2.tick_params(axis="y", labelsize=10, colors="#000000")

# ปรับ Spines & Grid
ax2.spines["top"].set_visible(False)
ax2.spines["right"].set_visible(False)
ax2.spines["left"].set_color(SPINE_COLOR)
ax2.spines["bottom"].set_color(SPINE_COLOR)
ax2.grid(axis="x", linestyle="--", alpha=0.7, color=GRID_COLOR)

ax2.invert_yaxis()
ax2.legend(
    loc="lower right",
    frameon=True,
    facecolor="white",
    edgecolor=SPINE_COLOR,
    fontsize=9.5,
)

# ตัวเลขจำนวน n ท้ายแท่งกราฟ
max_c = max(df_genre["count"]) if not df_genre.empty else 100
for bar in bars:
  width = bar.get_width()
  is_small = width < 100
  ax2.text(
      width + (max_c * 0.01),
      bar.get_y() + bar.get_height() / 2,
      f"n={int(width)}",
      va="center",
      ha="left",
      fontsize=9.5,
      color=ALERT_RED if is_small else "#000000",
      fontweight="bold",
  )

ax2.set_xlim(0, max_c * 1.15)
plt.tight_layout()
plt.show()
````

#### Step 1 — กำหนดเว็บไซต์และสิ่งที่ต้องการเก็บ

เริ่มจากกำหนดให้ชัดก่อนว่าเราจะเก็บอะไร

ในตัวอย่างนี้
* `LISTING_URL` = หน้ารวมข่าว ใช้สำหรับ **หา link**
* `ARTICLE_PATH` = รูปแบบ URL ที่ถือว่าเป็น **บทความจริง**
* `MAX_ARTICLES` = จำนวนข่าวที่ทดลองเก็บ
* `DELAY` = เวลาพักระหว่างการเรียกแต่ละหน้า เพื่อไม่ยิง request ถี่เกินไป

จุดสำคัญคือ **หน้ารวมข่าวไม่ใช่ record ที่จะเอาไปวิเคราะห์** แต่เป็นเพียงทางเข้าไปหา article URLs


````python
# ===== Step 1: ตั้งค่า =====
# หากยังไม่มี package:
# !pip install -q requests beautifulsoup4

import re
import time
import requests
import pandas as pd

from bs4 import BeautifulSoup
from urllib.parse import urljoin, urlparse, urldefrag
from urllib.robotparser import RobotFileParser

BASE_URL = "https://www.thaipbs.or.th"
LISTING_URL = "https://www.thaipbs.or.th/news"

# Thai PBS ใช้ URL ข่าวลักษณะ /news/content/509281
ARTICLE_PATH_PATTERN = r"^/news/content/\d+$"

MAX_ARTICLES = 20     # เริ่มน้อย ๆ เพื่อทดสอบก่อน
DELAY = 1.5           # พัก 1.5 วินาทีระหว่างหน้า
BOT_USER_AGENT = "KKU-TextAnalytics-StudentBot/1.0"

session = requests.Session()
session.headers.update({"User-Agent": BOT_USER_AGENT})

print("Listing page:", LISTING_URL)
print("เป้าหมาย: 1 row = 1 news article")
````

#### Step 2 — ทำความสะอาด URL และตรวจ `robots.txt`

URL เดียวกันอาจเขียนได้หลายแบบ เช่น

`https://www.thaipbs.or.th/news`  
`https://www.thaipbs.or.th/news#main-content`

ส่วนหลัง `#...` เรียกว่า **fragment** และโดยทั่วไปไม่ได้หมายถึงหน้าใหม่ จึงควรตัดออกก่อนตรวจ URL ซ้ำ

จากนั้นตรวจ `robots.txt` ก่อนเข้าแต่ละส่วนของเว็บ ถ้าไม่อนุญาตให้ bot เข้า path นั้น ให้ **เปลี่ยนแหล่งข้อมูล** ไม่ใช่หาวิธีหลบ


````python
# ===== Step 2: normalize URL + robots.txt =====

_robots_cache = {}

def normalize_url(url):
    """ตัด #fragment ออกจาก URL เพื่อไม่ให้หน้าเดียวกันถูกนับซ้ำ"""
    clean_url, _ = urldefrag(url)
    return clean_url.rstrip("/")


def robots_allowed(url, user_agent=BOT_USER_AGENT):
    """คืน True เมื่อ robots.txt อนุญาตให้ bot เข้าถึง URL"""
    p = urlparse(url)
    root = f"{p.scheme}://{p.netloc}"

    if root not in _robots_cache:
        rp = RobotFileParser()
        rp.set_url(urljoin(root, "/robots.txt"))
        try:
            rp.read()
            _robots_cache[root] = rp
        except Exception as e:
            print("อ่าน robots.txt ไม่ได้:", e)
            return False

    return _robots_cache[root].can_fetch(user_agent, url)


print("robots.txt อนุญาตหน้ารวมข่าวหรือไม่:", robots_allowed(LISTING_URL))
````

#### Step 3 — เข้า “หน้ารวมข่าว” เพื่อหาเฉพาะ Article URLs

ขั้นนี้ **ยังไม่เก็บเนื้อหาข่าว**

Robot เปิดหน้ารวมข่าว แล้วดู `<a href="...">` ทุก link จากนั้นกรองให้เหลือเฉพาะ URL ที่ตรงกับรูปแบบบทความ เช่น

`/news/content/509281`

การกรองนี้คือสิ่งที่ทำให้ Robot ต่างจาก generic crawler ที่วิ่งไปทุกหน้าใน domain


````python
# ===== Step 3: หา link ของ "บทความจริง" =====

def find_article_links(listing_url):
    if not robots_allowed(listing_url):
        raise PermissionError("robots.txt ไม่อนุญาตให้เข้าหน้ารวมข่าวนี้")

    r = session.get(listing_url, timeout=20)
    r.raise_for_status()

    soup = BeautifulSoup(r.text, "html.parser")
    article_urls = set()

    for a in soup.find_all("a", href=True):
        url = urljoin(listing_url, a["href"])
        url = normalize_url(url)

        p = urlparse(url)

        # 1) เอาเฉพาะ domain เดียวกับเว็บไซต์หลัก
        same_domain = (p.netloc == urlparse(BASE_URL).netloc)

        # 2) เอาเฉพาะ path ที่มีรูปแบบเป็นบทความข่าว
        is_article = bool(re.match(ARTICLE_PATH_PATTERN, p.path))

        if same_domain and is_article:
            article_urls.add(url)

    return sorted(article_urls)


article_urls = find_article_links(LISTING_URL)

print("พบ article URLs =", len(article_urls))
print("ตัวอย่าง 5 URLs:")
for u in article_urls[:5]:
    print(" -", u)
````

#### Step 4 — เปิดบทความทีละหน้าแล้วดึง “เนื้อหาที่ต้องการ”

ตอนนี้เรามี URL ของบทความแล้ว จึงเปิดทีละข่าวและแยก field ที่จะใช้วิเคราะห์ เช่น

* `url`
* `title`
* `published_at`
* `text`
* `word_count`

หลักสำคัญคือ **พยายามดึงเฉพาะตัวบทความ** ไม่ใช่ menu/footer/ข่าวแนะนำทั้งหน้า

ตัวอย่างใช้ `<article>` ก่อน ถ้าเว็บไซต์ไม่มี tag นี้จึง fallback ไปที่ `<main>` และนำข้อความจาก `<p>` มาต่อกัน  
เว็บไซต์อื่นอาจต้องเปลี่ยน selector เช่น `.article-body`, `.post-content` หรือ selector เฉพาะของเว็บนั้น


````python
# ===== Step 4: ดึงข้อมูลจากบทความ 1 หน้า =====

def extract_article(url):
    if not robots_allowed(url):
        print("robots.txt ไม่อนุญาต ข้าม:", url)
        return None

    r = session.get(url, timeout=20)
    r.raise_for_status()

    soup = BeautifulSoup(r.text, "html.parser")

    # ---- title ----
    h1 = soup.find("h1")
    title = h1.get_text(" ", strip=True) if h1 else ""

    # ---- published date: พยายามอ่านจาก meta หรือ time ----
    published_at = ""

    meta_time = soup.find("meta", attrs={"property": "article:published_time"})
    if meta_time and meta_time.get("content"):
        published_at = meta_time["content"]
    else:
        time_tag = soup.find("time")
        if time_tag:
            published_at = time_tag.get("datetime") or time_tag.get_text(" ", strip=True)

    # ---- article text ----
    container = soup.find("article") or soup.find("main")

    if container is None:
        return None

    paragraphs = []
    for p in container.find_all("p"):
        txt = p.get_text(" ", strip=True)

        # ตัดบรรทัดสั้น ๆ ที่มักไม่ใช่เนื้อหาหลัก
        if len(txt) >= 30:
            paragraphs.append(txt)

    text = " ".join(paragraphs).strip()

    # ถ้าดึงมาแล้วสั้นมาก มีโอกาสว่า selector ไม่ถูก
    if len(text) < 200:
        return None

    return {
        "doc_id": normalize_url(url),
        "url": normalize_url(url),
        "title": title,
        "published_at": published_at,
        "text": text,
        "word_count": len(text.split()),
        "source_domain": urlparse(url).netloc,
    }
````

#### Step 5 — ให้ Robot ทำซ้ำกับหลายบทความ

นี่คือส่วนที่ทำให้มันเป็น “Robot”

แทนที่จะเปิดข่าวด้วยมือทีละหน้า โปรแกรมจะวน URL ที่หาได้ใน Step 3 แล้วเรียก `extract_article()` ซ้ำ ๆ

มี `time.sleep(DELAY)` ทุกครั้งเพื่อไม่ส่ง request ถี่เกินไป

ในช่วงพัฒนาให้เริ่มจาก 5–20 ข่าวก่อน เมื่อดูตัวอย่างแล้วแน่ใจว่า `text` เป็นเนื้อหาข่าวจริง ค่อยเพิ่มจำนวน


````python
# ===== Step 5: เก็บหลายบทความอัตโนมัติ =====

def collect_articles(article_urls, max_articles=20, delay=1.5):
    rows = []

    for i, url in enumerate(article_urls[:max_articles], start=1):
        try:
            item = extract_article(url)

            if item is not None:
                rows.append(item)
                print(f"[{i}] OK   :", item["title"][:70])
            else:
                print(f"[{i}] SKIP :", url)

        except Exception as e:
            print(f"[{i}] ERROR:", url, e)

        time.sleep(delay)

    return rows


rows = collect_articles(
    article_urls,
    max_articles=MAX_ARTICLES,
    delay=DELAY
)

print("\nเก็บบทความที่ใช้ได้ =", len(rows))
````

#### Step 6 — ตรวจว่า “1 row = 1 ข่าว” จริงหรือไม่

อย่าเชื่อเพียงว่าโค้ดรันผ่าน ต้องเปิดดูตัวอย่างข้อมูลด้วย

สิ่งที่ต้องตรวจ:
1. URL แต่ละแถวเป็น article URL จริงหรือไม่
2. `title` ตรงกับข่าวหรือไม่
3. `text` เป็นเนื้อหาข่าว ไม่ใช่ menu/footer
4. มีข่าวซ้ำหรือไม่
5. ข้อความสั้นผิดปกติหรือไม่

หลังผ่านขั้นนี้จึงค่อยบันทึกเป็น corpus และนำไป Text Analytics


````python
# ===== Step 6: Quality Check =====

df_robot = pd.DataFrame(rows)

if len(df_robot) > 0:
    print("จำนวนก่อน clean =", len(df_robot))
    print("URL ซ้ำ =", df_robot["url"].duplicated().sum())
    print("ข้อความซ้ำ =", df_robot["text"].duplicated().sum())

    # ตัด URL ซ้ำและข้อความซ้ำ
    df_robot = (
        df_robot
        .drop_duplicates(subset=["url"])
        .drop_duplicates(subset=["text"])
        .copy()
    )

    # ตัวอย่างเกณฑ์ง่าย ๆ: ข่าวต้องมีอย่างน้อย 80 คำ
    df_robot = df_robot[df_robot["word_count"] >= 80].copy()

    print("จำนวนหลัง clean =", len(df_robot))

    display(
        df_robot[
            ["title", "published_at", "word_count", "url"]
        ].head(10)
    )

    # อ่านเนื้อหาจริง 1 ข่าวเพื่อตรวจด้วยตา
    print("\nตัวอย่าง text:\n")
    print(df_robot.iloc[0]["text"][:1500])
else:
    print("ยังไม่มีข้อมูล ตรวจ article pattern / selector / robots.txt")
````
