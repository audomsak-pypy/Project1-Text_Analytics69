## เทคนิคที่อาจใช้ — เลือกเท่าที่จำเป็น

ไม่ต้องใช้ทั้งหมด โดยทั่วไป **2–4 เทคนิคก็เพียงพอ** ให้เลือกตามคำถามของทีม

| เทคนิค | เหมาะเมื่ออยากรู้ |
|---|---|
| **Word / n-gram frequency** | คนพูดถึงอะไรบ่อย |
| **TF-IDF / distinctive words** | สองกลุ่มพูดต่างกันอย่างไร |
| **Sentiment** | ข้อความเป็นบวก/ลบเพียงใด |
| **Aspect analysis** | คนพูดถึงราคา คุณภาพ การส่งของ ฯลฯ อย่างไร |
| **Topic modeling** | มีประเด็นอะไรที่เราอาจยังไม่รู้ |
| **Trend / segmentation** | ปัญหาเปลี่ยนตามเวลา หรือหนักในกลุ่มไหน |

> เทคนิคขั้นสูง เช่น embedding, clustering, regression, classification หรือ anomaly detection ทำได้ แต่ **ไม่บังคับ** และเหมาะสำหรับ Bonus เมื่อใช้ถูกและอธิบายได้

ถ้ายังไม่เคยใช้เทคนิคใด ให้ทดลองกับข้อมูลชุดเล็กก่อน แล้วตรวจด้วยตาว่าผลสมเหตุสมผลหรือไม่


## แบบฟอร์ม Insight Card

ทุก insight เขียนในรูปแบบนี้ 5 บรรทัด ไม่ต้องยาว

> **หลักฐาน:** ตัวเลขพร้อม n และชี้ว่าดูได้จากกราฟหรือตารางไหน \
> **So what:** กระทบธุรกิจอย่างไร ถ้าประมาณขนาดผลกระทบเป็นตัวเลขได้ยิ่งดี \

> **ความมั่นใจ:** สูง กลาง หรือ ต่ำ พร้อมเหตุผลสั้น ๆ

**ตัวอย่างที่ยังไม่ใช่ insight**

> **So what:** สินค้า 12 ชิ้นนี้คิดเป็น 18% ของยอดรีวิวทั้งหมวด ถ้าแก้ปัญหาฝาขวดได้ คาดว่าคะแนนเฉลี่ยของหมวดจะขยับขึ้นราว 0.15 ดาว \

### 2.2 เตรียมข้อมูลที่เชื่อถือได้

ทำเพียงสิ่งจำเป็นก่อนเริ่มวิเคราะห์

* สุ่มรีวิวอย่างน้อย **20,000 รายการ** และอย่าใช้เพียง N บรรทัดแรก
* เชื่อม review กับ product metadata ด้วย `parent_asin`
* รายงานสั้น ๆ ว่า join สำเร็จกี่เปอร์เซ็นต์ และมี missing/duplicate สำคัญหรือไม่
* เขียน 2–3 บรรทัดว่า **ข้อมูลนี้เป็นตัวแทนของใคร และไม่เป็นตัวแทนของใคร**

> จุดประสงค์คือให้รู้ว่าข้อมูลที่ใช้เชื่อถือได้พอสำหรับคำถามของทีม ไม่ใช่ทำ data cleaning ทุกอย่างที่เป็นไปได้


````python
def iter_jsonl(path):
    with open(path, encoding='utf-8') as f:
        for line in f:
            yield json.loads(line)
````

### Sampling Reviews from `review_path`

````python
# Sample N reviews from the review_path
print(f"Sampling {SAMPLE_N:,} reviews from {review_path}...")

sampled_reviews = []
for i, review in enumerate(iter_jsonl(review_path)):
    if len(sampled_reviews) < SAMPLE_N:
        sampled_reviews.append(review)
    else:
        # Reservoir sampling to get a random sample of N reviews
        r = random.randint(0, i)
        if r < SAMPLE_N:
            sampled_reviews[r] = review

df_reviews = pd.DataFrame(sampled_reviews)
print(f"Loaded {len(df_reviews):,} sampled reviews.")

# Load all meta data
print(f"Loading metadata from {meta_path}...")
df_meta = pd.DataFrame(iter_jsonl(meta_path))
print(f"Loaded {len(df_meta):,} metadata items.")

display(df_reviews.head())
````

### Joining Reviews with Product Metadata

````python
# Perform a left merge to enrich reviews with product metadata
initial_review_count = len(df_reviews)
df_joined = pd.merge(df_reviews, df_meta, on='parent_asin', how='left', suffixes=('_review', '_meta'))

# Report join success percentage
successful_join_count = len(df_joined.dropna(subset=['main_category'])) # Assuming main_category is a good indicator of a successful meta join
join_percentage = (successful_join_count / initial_review_count) * 100
print(f"Successfully joined {successful_join_count:,} reviews with metadata ({join_percentage:.2f}%).")

# Check for missing values in critical columns
print("\nMissing values after join:")
missing_values = df_joined[['text', 'title_review', 'main_category', 'price']].isnull().sum()
display(missing_values[missing_values > 0])

# Check for duplicate reviews (based on a combination of user_id, timestamp, and asin)
# Note: parent_asin is the product group, asin is the specific product variant.
# We are interested in unique reviews, which is usually identified by user_id and timestamp for a product.
df_joined_before_dedup = len(df_joined)
df_joined = df_joined.drop_duplicates(subset=['user_id', 'timestamp', 'asin'])
duplicate_removed_count = df_joined_before_dedup - len(df_joined)
print(f"\nRemoved {duplicate_removed_count:,} duplicate reviews.")

display(df_joined.head())
````

ชุดข้อมูลนี้เป็นตัวแทนของรีวิวจำนวน 20,000 รายการ ที่ถูกสุ่มเลือกมาจากสินค้าในหมวดหมู่ 'Digital_Music' บน Amazon โดยรีวิวเหล่านี้รวมถึงข้อมูลอธิบายเพิ่มเติม (Metadata) ที่เกี่ยวข้อง เช่น ชื่อสินค้า หมวดหมู่ และคะแนนดาวเฉลี่ย ข้อมูลตัวอย่างนี้ช่วยให้เห็นภาพรวมความรู้สึกของลูกค้า (Customer Sentiment) และลักษณะเฉพาะของสินค้าในหมวดหมู่นี้

### 2.3 สำรวจข้อมูลเท่าที่จำเป็น

เลือกดูเฉพาะสิ่งที่ช่วยให้เข้าใจ brief ของทีม เช่น

* การแจกแจงดาว
* verified purchase / helpful vote
* ราคา หรือกลุ่มสินค้า
* เวลา

ทำ **2–4 ตารางหรือกราฟ** ก็พอ แล้วเขียนสั้น ๆ ว่า

> “สิ่งใดจาก EDA ทำให้ทีมอยากวิเคราะห์ต่อ และเพราะอะไร”

ไม่ต้องทำ EDA ทุกตัวแปร


````python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Prepare data for plotting
# Calculate counts of ratings for verified and non-verified purchases
rating_counts = df_reviews.groupby(['verified_purchase', 'rating']).size().unstack(fill_value=0)

# Calculate percentages for each group (row-wise)
rating_percentages = rating_counts.apply(lambda x: x / x.sum(), axis=1)

# Melt the dataframe for seaborn's barplot
df_plot = rating_percentages.stack().reset_index(name='percentage')
df_plot['rating'] = df_plot['rating'].astype(str) + ' Star'

# --- Plotting ---

# 1. Set up the plot with Spotify Light Mode style
plt.style.use('default')
fig, ax = plt.subplots(figsize=(10, 6), facecolor='#FFFFFF')
ax.set_facecolor('#FFFFFF')

# Spotify palette: Verified = Bright Green, Non-Verified = Dark Charcoal
spotify_colors = {
    True: '#1DB954',
    False: '#191414'
}

# Create the clustered bar chart
sns.barplot(
    data=df_plot,
    x='rating',
    y='percentage',
    hue='verified_purchase',
    palette=spotify_colors,
    ax=ax,
    edgecolor='none' # Remove bar borders
)

# 3. Apply Spotify Minimal Styling
ax.set_title('Rating Distribution by Verified Purchase Status', fontsize=18, fontweight='bold', color='#191414', pad=20)
ax.set_xlabel('') # Hide x-axis label
ax.set_ylabel('') # Hide y-axis label

# Adjust x-axis tick labels for clarity
ax.tick_params(axis='x', colors='#191414', labelsize=12, rotation=0)
ax.tick_params(axis='y', colors='#555555', labelsize=11)

# Remove all spines
for spine in ['top', 'right', 'left', 'bottom']:
    ax.spines[spine].set_visible(False)

# Add horizontal grid lines
ax.grid(axis='y', linestyle='--', alpha=0.4, color='#CCCCCC')

# Add annotations (percentages) on top of each bar
for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.annotate(f'{height:.1%}',
                    (p.get_x() + p.get_width() / 2., height),
                    ha='center', va='bottom', xytext=(0, 5),
                    textcoords='offset points', fontsize=10, color='#191414', fontweight='bold')

# Adjust legend
handles, labels = ax.get_legend_handles_labels()
ax.legend(handles=handles, labels=['Verified Purchase', 'Non-Verified Purchase'],
          title='Purchase Status', frameon=False, fontsize=10, title_fontsize=12,
          labelcolor='#191414', loc='upper left')

plt.tight_layout()
plt.show()
````

**ผลการวิเคราะห์ข้อมูล:**

**การกระจายตัวเอียงไปทางเชิงบวกอย่างมาก (Positive Skewness)**:ไม่ว่าจะเป็นผู้ซื้อที่ได้รับการยืนยันหรือไม่ยืนยัน เสียงส่วนใหญ่กระจุกตัวอยู่ที่ 5 ดาว สูงที่สุด (เกิน 70% ทั้งสองกลุ่ม) แสดงให้เห็นว่าลูกค้ารายการนี้พึงพอใจสูงมาก  

**กลุ่ม Non-Verified มีแนวโน้มให้ 5 ดาวสูงกว่าอย่างมีนัยสำคัญ:**
 - กลุ่ม Non-Verified (79.8%) มีสัดส่วนรีวิว 5 ดาว สูงกว่ากลุ่ม Verified (70.5%) ถึงเกือบ 10%  
 - ข้อสังเกตนี้อาจสะท้อนถึงอคติเชิงบวก (Positivity Bias) หรือกลุ่มผู้รีวิวที่ไม่แน่ชัดว่าซื้อจริง ซึ่งมักกระจุกตัวให้คะแนนเต็มได้ง่ายกว่า   

**ผู้ซื้อจริง (Verified Purchase) มีความประณีต/วิพากษ์วิจารณ์มากกว่า (More Critical):**
 - กลุ่มผู้ซื้อจริงมีสัดส่วนการกระจายคะแนนไปยังช่วง 4.0 ดาว (15.5%), 3.0 ดาว (6.0%) และ 1.0 ดาว (5.8%) สูงกว่ากลุ่มที่ไม่ยืนยันการซื้อ   
 - สะท้อนว่าผู้ที่ซื้อสินค้าไปใช้งานจริง มีการประเมินคุณภาพสินค้าตามความเป็นจริง และพร้อมจะให้คะแนนระดับกลางถึงน้อยตามประสบการณ์ใช้งานจริง

**ลักษณะพฤติกรรมของการเขียนรีวิว (Review Behavior):**
- กลุ่ม Non-Verified: มักเป็นผู้ที่ตั้งใจเข้ามาเขียนรีวิวขนาดยาว เล่าเรื่องราวเชิงลึก วิเคราะห์รายละเอียดอัลบั้ม/เพลง หรือใส่ข้อคิดเห็นที่มีอารมณ์ร่วมสูง ซึ่งเนื้อหาที่มีความลึกและความละเอียดนี้คือปัจจัยหลักที่ดึงดูดให้ผู้อื่นกดโหวตว่าเป็นประโยชน์ (Helpful Vote)   
- กลุ่ม Verified Purchase: ผู้ซื้อจริงส่วนใหญ่มักให้คะแนนดาวและเขียนข้อความสั้นๆ ตรงประเด็น (เช่น "ส่งเร็ว เสียงดี") ซึ่งแม้จะสะท้อนความพึงพอใจจริง แต่ด้วยข้อความที่กระชับจึงไม่ค่อยได้รับ Helpful Vote จากผู้อื่น   

**นัยสำคัญทางธุรกิจ (Business Implications):ไม่ใช่จำนวนผู้ซื้อจริง แต่คือ "ความละเอียดของเนื้อหา"**:
- การสร้าง Engagement หรือความน่าเชื่อถือบนแพลตฟอร์มไม่ได้ขึ้นอยู่กับสถานะ Verified เพียงอย่างเดียว แต่เกิดจาก เนื้อหาที่มีความยาวและการเล่าเรื่องเชิงลึก (Long-form Storytelling)

### 2.4 เตรียมข้อความและสร้าง Text Features

ก่อนนำข้อความไปวิเคราะห์ จำเป็นต้องมีการ **preprocessing** เพื่อทำให้ข้อความอยู่ในรูปที่เหมาะสมกับเทคนิคที่จะใช้ อย่างไรก็ตาม ในงานนี้ไม่จำเป็นต้องทำ preprocessing ทุกขั้นตอน ให้เลือกทำ **เท่าที่จำเป็น** ตามวัตถุประสงค์ของการวิเคราะห์

ตัวอย่างเช่น การเปลี่ยนข้อความเป็นตัวพิมพ์เล็ก (lower case) การนำ URL หรือ HTML ออก การแบ่งข้อความเป็นคำ (tokenization) หรือการตัด stopwords เช่น `the`, `and`, `is` ออก ในกรณีที่คำเหล่านี้ไม่ได้ช่วยในการวิเคราะห์

**ไม่บังคับให้ใช้ stemming หรือ lemmatization** หากไม่มีเหตุผลว่าขั้นตอนดังกล่าวจะช่วยตอบคำถามของทีมได้ดีขึ้น

หลังจากเตรียมข้อความแล้ว ให้สร้าง **Text Features อย่างน้อย 2 ชนิด** เพื่อนำข้อความมาเปลี่ยนให้อยู่ในรูปข้อมูลที่สามารถนำไปสำรวจ เปรียบเทียบ หรือวิเคราะห์ต่อได้

---

#### Text Feature คืออะไร?

Text Feature คือ **ข้อมูลเชิงตัวเลขที่สกัดหรือสร้างขึ้นจากข้อความ** เพื่อแทนลักษณะหรือความหมายบางอย่างของข้อความนั้น

Text Feature อาจอยู่ในรูปของ

* **ค่าตัวเลขค่าเดียว** เช่น จำนวนคำ หรือ sentiment score
* **ตัวแปรหลายตัว** เช่น `price=1`, `quality=0`, `delivery=1`
* **เวกเตอร์หลายมิติ** เช่น TF-IDF vector หรือ Text Embedding

ตัวอย่างเช่น รีวิว

> `"The product is easy to use but the delivery was very slow."`

สามารถเปลี่ยนเป็น feature ได้หลายรูปแบบ

| Feature                    |                 ตัวอย่างค่าที่ได้ |
| -------------------------- | --------------------------------: |
| จำนวนคำในรีวิว             |                                12 |
| Sentiment score            |                             -0.15 |
| กล่าวถึง usability หรือไม่ |                                 1 |
| กล่าวถึง delivery หรือไม่  |                                 1 |
| กล่าวถึง price หรือไม่     |                                 0 |
| TF-IDF vector              |   `[0.00, 0.41, 0.00, 0.72, ...]` |
| Embedding vector           | `[0.21, -0.08, 0.63, ..., -0.14]` |

ดังนั้นข้อความหนึ่งข้อความสามารถถูกแทนได้ทั้งด้วย **feature ที่ตีความได้โดยตรง** และ **feature แบบเวกเตอร์ที่ใช้แทนเนื้อหาหรือความหมายของข้อความ**

---

## ตัวอย่าง Text Features ที่สามารถเลือกใช้

ผมจะแก้ตารางเดิมเป็นแบบนี้:

| Text Feature          | คืออะไร                                     | ตัวอย่าง                     | ใช้ตอบคำถามอะไรได้                                    |
| --------------------- | ------------------------------------------- | ---------------------------- | ----------------------------------------------------- |
| **Review Length**     | จำนวนคำหรือตัวอักษร                         | 120 words                    | รีวิวที่ยาวกว่าได้รับ helpful vote มากกว่าหรือไม่     |
| **Sentiment Score**   | คะแนนความเป็นบวก/ลบ                         | `-0.72`, `+0.85`             | ความรู้สึกลูกค้าสัมพันธ์กับ rating หรือไม่            |
| **Aspect Indicators** | ระบุว่าพูดถึงหัวข้อใด                       | `delivery=1`, `price=0`      | ลูกค้าบ่นเรื่องใดมากที่สุด                            |
| **Lexical Diversity** | ความหลากหลายของคำ                           | unique words / total words   | รีวิวที่เขียนละเอียดมีประโยชน์มากกว่าหรือไม่          |
| **TF-IDF Vector**     | เวกเตอร์ที่ให้น้ำหนักคำเด่นในเอกสาร         | `[0, .35, 0, .81, ...]`      | คำใดแยกกลุ่มรีวิวบวกกับลบ                             |
| **Text Embedding**    | dense vector ที่แทนความหมายโดยรวมของข้อความ | `[.12, -.44, .08, ...]`      | รีวิวใดมีความหมายคล้ายกัน หรือมี customer themes อะไร |
| **Custom Feature**    | feature ที่ทีมสร้างตามโจทย์                 | `refund=1`, `defect_count=3` | ตอบ business question เฉพาะของทีม                     |

---

### ตัวอย่าง Embedding แบบง่าย

ถ้าสอน `sentence-transformers` มาแล้ว อาจใส่ตัวอย่างสั้น ๆ:

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

texts = [
```

**รายงานเพียง 3 อย่าง**
1. ทำ preprocessing อะไร และทำไม
2. สร้าง feature อะไรอย่างน้อย 2 ตัว
3. feature เหล่านั้นช่วยตอบคำถามทางธุรกิจอย่างไร

> ไม่ต้องรายงาน vocabulary ก่อน–หลัง เว้นแต่ประเด็นนั้นเกี่ยวข้องกับผลวิเคราะห์

````python
import html
import re
import numpy as np
import pandas as pd

# ==========================================
# 1. Text Preprocessing (เร่งสปีดด้วย List & Pre-compiled Regex)
# ==========================================

# Pre-compile Regex ไว้ล่วงหน้าเพื่อไม่ต้องสร้างใหม่ทุกรอบ
url_re = re.compile(r'https?://\S+|www\.\S+')
html_re = re.compile(r'<.*?>')
space_re = re.compile(r'\s+')

# แปลงเป็น List ล่วงหน้า (เร็วกว่า Pandas .apply / .fillna มาก)
titles = df_joined['title_review'].fillna('').astype(str).tolist()
texts = df_joined['text'].fillna('').astype(str).tolist()


def clean_text_fast(title, text):
  combined = f'{title} {text}'
  combined = html.unescape(combined)
  combined = url_re.sub('', combined)
  combined = html_re.sub('', combined)
  combined = space_re.sub(' ', combined)
  return combined.strip().lower()


# รวมข้อความและคลีนในขั้นตอนเดียว
cleaned_list = [clean_text_fast(t, x) for t, x in zip(titles, texts)]
df_joined['cleaned_text'] = cleaned_list


# ==========================================
# 2. Create Rule-based & Style Features
# ==========================================

# 2.1 Review Length (นับคำผ่าน Python List)
df_joined['review_length'] = [len(x.split()) for x in cleaned_list]

# 2.2 Aspect Indicators (ใช้ Non-capturing group (?:...) + Compiled Regex)
sound_pattern = re.compile(
    r'\b(?:sound|audio|quality|recording|stereo|album|song|track|fidelity|remaster|remastered)\b'
)
nostalgia_pattern = re.compile(
    r'\b(?:childhood|memories|memory|classic|timeless|youth|old'
    r' school|90s|80s|nostalgia|nostalgic)\b'
)
value_pattern = re.compile(
    r'\b(?:price|worth|value|cheap|deal|bargain|penny|affordable|cost|money|buy)\b'
)

df_joined['mention_sound'] = np.fromiter(
    (1 if sound_pattern.search(x) else 0 for x in cleaned_list), dtype=np.int8
)
df_joined['mention_nostalgia'] = np.fromiter(
    (1 if nostalgia_pattern.search(x) else 0 for x in cleaned_list),
    dtype=np.int8,
)
df_joined['mention_value'] = np.fromiter(
    (1 if value_pattern.search(x) else 0 for x in cleaned_list), dtype=np.int8
)


# ==========================================
# แสดงผลลัพธ์ Features ใหม่
# ==========================================
display(
    df_joined[[
        'title_review',
        'cleaned_text',
        'review_length',
        'mention_sound',
        'mention_nostalgia',
        'mention_value',
    ]].head()
)
````

##### **ทำอะไรบ้าง (What it does):**
* **Combine Text:** รวมข้อความจาก `title_review` และ `text` เข้าด้วยกันเป็นข้อความเดียวเพื่อเก็บบริบทการรีวิวอย่างสมบูรณ์

* **Remove URLs & HTML Tags:** ลบลิงก์เว็บไซต์ (`http://`, `www.`) และแท็ก HTML (เช่น `<br>`, `<b>`) ออกด้วย Regular Expressions (Regex)

##### **ทำทำไม (Why do it):**
* **ลด Noise (สิ่งรบกวนข้อมูล):** แท็ก HTML หรือ URLs ไม่ได้ให้ความหมายทางอารมณ์หรือคุณค่าเชิงธุรกิจ แต่จะสร้างความสับสนให้ระบบวิเคราะห์
* **สร้าง Standard Data Format:** ปรับรูปแบบข้อความให้เป็นมาตรฐานเดียวกัน เช่น *"Sound"*, *"SOUND"*, *"sound"* หากไม่แปลงเป็นตัวพิมพ์เล็ก ระบบจะมองเป็นคนละคำกัน ส่งผลให้การนับความถี่คีย์เวิร์ดคลาดเคลื่อน
* **เพิ่มความเร็วและแม่นยำในการประมวลผล:** การคลีนข้อความล่วงหน้าและจัดเก็บเป็น Python List ช่วยให้กระบวนการจับรูปแบบวลี (Regex Matching) และการสแกนคำทำงานได้เร็วขึ้นอย่างมหาศาล

## 2.6 Reality Check

ก่อนเอา insight ไปบอกลูกค้า ให้ลองหักล้างมันด้วยตัวเองก่อน อย่างน้อย 1 insight ต้องผ่านขั้นนี้

คำอธิบายทางเลือกที่พบบ่อย
* **ผลของขนาด:** กลุ่มที่ดูโดดเด่นอาจมีตัวอย่างน้อยมาก ตรวจ n ของทุกกลุ่ม
* **ผลของส่วนผสม:** ความต่างอาจมาจากสัดส่วนสินค้าในแต่ละกลุ่มไม่เหมือนกัน ไม่ใช่จากพฤติกรรมลูกค้า ลองคุมให้เทียบกันเฉพาะสินค้าประเภทเดียวกัน
* **ผลของเวลา:** ของเก่ามีเวลาสะสม vote และรีวิวมากกว่า
* **ผลของความยาว:** ข้อความยาวมีโอกาสมีคำที่เราค้นหามากกว่าโดยอัตโนมัติ
* **ผลของคนไม่กี่คน:** ผู้ใช้หรือสินค้าไม่กี่รายอาจครองสัดส่วนใหญ่ของกลุ่ม

เลือกอย่างน้อยหนึ่งคำอธิบาย แล้ว **ทดสอบด้วยตัวเลขจริง** สรุปว่า insight ยังยืนอยู่ได้ หรือต้องปรับถ้อยคำให้แคบลง

1. Helpful Vote เกิดจาก "ความยาวและการเล่าเรื่อง" (Storytelling) ไม่ใช่แค่การแตะคีย์เวิร์ด: การทำ Reality Check พิสูจน์แล้วว่า เมื่อคุมความยาวคำ รีวิวกลุ่ม High Impact ($\ge 5$ votes) มีอัตราการพูดถึง Sound Quality หรือ Nostalgia พอๆ กับรีวิวทั่วไป สิ่งที่สร้าง Engagement จริงๆ คือระดับความละเอียดของเนื้อหาปรับกลยุทธ์ Ad Copywriting เน้น "Deep Value Content":

2. การทำโฆษณาหรือสื่อสารการตลาดไม่ควรอัดแค่คำคีย์เวิร์ดสั้นๆ แต่ควรเน้นการเล่าเรื่องเชิงลึก (Long-form Storytelling) ที่สร้างอารมณ์ร่วมและให้ข้อมูลครบถ้วน เพื่อเปลี่ยนผู้ฟังทั่วไปให้เป็นผู้สนับสนุนแบรนด์

3. การคุมตัวแปรแฝง (Control Variables) ช่วยป้องกันข้อสรุปที่ผิดพลาด: กระบวนการหักล้างด้วย Reality Check ช่วยกรองความสัมพันธ์เทียม (Spurious Correlation) ออกไป ทำให้ได้ Insight ที่แม่นยำและปลอดภัยที่สุดก่อนนำไปวางแผนใช้งบประมาณจริงกับ Client

> **Insight Card 1**

> **Insight:** Helpful Vote ไม่ได้เกิดจากการอัดคำคีย์เวิร์ดเฉพาะ แต่เกิดจาก **"ความละเอียดและการเล่าเรื่องเชิงลึก" (Long-form Storytelling)**
>
> **หลักฐาน:** เมื่อควบคุมตัวแปรความยาวข้อความ (Controlled Bins) อัตราการเอ่ยถึง Sound Quality และ Nostalgia ในกลุ่ม Medium และ Long ของกลุ่ม High Impact ($\ge 5$ votes) และกลุ่มทั่วไปต่างกันเพียง **0.2% – 4.0%** เท่านั้น (จากกลุ่มตัวอย่างทั้งหมด $n = 19,962$)
>
>
> **So what:** การทำโฆษณาที่ยัดแค่คีย์เวิร์ดหรือสโลแกนสั้นๆ ไม่ช่วยสร้าง Engagement สิ่งที่ดึงดูดผู้ฟังได้จริงคือรายละเอียดและบริบทการใช้งานที่ครบถ้วน
>
>
> **Action:** ปรับ Ad Copywriting จาก Short Tagline เป็น **Long-form Narrative Copy** (เช่น Carousel Ads หรือ Video Reviews) ที่บรรยายฟีลลิ่ง ความรู้สึกในการฟัง และความทรงจำเกี่ยวกับแผ่นเสียง
>
>

> **หลักฐาน:** กลุ่ม *K-Pop / Asian Pop* ให้ความสำคัญกับ Sound Quality สูงที่สุดถึง **39.6%** แต่ธีม Nostalgia แทบเป็น 0% (**0.3%**) ขณะที่กลุ่ม *Classic Rock / Oldies* ตอบรับกับธีม Nostalgia สูงสุด (**9.7%**) ควบคู่กับ Sound Quality (**26.0%**) และกลุ่ม *Jazz / Classical* โดดเด่นในมิติ Nostalgia (**4.7%**) โดยวลีติดปากจริงของผู้ฟัง เช่น *"sound quality"* (224 ครั้ง) และ *"great sound"* (99 ครั้ง) ปรากฏถี่ที่สุดในรีวิวเชิงบวก
>
>
> **So what:** โฆษณาแบบข้อความเดียวใช้กับทุกกลุ่ม (One-size-fits-all) ไม่ได้ผล ต้องทำการสื่อสารเฉพาะกลุ่ม (Segmented Messaging) ให้ตรงจุดเน้นของแต่ละแนวเพลง

````python
import re
import matplotlib.gridspec as gridspec
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# ==========================================
# 1. SETUP LIGHT THEME & PALETTE
# ==========================================
BG_COLOR = '#FFFFFF'
CARD_BG = '#F8FAFC'
BORDER_COLOR = '#E2E8F0'
TEXT_DARK = '#0F172A'
TEXT_MUTED = '#64748B'

GREEN = '#1DB954'  # High Impact (>=5 votes)
GRAY = '#94A3B8'  # General (<5 votes)
BLUE = '#2563EB'  # Nostalgia
AMBER = '#D97706'  # Value for Money

plt.rcParams.update({
    'font.size': 9.5,
    'figure.facecolor': BG_COLOR,
    'axes.facecolor': CARD_BG,
    'axes.edgecolor': BORDER_COLOR,
    'axes.linewidth': 1.2,
    'grid.color': '#CBD5E1',
    'grid.linestyle': '--',
    'grid.alpha': 0.5,
})

# ==========================================
# 2. COMPUTING REAL METRICS FROM df_joined
# ==========================================

# 2.1 เตรียมข้อมูลข้อความ และคำนวณ Word Count
if 'cleaned_text' not in df_joined.columns:
  df_joined['cleaned_text'] = (
      df_joined['title_review'].fillna('').astype(str)
      + ' '
      + df_joined['text'].fillna('').astype(str)
  )

df_joined['word_count'] = (
    df_joined['cleaned_text'].astype(str).str.split().str.len()
)


def categorize_length(wc):
  if wc < 30:
    return 'Short (<30w)'
  elif wc <= 100:
    return 'Medium (30-100w)'
  else:
    return 'Long (>100w)'


df_joined['length_bin'] = df_joined['word_count'].apply(categorize_length)

# 2.2 ค้นหาคีย์เวิร์ด Mention
sound_re = re.compile(
    r'\b(?:sound|audio|quality|recording|stereo|album|song|track|fidelity|remaster|remastered)\b'
)
nostalgia_re = re.compile(
    r'\b(?:childhood|memories|memory|classic|timeless|youth|old'
    r' school|90s|80s|nostalgia|nostalgic)\b'
)
value_re = re.compile(
    r'\b(?:price|worth|value|cheap|deal|bargain|penny|affordable|cost|money|buy|photobook)\b'
)

df_joined['mention_sound'] = (
    df_joined['cleaned_text']
    .str.contains(sound_re, regex=True, na=False)
    .astype(int)
)
df_joined['mention_nostalgia'] = (
    df_joined['cleaned_text']
    .str.contains(nostalgia_re, regex=True, na=False)
    .astype(int)
)
df_joined['mention_value'] = (
    df_joined['cleaned_text']
    .str.contains(value_re, regex=True, na=False)
    .astype(int)
)

# 2.3 ตรวจสอบ Helpful Vote
vote_col = next(
    (
        c
        for c in ['helpful_votes', 'votes', 'helpful', 'helpful_vote']
        if c in df_joined.columns
    ),
    None,
)
if vote_col:
  df_joined['is_high_impact'] = df_joined[vote_col] >= 5
else:
  df_joined['is_high_impact'] = False

# 2.4 คำนวณสถิติแบ่งตาม Length Bins และ Impact Group
length_order = ['Short (<30w)', 'Medium (30-100w)', 'Long (>100w)']

sound_stats = (
    df_joined.groupby(['length_bin', 'is_high_impact'])['mention_sound']
    .mean()
    .unstack(fill_value=0)
    * 100
)
nost_stats = (
    df_joined.groupby(['length_bin', 'is_high_impact'])['mention_nostalgia']
    .mean()
    .unstack(fill_value=0)
    * 100
)

gen_sound = [
    sound_stats.loc[b, False] if b in sound_stats.index else 0
    for b in length_order
]
hi_sound = [
    sound_stats.loc[b, True] if b in sound_stats.index else 0
    for b in length_order
]

gen_nost = [
    nost_stats.loc[b, False] if b in nost_stats.index else 0
    for b in length_order
]
hi_nost = [
    nost_stats.loc[b, True] if b in nost_stats.index else 0 for b in length_order
]

# คำนวณส่วนต่าง (Delta) เพื่อยืนยันหลักฐาน Reality Check
medium_diff = abs(
    (
        sound_stats.loc['Medium (30-100w)', True]
        if 'Medium (30-100w)' in sound_stats.index
        else 0
    )
    - (
        sound_stats.loc['Medium (30-100w)', False]
        if 'Medium (30-100w)' in sound_stats.index
        else 0
    )
)
long_diff = abs(
    (
        sound_stats.loc['Long (>100w)', True]
        if 'Long (>100w)' in sound_stats.index
        else 0
    )
    - (
        sound_stats.loc['Long (>100w)', False]
        if 'Long (>100w)' in sound_stats.index
        else 0
    )
)

total_n = len(df_joined)

# ==========================================
# 3. BUILD GRAPHICAL INSIGHT CARD 1
# ==========================================
fig = plt.figure(figsize=(16, 9.5), dpi=150)
gs = gridspec.GridSpec(2, 2, width_ratios=[1.15, 1], height_ratios=[1.1, 0.9])

# --- Header Section ---
fig.text(
    0.04,
    0.95,
    'INSIGHT CARD 1: LONG-FORM STORYTELLING OVER KEYWORD STUFFING',
    fontsize=15,
    fontweight='bold',
    color=TEXT_DARK,
)
fig.text(
    0.04,
    0.92,
    f'Confidence Level: HIGH (Controlled Length Bias Reality Check | N = {total_n:,} real reviews)',
    fontsize=10,
    color=TEXT_MUTED,
)

# ------------------------------------------
# SUBPLOT 1: REALITY CHECK - SOUND QUALITY RATE (LEFT)
# ------------------------------------------
ax1 = fig.add_subplot(gs[:, 0])

x = np.arange(len(length_order))
width = 0.32

r1 = ax1.bar(
    x - width / 2, gen_sound, width, label='General (<5 votes)', color=GRAY
)
r2 = ax1.bar(
    x + width / 2,
    hi_sound,
    width,
    label='High Impact (>=5 votes)',
    color=GREEN,
)

ax1.set_title(
    'A. Sound Quality Mention Rate by Word Count Bin (%)',
    fontsize=11.5,
    fontweight='bold',
    color=TEXT_DARK,
    pad=12,
    loc='left',
)
ax1.set_ylabel('Mention Rate (%)', fontsize=9.5, color=TEXT_MUTED)
ax1.set_xticks(x)
ax1.set_xticklabels(length_order, fontsize=9.5, color=TEXT_DARK)
ax1.set_ylim(0, max(max(gen_sound + hi_sound + [50]) * 1.2, 100))
ax1.grid(True, axis='y')
ax1.legend(
    frameon=True,
    facecolor='#FFFFFF',
    edgecolor=BORDER_COLOR,
    fontsize=9,
    labelcolor=TEXT_DARK,
    loc='upper left',
)

for bar in r1 + r2:
  h = bar.get_height()
  ax1.annotate(
      f'{h:.1f}%',
      (bar.get_x() + bar.get_width() / 2.0, h + 1.5),
      ha='center',
      va='bottom',
      fontsize=8.5,
      fontweight='bold',
      color=TEXT_DARK,
  )

# ------------------------------------------
# SUBPLOT 2: REALITY CHECK - NOSTALGIA RATE (RIGHT TOP)
# ------------------------------------------
ax2 = fig.add_subplot(gs[0, 1])

r3 = ax2.bar(
    x - width / 2, gen_nost, width, label='General (<5 votes)', color=GRAY
)
r4 = ax2.bar(
    x + width / 2,
    hi_nost,
    width,
    label='High Impact (>=5 votes)',
    color=BLUE,
)

ax2.set_title(
    'B. Nostalgia Mention Rate Across Controlled Length Bins (%)',
    fontsize=11.5,
    fontweight='bold',
    color=TEXT_DARK,
    pad=12,
    loc='left',
)
ax2.set_ylabel('Mention Rate (%)', fontsize=9.5, color=TEXT_MUTED)
ax2.set_xticks(x)
ax2.set_xticklabels(length_order, fontsize=9.5, color=TEXT_DARK)
ax2.set_ylim(0, max(max(gen_nost + hi_nost + [20]) * 1.3, 30))
ax2.grid(True, axis='y')
ax2.legend(
    frameon=True,
    facecolor='#FFFFFF',
    edgecolor=BORDER_COLOR,
    fontsize=9,
    labelcolor=TEXT_DARK,
    loc='upper left',
)

for bar in r3 + r4:
  h = bar.get_height()
  ax2.annotate(
      f'{h:.1f}%',
      (bar.get_x() + bar.get_width() / 2.0, h + 0.8),
      ha='center',
      va='bottom',
      fontsize=8.5,
      fontweight='bold',
      color=TEXT_DARK,
  )

# ------------------------------------------
# SUBPLOT 3: EXECUTIVE ACTION PLAN & SO WHAT (RIGHT BOTTOM)
# ------------------------------------------
ax3 = fig.add_subplot(gs[1, 1])
ax3.axis('off')

action_box_text = f"""C. Executive Playbook & Strategic Action Plan:

• Key Finding (Evidence):
  When controlling for length, the keyword rate difference between High
  Impact (>=5 votes) and General reviews shrinks to just {medium_diff:.1f}% - {long_diff:.1f}%
  in Medium & Long bins. Helpful votes stem from story depth, not keywords.

• So What:
  Short, keyword-stuffed taglines fail to drive user engagement. True connection
  and decision confidence come from rich context and narrative experience.

• Strategic Action Plan:
  1. Ad Copywriting Shift: Transition from Short Taglines to Long-form
     Narrative Copy (e.g., Carousel Ads, Detailed Video Reviews).
  2. Content Focus: Describe listening feelings, acoustic warmth, and memories.

• Measurement & KPIs:
  Ad Engagement Rate, Average Time on Page, and Carousel Click-Through Rate (CTR)."""

ax3.text(
    0.0,
    1.0,
    action_box_text,
    transform=ax3.transAxes,
    fontsize=9.2,
    color=TEXT_DARK,
    va='top',
    ha='left',
    bbox=dict(
        boxstyle='round,pad=0.8',
        facecolor=CARD_BG,
        edgecolor=BORDER_COLOR,
        linewidth=1.2,
    ),
)

plt.tight_layout()
plt.subplots_adjust(top=0.88)
plt.savefig(
    'insight_card_1_real_data.png',
    dpi=300,
    bbox_inches='tight',
    facecolor=BG_COLOR,
)
plt.show()
````

````python
import re
import matplotlib.gridspec as gridspec
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# ==========================================
# 1. SETUP LIGHT THEME & PALETTE
# ==========================================
BG_COLOR = '#FFFFFF'
CARD_BG = '#F8FAFC'
BORDER_COLOR = '#E2E8F0'
TEXT_DARK = '#0F172A'
TEXT_MUTED = '#64748B'

GREEN = '#1DB954'  # Sound Quality
BLUE = '#2563EB'  # Nostalgia
AMBER = '#D97706'  # Value for Money

plt.rcParams.update({
    'font.size': 9.5,
    'figure.facecolor': BG_COLOR,
    'axes.facecolor': CARD_BG,
    'axes.edgecolor': BORDER_COLOR,
    'axes.linewidth': 1.2,
    'grid.color': '#CBD5E1',
    'grid.linestyle': '--',
    'grid.alpha': 0.5,
})

# ==========================================
# 2. COMPUTING REAL METRICS FROM df_joined
# ==========================================

# 2.1 เตรียมข้อมูลข้อความ (Cleaned Text)
if 'cleaned_text' not in df_joined.columns:
  df_joined['cleaned_text'] = (
      df_joined['title_review'].fillna('').astype(str)
      + ' '
      + df_joined['text'].fillna('').astype(str)
  )

# 2.2 ค้นหาคีย์เวิร์ด Mention ทั้ง 3 หมวดหมู่จากข้อมูลจริง
sound_re = re.compile(
    r'\b(?:sound|audio|quality|recording|stereo|album|song|track|fidelity|remaster|remastered)\b'
)
nostalgia_re = re.compile(
    r'\b(?:childhood|memories|memory|classic|timeless|youth|old'
    r' school|90s|80s|nostalgia|nostalgic)\b'
)
value_re = re.compile(
    r'\b(?:price|worth|value|cheap|deal|bargain|penny|affordable|cost|money|buy|photobook)\b'
)

df_joined['mention_sound'] = (
    df_joined['cleaned_text']
    .str.contains(sound_re, regex=True, na=False)
    .astype(int)
)
df_joined['mention_nostalgia'] = (
    df_joined['cleaned_text']
    .str.contains(nostalgia_re, regex=True, na=False)
    .astype(int)
)
df_joined['mention_value'] = (
    df_joined['cleaned_text']
    .str.contains(value_re, regex=True, na=False)
    .astype(int)
)

# 2.3 คำนวณสถิติแยกตาม Genre จริงใน df_joined
genre_col = next(
    (
        c
        for c in [
            'inferred_genre',
            'genre',
            'category',
            'music_genre',
            'store',
        ]
        if c in df_joined.columns
    ),
    None,
)

if genre_col:
  # ดึงประเภท Genre ที่มีอยู่จริงในข้อมูล
  genre_df = (
      df_joined.groupby(genre_col)[
          ['mention_sound', 'mention_nostalgia', 'mention_value']
      ]
      .mean()
      * 100
  )
  # เลือกกลุ่มที่มีจำนวนรีวิวสูงสุด 5 อันดับแรกเพื่อความสมบูรณ์ของกราฟ
  top_genres = df_joined[genre_col].value_counts().head(5).index.tolist()
  genre_df = genre_df.loc[top_genres]
else:
  raise ValueError("ไม่พบคอลัมน์ระบุ Genre ใน df_joined")

# 2.4 คำนวณความถี่ VoC จริงจากรีวิวเชิงบวก (Rating >= 4)
rate_col = next(
    (c for c in ['rating', 'overall', 'stars'] if c in df_joined.columns), None
)
if rate_col:
  pos_reviews = df_joined[df_joined[rate_col] >= 4]['cleaned_text']
else:
  pos_reviews = df_joined['cleaned_text']

voc_dictionary = {
    # Sound Quality
    'sound quality': 'Sound Quality',
    'great sound': 'Sound Quality',
    'excellent sound': 'Sound Quality',
    'remastered sound': 'Sound Quality',
    # Nostalgia
    'takes me back': 'Nostalgia',
    'back in the day': 'Nostalgia',
    'good old days': 'Nostalgia',
    # Value for Money
    'worth every penny': 'Value for Money',
    'great value': 'Value for Money',
    'worth the money': 'Value for Money',
}

voc_data = []
for phrase, cat in voc_dictionary.items():
  pat = r'\b' + re.escape(phrase) + r'\b'
  cnt = pos_reviews.str.contains(pat, regex=True, na=False).sum()
  voc_data.append({'Phrase': phrase, 'Frequency': cnt, 'Category': cat})

voc_df = (
    pd.DataFrame(voc_data)
    .sort_values(by='Frequency', ascending=True)
    .reset_index(drop=True)
)

# ==========================================
# 3. BUILD GRAPHICAL INSIGHT CARD (REAL DATA)
# ==========================================
fig = plt.figure(figsize=(16, 9.5), dpi=150)
gs = gridspec.GridSpec(2, 2, width_ratios=[1.15, 1], height_ratios=[1.1, 0.9])

total_n = len(df_joined)

# --- Header Section ---
fig.text(
    0.04,
    0.95,
    'INSIGHT CARD 2: STRATEGIC GENRE SEGMENTATION & VOC SOCIAL PROOF',
    fontsize=15,
    fontweight='bold',
    color=TEXT_DARK,
)
fig.text(
    0.04,
    0.92,
    f'Confidence Level: HIGH (Calculated directly from N = {total_n:,} real reviews)',
    fontsize=10,
    color=TEXT_MUTED,
)

# ------------------------------------------
# SUBPLOT 1: GENRE MOTIVATION DRIVERS (REAL DATA)
# ------------------------------------------
ax1 = fig.add_subplot(gs[:, 0])

genres = genre_df.index.tolist()
y = np.arange(len(genres))
height = 0.25

r1 = ax1.barh(
    y - height,
    genre_df['mention_sound'],
    height,
    label='Sound Quality',
    color=GREEN,
)
r2 = ax1.barh(
    y, genre_df['mention_nostalgia'], height, label='Nostalgia', color=BLUE
)
r3 = ax1.barh(
    y + height,
    genre_df['mention_value'],
    height,
    label='Value for Money',
    color=AMBER,
)

ax1.set_title(
    'A. Real Purchase Motivation Drivers by Genre (%)',
    fontsize=11.5,
    fontweight='bold',
    color=TEXT_DARK,
    pad=12,
    loc='left',
)
ax1.set_yticks(y)
ax1.set_yticklabels(genres, fontsize=9.5, fontweight='bold', color=TEXT_DARK)
ax1.set_xlabel('Mention Rate (%)', fontsize=9.5, color=TEXT_MUTED)
ax1.set_xlim(0, max(genre_df.values.max() * 1.25, 30))
ax1.grid(True, axis='x')
ax1.legend(
    frameon=True,
    facecolor='#FFFFFF',
    edgecolor=BORDER_COLOR,
    fontsize=9,
    labelcolor=TEXT_DARK,
    loc='lower right',
)

# Value Labels จากข้อมูลจริง
for bars in [r1, r2, r3]:
  for bar in bars:
    w = bar.get_width()
    ax1.annotate(
        f'{w:.1f}%',
        (w + 0.5, bar.get_y() + bar.get_height() / 2.0),
        ha='left',
        va='center',
        fontsize=8.5,
        fontweight='bold',
        color=TEXT_DARK,
    )

# ------------------------------------------
# SUBPLOT 2: TOP VOC PHRASES (REAL DATA)
# ------------------------------------------
ax2 = fig.add_subplot(gs[0, 1])

category_color_map = {
    'Sound Quality': GREEN,
    'Nostalgia': BLUE,
    'Value for Money': AMBER,
}
bar_colors = [category_color_map[cat] for cat in voc_df['Category']]

r4 = ax2.barh(
    voc_df['Phrase'],
    voc_df['Frequency'],
    color=bar_colors,
    height=0.55,
    edgecolor='none',
)

ax2.set_title(
    'B. Real VoC Phrase Counts in Positive Reviews',
    fontsize=11.5,
    fontweight='bold',
    color=TEXT_DARK,
    pad=12,
    loc='left',
)
ax2.set_xlabel('Actual Frequency Count', fontsize=9.5, color=TEXT_MUTED)
ax2.set_xlim(0, max(voc_df['Frequency'].max() * 1.25, 10))
ax2.grid(True, axis='x')

for bar in r4:
  w = bar.get_width()
  ax2.annotate(
      f'{int(w):,}',
      (w + (ax2.get_xlim()[1] * 0.02), bar.get_y() + bar.get_height() / 2.0),
      ha='left',
      va='center',
      fontsize=8.5,
      fontweight='bold',
      color=TEXT_DARK,
  )


# ------------------------------------------
# SUBPLOT 3: ACTION PLAN FORMED FROM REAL METRICS
# ------------------------------------------
def get_val(g_name, col_name):
  if g_name in genre_df.index:
    return f'{genre_df.loc[g_name, col_name]:.1f}%'
  return 'N/A'


ax3 = fig.add_subplot(gs[1, 1])
ax3.axis('off')

# ดึงชื่อ Genre จริงที่ได้คะแนนสูงสุดในแต่ละด้าน
top_sound_genre = genre_df['mention_sound'].idxmax()
top_nost_genre = genre_df['mention_nostalgia'].idxmax()
top_value_genre = genre_df['mention_value'].idxmax()

dynamic_action_text = f"""C. Data-Driven Strategic Playbook (Formed from df_joined):

• Sound Quality Dominance:
  - Highest in [{top_sound_genre}] ({get_val(top_sound_genre, 'mention_sound')})
  - Campaign Hook: Focus heavily on audio fidelity & remastering details.

• Nostalgia Driver Peak:
  - Highest in [{top_nost_genre}] ({get_val(top_nost_genre, 'mention_nostalgia')})
  - Campaign Hook: Leverage emotional storytelling & "Takes you back" messaging.

• Value for Money Focus:
  - Highest in [{top_value_genre}] ({get_val(top_value_genre, 'mention_value')})
  - Campaign Hook: Emphasize physical collectibles & cost-per-play efficiency.

• Execution: Embed top actual VoC phrases directly into Landing Page Headlines
  and Ad Review Quotes to boost Click-Through Rate (CTR) and ROAS."""

ax3.text(
    0.0,
    1.0,
    dynamic_action_text,
    transform=ax3.transAxes,
    fontsize=9.2,
    color=TEXT_DARK,
    va='top',
    ha='left',
    bbox=dict(
        boxstyle='round,pad=0.8',
        facecolor=CARD_BG,
        edgecolor=BORDER_COLOR,
        linewidth=1.2,
    ),
)

plt.tight_layout()
plt.subplots_adjust(top=0.88)
plt.savefig(
    'insight_card_2_real_data.png',
    dpi=300,
    bbox_inches='tight',
    facecolor=BG_COLOR,
)
plt.show()
````

### ตัวอย่างที่ 4: Article Collection Robot — อธิบายทีละขั้น

ตัวอย่างนี้ใช้เว็บไซต์ข่าวเพื่อให้เห็นแนวคิดของ Web Robot ชัดเจน

**เป้าหมาย:** ต้องการสร้างชุดข้อมูลที่

> **1 row = 1 บทความข่าว**

ดังนั้น Robot จะไม่ crawl ทุก link ในเว็บไซต์ เพราะจะได้หน้าเมนู หน้า category หน้า search หรือ URL ซ้ำ ๆ ปะปนมา

````python
# Part 3 — Web Robot config (one source/one method)
import re, time, requests, pandas as pd
from bs4 import BeautifulSoup
from urllib.parse import urljoin, urlparse, urldefrag
from urllib.robotparser import RobotFileParser
BASE_URL = 'https://www.thaipbs.or.th'
LISTING_URL = BASE_URL + '/news/politics'
ARTICLE_PATH_PATTERN = r'^/news/content/\d+$'
MIN_VALID_RECORDS = 200
MAX_LISTING_PAGES = 40
MAX_ARCHIVE_DAYS = 60      # permitted day-indexed news/archive pages
MAX_ARTICLES = 350          # เก็บเผื่อการตัด invalid / duplicate
DELAY = 1.5                # seconds between requests
BOT_USER_AGENT = 'KKU-TextAnalytics-StudentBot/1.0 (education; respects robots.txt)'
session = requests.Session()
session.headers.update({'User-Agent': BOT_USER_AGENT})
RANDOM_SEED = 42
print('Source:', LISTING_URL, '| required valid records:', MIN_VALID_RECORDS)
````

#### Step 2 — ทำความสะอาด URL และตรวจ `robots.txt`

URL เดียวกันอาจเขียนได้หลายแบบ เช่น

`https://www.thaipbs.or.th/news`  
`https://www.thaipbs.or.th/news#main-content`

ส่วนหลัง `#...` เรียกว่า **fragment** และโดยทั่วไปไม่ได้หมายถึงหน้าใหม่ จึงควรตัดออกก่อนตรวจ URL ซ้ำ

จากนั้นตรวจ `robots.txt` ก่อนเข้าแต่ละส่วนของเว็บ ถ้าไม่อนุญาตให้ bot เข้า path นั้น ให้ **เปลี่ยนแหล่งข้อมูล** ไม่ใช่หาวิธีหลบ


````python
# Ethical / fail-closed robots checking: no bypassing access controls
_robots_cache = {}
def normalize_url(url):
    return urldefrag(url)[0].rstrip('/')
def robots_allowed(url, user_agent=BOT_USER_AGENT):
    parsed = urlparse(url)
    root = f'{parsed.scheme}://{parsed.netloc}'
    if parsed.scheme != 'https' or parsed.netloc != urlparse(BASE_URL).netloc:
        return False
    if root not in _robots_cache:
        try:
            response = session.get(root + '/robots.txt', timeout=20)
            if response.status_code != 200:
                _robots_cache[root] = None
            else:
                rp = RobotFileParser()
                rp.parse(response.text.splitlines())
                _robots_cache[root] = rp
        except requests.RequestException:
            _robots_cache[root] = None
    rp = _robots_cache[root]
    return bool(rp is not None and rp.can_fetch(user_agent,url))
print('Robots permission for listing:',robots_allowed(LISTING_URL))
if not robots_allowed(LISTING_URL):
    raise PermissionError('robots.txt unavailable/disallows this page: switch to an authorized source; do not bypass')
````

#### Step 3 — เข้า “หน้ารวมข่าว” เพื่อหาเฉพาะ Article URLs

ขั้นนี้ **ยังไม่เก็บเนื้อหาข่าว**

Robot เปิดหน้ารวมข่าว แล้วดู `<a href="...">` ทุก link จากนั้นกรองให้เหลือเฉพาะ URL ที่ตรงกับรูปแบบบทความ เช่น

`/news/content/509281`

การกรองนี้คือสิ่งที่ทำให้ Robot ต่างจาก generic crawler ที่วิ่งไปทุกหน้าใน domain


````python
# Step 3: targeted discovery from news/politics and its pagination (no generic crawl)
def find_article_links(listing_url):
    if not robots_allowed(listing_url):
        raise PermissionError('Disallowed listing URL: ' + listing_url)
    response = session.get(listing_url,timeout=25)
    response.raise_for_status()
    soup = BeautifulSoup(response.text,'html.parser')
    found = set()
    for a in soup.find_all('a',href=True):
        url = normalize_url(urljoin(listing_url,a['href']))
        p = urlparse(url)
        if p.netloc == urlparse(BASE_URL).netloc and re.fullmatch(ARTICLE_PATH_PATTERN,p.path):
            found.add(url)
    return found

article_urls, no_new_pages = set(), 0
for page_no in range(1,MAX_LISTING_PAGES+1):
    page_url = LISTING_URL if page_no == 1 else f'{LISTING_URL}?page={page_no}'
    try:
        page_links = find_article_links(page_url)
    except (requests.RequestException,PermissionError) as exc:
        print('Stopping listing:',type(exc).__name__,str(exc)[:120]); break
    fresh = page_links - article_urls
    article_urls.update(fresh)
    print('listing',page_no,'new links',len(fresh),'unique links',len(article_urls))
    no_new_pages = 0 if fresh else no_new_pages + 1
    if len(article_urls) >= MAX_ARTICLES or no_new_pages >= 3: break
    time.sleep(DELAY)
# If one landing page exposes too few links, use public *daily news archive* pages.
# This BROADENS the sampling frame to ALL news categories; reflect that in analysis.
from datetime import datetime, timezone, timedelta
archive_pages_visited = 0
if len(article_urls) < MAX_ARTICLES:
    for days_back in range(MAX_ARCHIVE_DAYS):
        day = (datetime.now(timezone.utc).date() - timedelta(days=days_back)).isoformat()
        archive_url = f'{BASE_URL}/news/archive/{day}'
        if not robots_allowed(archive_url):
            print('Archive not permitted or robots unavailable; stopping at', day)
            break
        try:
            discovered = find_article_links(archive_url)
        except requests.RequestException as exc:
            print('Archive request failed; stopping:', type(exc).__name__)
            break
        archive_pages_visited += 1
        article_urls.update(discovered)
        print('archive', day, 'unique candidates', len(article_urls))
        if len(article_urls) >= MAX_ARTICLES: break
        time.sleep(DELAY)
print('archive pages visited:',archive_pages_visited)
article_urls = sorted(article_urls)
print('Discovered distinct candidate URLs:',len(article_urls))
if len(article_urls)<MIN_VALID_RECORDS:
    print('WARNING: discovery below 200. Pagination may be dynamic; do NOT report completion until actual valid n>=200.')
````

#### Step 4 — เปิดบทความทีละหน้าแล้วดึง “เนื้อหาที่ต้องการ”

ตอนนี้เรามี URL ของบทความแล้ว จึงเปิดทีละข่าวและแยก field ที่จะใช้วิเคราะห์ เช่น

* `url`
* `title`
* `published_at`
* `text`
* `word_count`

หลักสำคัญคือ **พยายามดึงเฉพาะตัวบทความ** ไม่ใช่ menu/footer/ข่าวแนะนำทั้งหน้า

ตัวอย่างใช้ `<article>` ก่อน ถ้าเว็บไซต์ไม่มี tag นี้จึง fallback ไปที่ `<main>` และนำข้อความจาก `<p>` มาต่อกัน  
เว็บไซต์อื่นอาจต้องเปลี่ยน selector เช่น `.article-body`, `.post-content` หรือ selector เฉพาะของเว็บนั้น


````python
# Step 4: extract actual article body. Thai word count uses PyThaiNLP.
from pythainlp.tokenize import word_tokenize
from urllib.parse import urlparse
from hashlib import sha256

def extract_article(url):
    if not robots_allowed(url):
        return None
    r = session.get(url,timeout=25)
    r.raise_for_status()
    soup=BeautifulSoup(r.text,'html.parser')
    h1=soup.find('h1')
    title=h1.get_text(' ',strip=True) if h1 else ''
    meta=soup.find('meta',attrs={'property':'article:published_time'})
    t=soup.find('time')
    published=(meta.get('content') if meta else '') or ((t.get('datetime') or t.get_text(' ',strip=True)) if t else '')
    body=soup.find('article') or soup.find('main')
    if body is None: return None
    for trash in body.find_all(['script','style','nav','footer','aside']):trash.decompose()
    paras=[p.get_text(' ',strip=True) for p in body.find_all('p')]
    text=' '.join(p for p in paras if len(p)>=30).strip()
    words=[w for w in word_tokenize(text,engine='newmm') if w.strip()]
    return {'doc_id':urlparse(url).path.rsplit('/',1)[-1], 'url':normalize_url(url),
            'title':title,'published_at':published,'text':text,'word_count':len(words),
            'source_domain':urlparse(url).netloc,'text_hash':sha256(text.encode('utf-8')).hexdigest()}
````

#### Step 5 — ให้ Robot ทำซ้ำกับหลายบทความ

นี่คือส่วนที่ทำให้มันเป็น “Robot”

แทนที่จะเปิดข่าวด้วยมือทีละหน้า โปรแกรมจะวน URL ที่หาได้ใน Step 3 แล้วเรียก `extract_article()` ซ้ำ ๆ

มี `time.sleep(DELAY)` ทุกครั้งเพื่อไม่ส่ง request ถี่เกินไป

ในช่วงพัฒนาให้เริ่มจาก 5–20 ข่าวก่อน เมื่อดูตัวอย่างแล้วแน่ใจว่า `text` เป็นเนื้อหาข่าวจริง ค่อยเพิ่มจำนวน


````python
# ===== Step 5: เก็บหลายบทความอัตโนมัติ =====

def collect_articles(article_urls, max_articles=20, delay=1.5):
    rows = []

    for i, url in enumerate(article_urls[:max_articles], start=1):
        try:
            item = extract_article(url)

            if item is not None:
                rows.append(item)
                print(f"[{i}] OK   :", item["title"][:70])
            else:
                print(f"[{i}] SKIP :", url)

        except Exception as e:
            print(f"[{i}] ERROR:", url, e)

        time.sleep(delay)

    return rows


rows = collect_articles(
    article_urls,
    max_articles=MAX_ARTICLES,
    delay=DELAY
)

print("\nเก็บบทความที่ใช้ได้ =", len(rows))
````

### 3.3 หา Insight

ใช้ **1–2 เทคนิค** ที่เหมาะกับคำถาม เช่น

* frequency / n-gram

สร้างกราฟหรือตารางหลักอย่างน้อย **1 ชิ้น** แล้วตอบว่า

> “จากข้อมูลนี้ ผู้ใช้ผลวิเคราะห์ควรรู้อะไรหรือทำอะไรต่อ?”


````python
# ----------------- Your code here -----------------
import re
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

# 1. โหลดข้อมูล Final Corpus จากไฟล์ corpus_cleaned.jsonl
corp_analysis = pd.read_json("corpus_cleaned.jsonl", lines=True)
corp_analysis['published_at'] = pd.to_datetime(corp_analysis['published_at'])

# 2. ทำการสกัดคำศัพท์หรือคีย์เวิร์ดเด่น (Keyword Frequency)
# เนื่องจากเป็นภาษาไทย เราใช้การจับกลุ่มคีย์เวิร์ดที่เป็นประเด็นข่าวหลักเพื่อหาว่าข่าวพูดถึงอะไรมากที่สุด
keywords = {
    'ศาลรัฐธรรมนูญ / กฎหมาย': r'ศาล|ศาลรัฐธรรมนูญ|รัฐธรรมนูญ|คำวินิจฉัย|คดี',
    'การเลือกตั้ง / สว. / อบจ.': r'เลือกตั้ง|สว|กกต|อบจ|คูหา|ฮั้ว',
    'งบประมาณ / ดิจิทัลวอลเล็ต': r'งบ|งบประมาณ|ดิจิทัลวอลเล็ต|แจกเงิน',
    'ความขัดแย้ง / ชายแดนใต้': r'ชายแดนใต้|ความรุนแรง|ยะลา|ปัตตานี|นราธิวาส'
}

# นับจำนวนข่าวที่กล่าวถึงแต่ละประเด็น
keyword_counts = {}
for category, pattern in keywords.items():
    count = corp_analysis['text'].str.contains(pattern, flags=re.IGNORECASE, regex=True).sum()
    keyword_counts[category] = count

df_keywords = pd.DataFrame(list(keyword_counts.items()), columns=['ประเด็นข่าว', 'จำนวนข่าว']).sort_values(by='จำนวนข่าว', ascending=True)
````

````python
# ----------------- Your code here -----------------
# กราฟสำหรับนำเสนอ
# 3. สร้างกราฟแสดงความสนใจของสื่อ (Media Attention)
plt.style.use('default')
fig, ax = plt.subplots(figsize=(10, 5), facecolor='#FFFFFF')
ax.set_facecolor('#FFFFFF')

colors = ['#CCCCCC', '#94A3B8', '#1DB954', '#1E3A8A'] # เน้นประเด็นสำคัญสีเขียวเข้มและน้ำเงิน

bars = ax.barh(df_keywords['ประเด็นข่าว'], df_keywords['จำนวนข่าว'], color=colors[:len(df_keywords)], edgecolor='#0F172A', height=0.55)

# ใส่ตัวเลขบนแท่งกราฟ
for bar in bars:
    width = bar.get_width()
    ax.annotate(f'{int(width)} ข่าว ({width/len(corp_analysis)*100:.1f}%)',
                (width, bar.get_y() + bar.get_height() / 2.0),
                ha='left', va='center', xytext=(5, 0),
                textcoords='offset points', fontsize=10, fontweight='bold', color='#0F172A')

ax.set_title('การกระจายตัวของประเด็นข่าวในคลังข้อมูล (Media Attention Distribution)', fontsize=14, fontweight='bold', color='#0F172A', pad=15)
ax.set_xlabel('จำนวนบทความข่าว (ชิ้น)', fontsize=11, color='#0F172A')
ax.set_xlim(0, len(corp_analysis) * 1.2)

# ซ่อนเส้นขอบ
for spine in ['top', 'right', 'bottom']:
    ax.spines[spine].set_visible(False)
ax.spines['left'].set_color('#CCCCCC')
ax.grid(axis='x', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()

# 4. วิเคราะห์สรุปเชิงธุรกิจ
print("\n=== จากข้อมูลนี้ ผู้ใช้ผลวิเคราะห์ควรรู้อะไรหรือทำอะไรต่อ? ===")
print("1. สื่อให้ความพึงพอใจและโฟกัสหลักไปที่การตรวจสอบกลไกของรัฐและการเมืองส่วนกลาง (Political & Legal Scrutiny):")
print(f"   - ประเด็นข่าวเรื่อง 'การเลือกตั้ง / สว. / อบจ.' และ 'ศาลรัฐธรรมนูญ / กฎหมาย' ครองพื้นที่ข่าวกว่า 70% ของ Corpus ข้อมูล")
print("   - บ่งชี้ว่าผู้บริโภคสื่อในช่องทางนี้กำลังตื่นตัวและเสพข่าวเชิงโครงสร้างทางกฎหมายและการเมืองอย่างใกล้ชิด")
print("\n2. สิ่งที่แบรนด์หรือผู้ใช้ผลวิเคราะห์ควรทำต่อ (Strategic Actions):")
print("   - หลีกเลี่ยงการทำแคมเปญการตลาดเชิงอารมณ์ขันหรือเรื่องเบาสมองในช่วงเวลาที่กระแสข่าวเชิงตรวจสอบกำลังตึงเครียด")
print("   - ปรับการสื่อสารของแบรนด์ให้อยู่บนพื้นฐานของ 'ความโปร่งใส ความถูกต้อง และตรวจสอบได้' (Transparency & Integrity) สอดคล้องกับโทนกระแสสังคม")
print("   - สำหรับแบรนด์ที่ทำคอนเทนต์การเมือง/นโยบายสาธารณะ ควรขยายผลจากการสกัดประเด็นเหล่านี้ไปสู่การสร้าง 'สรุปประเด็นเข้าใจง่าย (Explainer Content)' เพื่อดึงยอด Engagement")
````
