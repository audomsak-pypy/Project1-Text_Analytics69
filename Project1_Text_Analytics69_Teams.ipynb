## เทคนิคที่อาจใช้ — เลือกเท่าที่จำเป็น

ไม่ต้องใช้ทั้งหมด โดยทั่วไป **2–4 เทคนิคก็เพียงพอ** ให้เลือกตามคำถามของทีม

| เทคนิค | เหมาะเมื่ออยากรู้ |
|---|---|
| **Word / n-gram frequency** | คนพูดถึงอะไรบ่อย |
| **TF-IDF / distinctive words** | สองกลุ่มพูดต่างกันอย่างไร |
| **Sentiment** | ข้อความเป็นบวก/ลบเพียงใด |
| **Aspect analysis** | คนพูดถึงราคา คุณภาพ การส่งของ ฯลฯ อย่างไร |
| **Topic modeling** | มีประเด็นอะไรที่เราอาจยังไม่รู้ |
| **Trend / segmentation** | ปัญหาเปลี่ยนตามเวลา หรือหนักในกลุ่มไหน |

> เทคนิคขั้นสูง เช่น embedding, clustering, regression, classification หรือ anomaly detection ทำได้ แต่ **ไม่บังคับ** และเหมาะสำหรับ Bonus เมื่อใช้ถูกและอธิบายได้

ถ้ายังไม่เคยใช้เทคนิคใด ให้ทดลองกับข้อมูลชุดเล็กก่อน แล้วตรวจด้วยตาว่าผลสมเหตุสมผลหรือไม่

ทุก insight เขียนในรูปแบบนี้ 5 บรรทัด ไม่ต้องยาว

> **หลักฐาน:** ตัวเลขพร้อม n และชี้ว่าดูได้จากกราฟหรือตารางไหน \

### 2.2 เตรียมข้อมูลที่เชื่อถือได้

ทำเพียงสิ่งจำเป็นก่อนเริ่มวิเคราะห์

* สุ่มรีวิวอย่างน้อย **20,000 รายการ** และอย่าใช้เพียง N บรรทัดแรก
* เชื่อม review กับ product metadata ด้วย `parent_asin`
* รายงานสั้น ๆ ว่า join สำเร็จกี่เปอร์เซ็นต์ และมี missing/duplicate สำคัญหรือไม่
* เขียน 2–3 บรรทัดว่า **ข้อมูลนี้เป็นตัวแทนของใคร และไม่เป็นตัวแทนของใคร**

> จุดประสงค์คือให้รู้ว่าข้อมูลที่ใช้เชื่อถือได้พอสำหรับคำถามของทีม ไม่ใช่ทำ data cleaning ทุกอย่างที่เป็นไปได้


````python
def iter_jsonl(path):
    with open(path, encoding='utf-8') as f:
        for line in f:
            yield json.loads(line)
````

### Sampling Reviews from `review_path`

````python
# Sample N reviews from the review_path
print(f"Sampling {SAMPLE_N:,} reviews from {review_path}...")

sampled_reviews = []
for i, review in enumerate(iter_jsonl(review_path)):
    if len(sampled_reviews) < SAMPLE_N:
        sampled_reviews.append(review)
    else:
        # Reservoir sampling to get a random sample of N reviews
        r = random.randint(0, i)
        if r < SAMPLE_N:
            sampled_reviews[r] = review

df_reviews = pd.DataFrame(sampled_reviews)
print(f"Loaded {len(df_reviews):,} sampled reviews.")

# Load all meta data
print(f"Loading metadata from {meta_path}...")
df_meta = pd.DataFrame(iter_jsonl(meta_path))
print(f"Loaded {len(df_meta):,} metadata items.")

display(df_reviews.head())
````

### Joining Reviews with Product Metadata

````python
# Perform a left merge to enrich reviews with product metadata
initial_review_count = len(df_reviews)
df_joined = pd.merge(df_reviews, df_meta, on='parent_asin', how='left', suffixes=('_review', '_meta'))

# Report join success percentage
successful_join_count = len(df_joined.dropna(subset=['main_category'])) # Assuming main_category is a good indicator of a successful meta join
join_percentage = (successful_join_count / initial_review_count) * 100
print(f"Successfully joined {successful_join_count:,} reviews with metadata ({join_percentage:.2f}%).")

# Check for missing values in critical columns
print("\nMissing values after join:")
missing_values = df_joined[['text', 'title_review', 'main_category', 'price']].isnull().sum()
display(missing_values[missing_values > 0])

# Check for duplicate reviews (based on a combination of user_id, timestamp, and asin)
# Note: parent_asin is the product group, asin is the specific product variant.
# We are interested in unique reviews, which is usually identified by user_id and timestamp for a product.
df_joined_before_dedup = len(df_joined)
df_joined = df_joined.drop_duplicates(subset=['user_id', 'timestamp', 'asin'])
duplicate_removed_count = df_joined_before_dedup - len(df_joined)
print(f"\nRemoved {duplicate_removed_count:,} duplicate reviews.")

display(df_joined.head())
````

ชุดข้อมูลนี้เป็นตัวแทนของรีวิวจำนวน 20,000 รายการ ที่ถูกสุ่มเลือกมาจากสินค้าในหมวดหมู่ 'Digital_Music' บน Amazon โดยรีวิวเหล่านี้รวมถึงข้อมูลอธิบายเพิ่มเติม (Metadata) ที่เกี่ยวข้อง เช่น ชื่อสินค้า หมวดหมู่ และคะแนนดาวเฉลี่ย ข้อมูลตัวอย่างนี้ช่วยให้เห็นภาพรวมความรู้สึกของลูกค้า (Customer Sentiment) และลักษณะเฉพาะของสินค้าในหมวดหมู่นี้

### 2.3 สำรวจข้อมูลเท่าที่จำเป็น

เลือกดูเฉพาะสิ่งที่ช่วยให้เข้าใจ brief ของทีม เช่น

* การแจกแจงดาว
* verified purchase / helpful vote
* ราคา หรือกลุ่มสินค้า
* เวลา
* ความยาวรีวิว

ทำ **2–4 ตารางหรือกราฟ** ก็พอ แล้วเขียนสั้น ๆ ว่า

> “สิ่งใดจาก EDA ทำให้ทีมอยากวิเคราะห์ต่อ และเพราะอะไร”

ไม่ต้องทำ EDA ทุกตัวแปร


````python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Prepare data for plotting
# Calculate counts of ratings for verified and non-verified purchases
rating_counts = df_joined.groupby(['verified_purchase', 'rating']).size().unstack(fill_value=0)

# Calculate percentages for each group (row-wise)
rating_percentages = rating_counts.apply(lambda x: x / x.sum(), axis=1)

# Melt the dataframe for seaborn's barplot
df_plot = rating_percentages.stack().reset_index(name='percentage')
df_plot['rating'] = df_plot['rating'].astype(str) + ' Star'

# --- Plotting ---

# 1. Set up the plot with Spotify Light Mode style
plt.style.use('default')
fig, ax = plt.subplots(figsize=(10, 6), facecolor='#FFFFFF')
ax.set_facecolor('#FFFFFF')

# Spotify palette: Verified = Bright Green, Non-Verified = Dark Charcoal
spotify_colors = {
    True: '#1DB954',
    False: '#191414'
}

# Create the clustered bar chart
sns.barplot(
    data=df_plot,
    x='rating',
    y='percentage',
    hue='verified_purchase',
    palette=spotify_colors,
    ax=ax,
    edgecolor='none' # Remove bar borders
)

# 3. Apply Spotify Minimal Styling
ax.set_title('Rating Distribution by Verified Purchase Status', fontsize=18, fontweight='bold', color='#191414', pad=20)
ax.set_xlabel('') # Hide x-axis label
ax.set_ylabel('') # Hide y-axis label

# Adjust x-axis tick labels for clarity
ax.tick_params(axis='x', colors='#191414', labelsize=12, rotation=0)
ax.tick_params(axis='y', colors='#555555', labelsize=11)

# Remove all spines
for spine in ['top', 'right', 'left', 'bottom']:
    ax.spines[spine].set_visible(False)

# Add horizontal grid lines
ax.grid(axis='y', linestyle='--', alpha=0.4, color='#CCCCCC')

# Add annotations (percentages) on top of each bar
for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.annotate(f'{height:.1%}',
                    (p.get_x() + p.get_width() / 2., height),
                    ha='center', va='bottom', xytext=(0, 5),
                    textcoords='offset points', fontsize=10, color='#191414', fontweight='bold')

# Adjust legend
handles, labels = ax.get_legend_handles_labels()
ax.legend(handles=handles, labels=['Verified Purchase', 'Non-Verified Purchase'],
          title='Purchase Status', frameon=False, fontsize=10, title_fontsize=12,
          labelcolor='#191414', loc='upper left')

plt.tight_layout()
plt.show()
````

**ผลการวิเคราะห์ข้อมูล:**

**การกระจายตัวเอียงไปทางเชิงบวกอย่างมาก (Positive Skewness)**:ไม่ว่าจะเป็นผู้ซื้อที่ได้รับการยืนยันหรือไม่ยืนยัน เสียงส่วนใหญ่กระจุกตัวอยู่ที่ 5 ดาว สูงที่สุด (เกิน 70% ทั้งสองกลุ่ม) แสดงให้เห็นว่าลูกค้ารายการนี้พึงพอใจสูงมาก  

**กลุ่ม Non-Verified มีแนวโน้มให้ 5 ดาวสูงกว่าอย่างมีนัยสำคัญ:**
 - กลุ่ม Non-Verified (79.8%) มีสัดส่วนรีวิว 5 ดาว สูงกว่ากลุ่ม Verified (70.5%) ถึงเกือบ 10%  
 - ข้อสังเกตนี้อาจสะท้อนถึงอคติเชิงบวก (Positivity Bias) หรือกลุ่มผู้รีวิวที่ไม่แน่ชัดว่าซื้อจริง ซึ่งมักกระจุกตัวให้คะแนนเต็มได้ง่ายกว่า   

**ผู้ซื้อจริง (Verified Purchase) มีความประณีต/วิพากษ์วิจารณ์มากกว่า (More Critical):**

- กลุ่ม Non-Verified: มักเป็นผู้ที่ตั้งใจเข้ามาเขียนรีวิวขนาดยาว เล่าเรื่องราวเชิงลึก วิเคราะห์รายละเอียดอัลบั้ม/เพลง หรือใส่ข้อคิดเห็นที่มีอารมณ์ร่วมสูง ซึ่งเนื้อหาที่มีความลึกและความละเอียดนี้คือปัจจัยหลักที่ดึงดูดให้ผู้อื่นกดโหวตว่าเป็นประโยชน์ (Helpful Vote)   
- กลุ่ม Verified Purchase: ผู้ซื้อจริงส่วนใหญ่มักให้คะแนนดาวและเขียนข้อความสั้นๆ ตรงประเด็น (เช่น "ส่งเร็ว เสียงดี") ซึ่งแม้จะสะท้อนความพึงพอใจจริง แต่ด้วยข้อความที่กระชับจึงไม่ค่อยได้รับ Helpful Vote จากผู้อื่น   

**นัยสำคัญทางธุรกิจ (Business Implications):ไม่ใช่จำนวนผู้ซื้อจริง แต่คือ "ความละเอียดของเนื้อหา"**:
- การสร้าง Engagement หรือความน่าเชื่อถือบนแพลตฟอร์มไม่ได้ขึ้นอยู่กับสถานะ Verified เพียงอย่างเดียว แต่เกิดจาก เนื้อหาที่มีความยาวและการเล่าเรื่องเชิงลึก (Long-form Storytelling)

### 2.4 เตรียมข้อความและสร้าง Text Features

ก่อนนำข้อความไปวิเคราะห์ จำเป็นต้องมีการ **preprocessing** เพื่อทำให้ข้อความอยู่ในรูปที่เหมาะสมกับเทคนิคที่จะใช้ อย่างไรก็ตาม ในงานนี้ไม่จำเป็นต้องทำ preprocessing ทุกขั้นตอน ให้เลือกทำ **เท่าที่จำเป็น** ตามวัตถุประสงค์ของการวิเคราะห์

ตัวอย่างเช่น การเปลี่ยนข้อความเป็นตัวพิมพ์เล็ก (lower case) การนำ URL หรือ HTML ออก การแบ่งข้อความเป็นคำ (tokenization) หรือการตัด stopwords เช่น `the`, `and`, `is` ออก ในกรณีที่คำเหล่านี้ไม่ได้ช่วยในการวิเคราะห์

**ไม่บังคับให้ใช้ stemming หรือ lemmatization** หากไม่มีเหตุผลว่าขั้นตอนดังกล่าวจะช่วยตอบคำถามของทีมได้ดีขึ้น

หลังจากเตรียมข้อความแล้ว ให้สร้าง **Text Features อย่างน้อย 2 ชนิด** เพื่อนำข้อความมาเปลี่ยนให้อยู่ในรูปข้อมูลที่สามารถนำไปสำรวจ เปรียบเทียบ หรือวิเคราะห์ต่อได้

---

#### Text Feature คืออะไร?

Text Feature คือ **ข้อมูลเชิงตัวเลขที่สกัดหรือสร้างขึ้นจากข้อความ** เพื่อแทนลักษณะหรือความหมายบางอย่างของข้อความนั้น

Text Feature อาจอยู่ในรูปของ

* **ค่าตัวเลขค่าเดียว** เช่น จำนวนคำ หรือ sentiment score
* **ตัวแปรหลายตัว** เช่น `price=1`, `quality=0`, `delivery=1`
* **เวกเตอร์หลายมิติ** เช่น TF-IDF vector หรือ Text Embedding

ตัวอย่างเช่น รีวิว

> `"The product is easy to use but the delivery was very slow."`

สามารถเปลี่ยนเป็น feature ได้หลายรูปแบบ

| Feature                    |                 ตัวอย่างค่าที่ได้ |
| -------------------------- | --------------------------------: |
| จำนวนคำในรีวิว             |                                12 |
| Sentiment score            |                             -0.15 |
| กล่าวถึง usability หรือไม่ |                                 1 |
| กล่าวถึง delivery หรือไม่  |                                 1 |
| กล่าวถึง price หรือไม่     |                                 0 |
| TF-IDF vector              |   `[0.00, 0.41, 0.00, 0.72, ...]` |
| Embedding vector           | `[0.21, -0.08, 0.63, ..., -0.14]` |

ดังนั้นข้อความหนึ่งข้อความสามารถถูกแทนได้ทั้งด้วย **feature ที่ตีความได้โดยตรง** และ **feature แบบเวกเตอร์ที่ใช้แทนเนื้อหาหรือความหมายของข้อความ**

---

## ตัวอย่าง Text Features ที่สามารถเลือกใช้

ผมจะแก้ตารางเดิมเป็นแบบนี้:

| Text Feature          | คืออะไร                                     | ตัวอย่าง                     | ใช้ตอบคำถามอะไรได้                                    |
| --------------------- | ------------------------------------------- | ---------------------------- | ----------------------------------------------------- |
| **Review Length**     | จำนวนคำหรือตัวอักษร                         | 120 words                    | รีวิวที่ยาวกว่าได้รับ helpful vote มากกว่าหรือไม่     |
| **Sentiment Score**   | คะแนนความเป็นบวก/ลบ                         | `-0.72`, `+0.85`             | ความรู้สึกลูกค้าสัมพันธ์กับ rating หรือไม่            |
| **Aspect Indicators** | ระบุว่าพูดถึงหัวข้อใด                       | `delivery=1`, `price=0`      | ลูกค้าบ่นเรื่องใดมากที่สุด                            |
| **Lexical Diversity** | ความหลากหลายของคำ                           | unique words / total words   | รีวิวที่เขียนละเอียดมีประโยชน์มากกว่าหรือไม่          |
| **TF-IDF Vector**     | เวกเตอร์ที่ให้น้ำหนักคำเด่นในเอกสาร         | `[0, .35, 0, .81, ...]`      | คำใดแยกกลุ่มรีวิวบวกกับลบ                             |
| **Text Embedding**    | dense vector ที่แทนความหมายโดยรวมของข้อความ | `[.12, -.44, .08, ...]`      | รีวิวใดมีความหมายคล้ายกัน หรือมี customer themes อะไร |
| **Custom Feature**    | feature ที่ทีมสร้างตามโจทย์                 | `refund=1`, `defect_count=3` | ตอบ business question เฉพาะของทีม                     |

---

### ตัวอย่าง Embedding แบบง่าย

ถ้าสอน `sentence-transformers` มาแล้ว อาจใส่ตัวอย่างสั้น ๆ:

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

texts = [
    "The delivery was very slow.",
    "Shipping took too long.",
    "Excellent sound quality."
]

embeddings = model.encode(texts)

print(embeddings.shape)
```

**รายงานเพียง 3 อย่าง**
1. ทำ preprocessing อะไร และทำไม
2. สร้าง feature อะไรอย่างน้อย 2 ตัว
3. feature เหล่านั้นช่วยตอบคำถามทางธุรกิจอย่างไร

> ไม่ต้องรายงาน vocabulary ก่อน–หลัง เว้นแต่ประเด็นนั้นเกี่ยวข้องกับผลวิเคราะห์

````python
import html
import re
import numpy as np
import pandas as pd

# ==========================================
# 1. Text Preprocessing (เร่งสปีดด้วย List & Pre-compiled Regex)
# ==========================================

# Pre-compile Regex ไว้ล่วงหน้าเพื่อไม่ต้องสร้างใหม่ทุกรอบ
url_re = re.compile(r'https?://\S+|www\.\S+')
html_re = re.compile(r'<.*?>')
space_re = re.compile(r'\s+')

# แปลงเป็น List ล่วงหน้า (เร็วกว่า Pandas .apply / .fillna มาก)
titles = df_joined['title_review'].fillna('').astype(str).tolist()
texts = df_joined['text'].fillna('').astype(str).tolist()


def clean_text_fast(title, text):
  combined = f'{title} {text}'
  combined = html.unescape(combined)
  combined = url_re.sub('', combined)
  combined = html_re.sub('', combined)
  combined = space_re.sub(' ', combined)
  return combined.strip().lower()


# รวมข้อความและคลีนในขั้นตอนเดียว
cleaned_list = [clean_text_fast(t, x) for t, x in zip(titles, texts)]
df_joined['cleaned_text'] = cleaned_list


# ==========================================
# 2. Create Rule-based & Style Features
# ==========================================

# 2.1 Review Length (นับคำผ่าน Python List)
df_joined['review_length'] = [len(x.split()) for x in cleaned_list]

# 2.2 Aspect Indicators (เวอร์ชันแคบ: เน้นคำที่สื่อ "คุณภาพเสียง" จริง)
ADJ_POS_NEG = (r'great|good|excellent|awesome|amazing|wonderful|superb|fantastic|'
               r'terrible|bad|poor|awful|clear|crisp|muddy|tinny|thin|flat')

sound_pattern = re.compile(
    r'\b(?:sound quality|audio quality'
    rf'|(?:{ADJ_POS_NEG})\s+(?:sound|audio)'                     # "awesome sound"
    rf'|sound(?:s|ed)?\s+(?:is |are |was |so |very |really )?(?:{ADJ_POS_NEG})'  # "sounds great"
    r'|crystal clear|distortion|distorted|hi-?fi|fidelity|remaster(?:ed)?|mastering'
    r'|vocals? (?:are|is|sound)s?)\b'
)

nostalgia_pattern = re.compile(
    r'\b(?:nostalgi(?:a|c)|childhood|brings? (?:me )?back|takes? me back|back in the day'
    r'|good old days|growing up|grew up|my youth|reminds? me of|memories)\b'
)

value_pattern = re.compile(
    r'\b(?:worth (?:every|the|it)|good (?:price|value)|great (?:price|value)|bargain|affordable'
    r'|cheap|overpriced|rip-?off|value for money|money well spent|waste of money|a steal)\b'
)

def flag(pattern, texts):
    return np.fromiter((1 if pattern.search(x) else 0 for x in texts), dtype=np.int8)

df_joined['mention_sound']     = flag(sound_pattern, cleaned_list)
df_joined['mention_nostalgia'] = flag(nostalgia_pattern, cleaned_list)
df_joined['mention_value']     = flag(value_pattern, cleaned_list)


# ==========================================
# แสดงผลลัพธ์ Features ใหม่
# ==========================================
display(
    df_joined[[
        'title_review',
        'cleaned_text',
        'review_length',
        'mention_sound',
        'mention_nostalgia',
        'mention_value',
    ]].head()
)
````

````python
def sample_for_review(feature_col, n=30, seed=42):
    """สุ่มรีวิวที่ match แล้วตัดข้อความมาให้อ่านเอง"""
    sub = df_joined[df_joined[feature_col] == 1]
    s = sub.sample(min(n, len(sub)), random_state=seed)[['cleaned_text']].copy()
    s['cleaned_text'] = s['cleaned_text'].str[:300]
    s['correct'] = ''   # <-- ให้คนในทีมกรอก 1 = ตรงความหมาย, 0 = ไม่ตรง
    return s.reset_index(drop=True)

review_sound     = sample_for_review('mention_sound')
review_nostalgia = sample_for_review('mention_nostalgia')
review_value     = sample_for_review('mention_value')

# บันทึกเป็น CSV ให้เปิดกรอกใน Excel/Sheets แล้วอัปโหลดกลับ
review_sound.to_csv('validate_sound.csv', index=False, encoding='utf-8-sig')
review_nostalgia.to_csv('validate_nostalgia.csv', index=False, encoding='utf-8-sig')
review_value.to_csv('validate_value.csv', index=False, encoding='utf-8-sig')
````

###  Text Preprocessing & Feature Engineering Breakdown

---

#### 1. Preprocessing (การเตรียมและทำความสะอาดข้อความ)

##### **ทำอะไรบ้าง (What it does):**
* **Combine Text:** รวมข้อความจาก `title_review` และ `text` เข้าด้วยกันเป็นข้อความเดียวเพื่อเก็บบริบทการรีวิวอย่างสมบูรณ์
* **Decode HTML Entities:** ถอดรหัสอักขระพิเศษของ HTML (เช่น เปลี่ยน `&amp;` ➔ `&`, `&quot;` ➔ `"`)
* **Remove URLs & HTML Tags:** ลบลิงก์เว็บไซต์ (`http://`, `www.`) และแท็ก HTML (เช่น `<br>`, `<b>`) ออกด้วย Regular Expressions (Regex)
* **Normalize Whitespace:** จัดการช่องว่างที่ซ้ำซ้อนให้เหลือเพียงช่องว่างเดียว
* **Lowercasing & Stripping:** แปลงข้อความทั้งหมดเป็นตัวพิมพ์เล็ก (Lowercasing) และตัดช่องว่างส่วนเกินหัว-ท้ายข้อความ

##### **ทำทำไม (Why do it):**
* **ลด Noise (สิ่งรบกวนข้อมูล):** แท็ก HTML หรือ URLs ไม่ได้ให้ความหมายทางอารมณ์หรือคุณค่าเชิงธุรกิจ แต่จะสร้างความสับสนให้ระบบวิเคราะห์
* **สร้าง Standard Data Format:** ปรับรูปแบบข้อความให้เป็นมาตรฐานเดียวกัน เช่น *"Sound"*, *"SOUND"*, *"sound"* หากไม่แปลงเป็นตัวพิมพ์เล็ก ระบบจะมองเป็นคนละคำกัน ส่งผลให้การนับความถี่คีย์เวิร์ดคลาดเคลื่อน
* **เพิ่มความเร็วและแม่นยำในการประมวลผล:** การคลีนข้อความล่วงหน้าและจัดเก็บเป็น Python List ช่วยให้กระบวนการจับรูปแบบวลี (Regex Matching) และการสแกนคำทำงานได้เร็วขึ้นอย่างมหาศาล

Part 1: Sentiment Contrast (เปรียบเทียบรีวิว 4–5 ดาว vs 1–2 ดาว)

````python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from scipy.stats import fisher_exact

# ==========================================
# 1. SET SPOTIFY THEME & PALETTE
# ==========================================
SPOTIFY_GREEN = '#1DB954'  # สีเขียวหลักสำหรับ Positive / HighImpact
NEGATIVE_RED = '#E24A4A'   # สีแดงอมส้มสำหรับ Negative / General
BG_COLOR = '#FFFFFF'
TEXT_COLOR = '#000000'
GRID_COLOR = '#CCCCCC'

plt.rcParams.update({
    'figure.facecolor': BG_COLOR,
    'axes.facecolor': BG_COLOR,
    'axes.edgecolor': GRID_COLOR,
    'axes.linewidth': 1.2,
    'axes.labelcolor': TEXT_COLOR,
    'axes.titlecolor': TEXT_COLOR,
    'xtick.color': TEXT_COLOR,
    'ytick.color': TEXT_COLOR,
    'grid.color': GRID_COLOR,
    'grid.linestyle': '--',
    'grid.alpha': 0.6,
    'font.size': 10,
})

# ==========================================
# 2. DATA CALCULATION FUNCTION
# ==========================================
FEATS = ['mention_sound', 'mention_nostalgia', 'mention_value']

def compare_rates(a, b, name_a, name_b, feats=FEATS):
    print(f'n {name_a} = {len(a):,} | n {name_b} = {len(b):,}')
    rows = []
    for c in feats:
        xa, xb = int(a[c].sum()), int(b[c].sum())
        _, p = fisher_exact([[xa, len(a)-xa], [xb, len(b)-xb]])
        rows.append({
            'feature': c,
            f'{name_a} %': round(xa/len(a)*100, 1), f'{name_a} count': xa,
            f'{name_b} %': round(xb/len(b)*100, 1), f'{name_b} count': xb,
            'p_value': round(p, 4)
        })
    return pd.DataFrame(rows)

# ==========================================
# 3. PLOTTING FUNCTION
# ==========================================
def plot_compare_rates(res_df, name_a, name_b, title="Aspect Mention Rate Comparison"):
    # แปลงข้อมูลเป็น Long Format สำหรับ Seaborn
    melted_df = res_df.melt(
        id_vars=['feature', 'p_value'],
        value_vars=[f'{name_a} %', f'{name_b} %'],
        var_name='Group',
        value_name='Mention Rate (%)'
    )
    melted_df['Group'] = melted_df['Group'].str.replace(' %', '')
    # ปรับชื่อ Feature ให้สวยงาม (เช่น mention_sound -> Sound)
    melted_df['Aspect'] = melted_df['feature'].str.replace('mention_', '').str.title()

    fig, ax = plt.subplots(figsize=(9, 5.5), dpi=120)

    sns.barplot(
        data=melted_df,
        x='Aspect',
        y='Mention Rate (%)',
        hue='Group',
        ax=ax,
        palette=[SPOTIFY_GREEN, NEGATIVE_RED],
        edgecolor=TEXT_COLOR,
    )

    # ตัวเลขเปอร์เซ็นต์บนหัวแท่งกราฟ
    for p in ax.patches:
        h = p.get_height()
        if h > 0:
            ax.annotate(
                f'{h:.1f}%',
                (p.get_x() + p.get_width() / 2.0, h),
                ha='center',
                va='bottom',
                color=TEXT_COLOR,
                fontweight='bold',
                fontsize=9.5,
                xytext=(0, 3),
                textcoords='offset points'
            )

    # แสดง p-value ด้านบนของแต่ละคู่อย่างสวยงาม
    max_y = melted_df['Mention Rate (%)'].max()
    for idx, row in res_df.iterrows():
        p_val = row['p_value']
        p_text = "p < 0.001" if p_val < 0.001 else f"p = {p_val:.4f}"
        feat_max = melted_df[melted_df['feature'] == row['feature']]['Mention Rate (%)'].max()
        ax.text(
            idx, feat_max + (max_y * 0.12), p_text,
            ha='center', va='bottom', fontsize=9, fontweight='bold', color='#444444'
        )

    # ตกแต่งรายละเอียดกราฟ
    ax.set_title(title, fontsize=13, color=TEXT_COLOR, fontweight='bold', pad=15)
    ax.set_xlabel('Marketing Aspect', fontsize=11, color=TEXT_COLOR)
    ax.set_ylabel('Mention Rate (%)', fontsize=11, color=TEXT_COLOR)
    ax.set_ylim(0, max_y * 1.35)  # ขยายขอบเขตแกน Y ให้พอดีกับข้อความ p-value
    ax.legend(
        title='Group',
        facecolor=BG_COLOR,
        edgecolor=GRID_COLOR,
        labelcolor=TEXT_COLOR,
    )
    ax.grid(True)

    plt.tight_layout()
    plt.show()

# ==========================================
# 4. EXECUTION
# ==========================================
# 1) เปรียบเทียบ Positive vs Negative
pos_df = df_joined[df_joined['rating'] >= 4]
neg_df = df_joined[df_joined['rating'] <= 2]
res_pos_neg = compare_rates(pos_df, neg_df, 'Positive', 'Negative')
display(res_pos_neg)

plot_compare_rates(
    res_pos_neg,
    'Positive',
    'Negative',
    title='Sentiment Contrast: Aspect Mention Rate (Positive vs. Negative Reviews)'
)

# 2) เปรียบเทียบ HighImpact vs General
high_df = df_joined[df_joined['helpful_vote'] >= 5]
gen_df  = df_joined[df_joined['helpful_vote'] < 5]
res_high_gen = compare_rates(high_df, gen_df, 'HighImpact', 'General')
display(res_high_gen)

plot_compare_rates(
    res_high_gen,
    'HighImpact',
    'General',
    title='Impact Contrast: Aspect Mention Rate (HighImpact vs. General Reviews)'
)
````

Part 3: Inferred Genre Segmentation (วิเคราะห์แยกตามแนวเพลง)

````python
import ast
import re
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# ==========================================
# 1. FEATURE ENGINEERING & GENRE INFERENCE (REGEX)
# ==========================================
def extract_clean_artist(store_val):
    if pd.isna(store_val):
        return 'Unknown'
    clean_val = re.sub(r'Format:.*$', '', str(store_val))
    clean_val = re.sub(r'\(Artist\)', '', clean_val)
    return clean_val.strip()

df_joined['clean_artist'] = df_joined['store'].apply(extract_clean_artist)

# กฎสำหรับจําแนกแนวเพลงด้วย Regular Expression
GENRE_RULES = [
    ('K-Pop / Asian Pop',        r'\b(?:bts|bangtan|k-?pop|hanteo|photobook)\b'),
    ('Classical / Jazz',         r'\b(?:jazz|classical|symphony|orchestra|concerto|opera|yo-yo ma)\b'),
    ('Classic Rock / Oldies',    r"\b(?:beatles|elvis|pink floyd|grateful dead|rock|oldies|bbc sessions|60s|70s|80s)\b"),
    ('Dance / R&B / Electronic', r'\b(?:r&b|house music|dj|remix|dance|electronic|edm)\b'),
    ('Audiobook / Spoken Word',  r'\b(?:unabridged|audiobook|spoken word)\b'),
]
GENRE_RULES = [(g, re.compile(p, flags=re.IGNORECASE)) for g, p in GENRE_RULES]

def infer_genre(title, desc, artist):
    d = ' '.join(map(str, desc)) if isinstance(desc, list) else str(desc or '')
    text = f'{title} {d} {artist}'.lower()
    for g, pat in GENRE_RULES:
        if pat.search(text):
            return g
    return 'General / Mainstream'

# สร้างคอลัมน์ inferred_genre
df_joined['inferred_genre'] = [
    infer_genre(t, d, a) for t, d, a in
    zip(df_joined['title_meta'], df_joined['description'], df_joined['clean_artist'])
]

print("=== Distribution of Inferred Genres ===")
print(df_joined['inferred_genre'].value_counts())
print("=" * 40)

# ==========================================
# 2. PREPARE DATA FOR PLOTTING
# ==========================================
# กรองเฉพาะรีวิว Verified Purchase และมี Rating >= 4
brief_df = df_joined[
    (df_joined['verified_purchase'] == True) & (df_joined['rating'] >= 4)
].copy()

# คำนวณ % การพูดถึงแต่ละ Aspect แยกตาม Inferred Genre
genre_matrix = (
    brief_df.groupby('inferred_genre', observed=False)
    .agg(
        Nostalgia=('mention_nostalgia', lambda x: x.mean() * 100 if len(x) > 0 else 0),
        Sound_Quality=('mention_sound', lambda x: x.mean() * 100 if len(x) > 0 else 0),
        Value_for_Money=('mention_value', lambda x: x.mean() * 100 if len(x) > 0 else 0),
    )
    .reset_index()
)

# แปลงข้อมูลเป็น Long-format สำหรับ Seaborn
genre_melted = genre_matrix.melt(
    id_vars=['inferred_genre'],
    value_vars=['Nostalgia', 'Sound_Quality', 'Value_for_Money'],
    var_name='Aspect',
    value_name='Mention_Rate',
)
genre_melted['Aspect'] = genre_melted['Aspect'].replace({
    'Sound_Quality': 'Sound Quality',
    'Value_for_Money': 'Value for Money',
})

# ==========================================
# 3. PLOT LIGHT THEME GRAPH (SPOTIFY STYLE)
# ==========================================
SPOTIFY_GREEN = '#1DB954'
SPOTIFY_LIGHT_GREEN = '#1ED760'
GENERAL_GRAY = '#535353'
BG_COLOR = '#FFFFFF'
TEXT_COLOR = '#000000'
GRID_COLOR = '#CCCCCC'

plt.rcParams.update({
    'figure.facecolor': BG_COLOR,
    'axes.facecolor': BG_COLOR,
    'axes.edgecolor': GRID_COLOR,
    'axes.linewidth': 1.2,
    'axes.labelcolor': TEXT_COLOR,
    'axes.titlecolor': TEXT_COLOR,
    'xtick.color': TEXT_COLOR,
    'ytick.color': TEXT_COLOR,
    'grid.color': GRID_COLOR,
    'grid.linestyle': '--',
    'grid.alpha': 0.6,
    'font.size': 10,
})

fig, ax = plt.subplots(figsize=(12, 6), dpi=120)

palette = [SPOTIFY_GREEN, SPOTIFY_LIGHT_GREEN, GENERAL_GRAY]

bar_plot = sns.barplot(
    data=genre_melted,
    x='inferred_genre',
    y='Mention_Rate',
    hue='Aspect',
    palette=palette,
    edgecolor=TEXT_COLOR,
    ax=ax,
)

# แสดงตัวเลข % บนหัวแท่งกราฟ
for p in ax.patches:
    height = p.get_height()
    if not np.isnan(height) and height > 0:
        ax.annotate(
            f'{height:.1f}%',
            (p.get_x() + p.get_width() / 2.0, height),
            ha='center',
            va='bottom',
            xytext=(0, 3),
            textcoords='offset points',
            color=TEXT_COLOR,
            fontweight='bold',
            fontsize=8.5,
        )

# ตกแต่งองค์ประกอบกราฟ
ax.set_title(
    'Category-to-Message Fit (Inferred from Title, Description & Store via Regex)',
    fontsize=13,
    fontweight='bold',
    color=TEXT_COLOR,
    pad=15,
)
ax.set_xlabel('Inferred Music Genre / Segment', fontsize=11, color=TEXT_COLOR)
ax.set_ylabel('Mention Rate in Positives (%)', fontsize=11, color=TEXT_COLOR)

max_rate = genre_melted['Mention_Rate'].max()
ax.set_ylim(0, (max_rate * 1.18) if max_rate > 0 else 100)

plt.xticks(rotation=15, ha='right', fontsize=10, color=TEXT_COLOR)

ax.legend(
    title='Marketing Aspect',
    facecolor=BG_COLOR,
    edgecolor=GRID_COLOR,
    labelcolor=TEXT_COLOR,
    fontsize=10,
)
ax.grid(True)

plt.tight_layout()
plt.show()
````

เมื่อดูตาม Category จะเห็นว่า phrase ที่เกี่ยวกับ Sound Quality ได้แก่ sound quality, great sound และ excellent sound

ส่วนด้าน Nostalgia ในกราฟนี้มี back in the day เป็นตัวแทนที่เห็นชัด โดยคำเหล่านี้ช่วยทำให้เรารู้ว่าไม่ได้มีเพียงการพูดถึง Aspect ในเชิงนามธรรม แต่มี ภาษาที่ลูกค้าใช้จริง ซึ่งสามารถนำไปเป็นวัตถุดิบในการคิด Marketing Message ได้

อย่างไรก็ตาม จุดสำคัญมากคือ 1.27% ไม่ได้หมายความว่า 1.27% ของทุกคำในรีวิวพูดว่า “sound quality” และไม่ได้หมายความว่า 1.27% ของลูกค้าทั้งหมดชอบ Sound Quality เพราะมันหมายถึงว่าใน Positive Reviews จำนวน 17,602 รีวิว มี 224 รีวิวที่ตรวจพบ phrase “sound quality” เท่านั้น ดังนั้น VoC นี้ควรใช้เป็นหลักฐานว่า วลีดังกล่าวปรากฏซ้ำในเสียงของลูกค้า ไม่ควรนำไปกล่าวว่าคนส่วนใหญ่พูดแบบนี้


````python
import re
import pandas as pd

# 1. คลีนชื่อศิลปินจาก store (ตัด 'Format: Audio CD' ออก)
def extract_artist(store_val):
    if pd.isna(store_val):
        return 'Unknown'
    # ลบข้อความเกี่ยวกับ Format และการเว้นวรรคส่วนเกิน
    clean_val = re.sub(r'Format:.*$', '', str(store_val))
    clean_val = re.sub(r'\(Artist\)', '', clean_val)
    return clean_val.strip()

df_joined['clean_artist'] = df_joined['store'].apply(extract_artist)

# 2. สร้าง Sub-genre และ Era ขึ้นมาใหม่จาก Description + Title + Artist
def infer_marketing_segment(row):
    text = f"{row.get('title_meta', '')} {row.get('description', '')} {row.get('clean_artist', '')}".lower()

    # Matching กฎการตลาด
    if any(k in text for k in ['bts', 'bangtan', 'k-pop', 'hanteo', 'photobook', 'mini album']):
        return 'K-Pop / Asian Pop'
    elif any(k in text for k in ['beatles', 'elvis', 'pink floyd', 'grateful dead', 'rock', '70s', '80s', '1977', '1971', 'classic']):
        return 'Classic Rock / Oldies'
    elif any(k in text for k in ['r&b', 'house', 'mix', 'dj', 'dance', 'party']):
        return 'Dance / R&B / Electronic'
    elif any(k in text for k in ['unabridged', 'pajamas', 'book', 'conversation']):
        return 'Audiobook / Spoken Word'
    else:
        return 'General / Mainstream'

df_joined['inferred_genre'] = df_joined.apply(infer_marketing_segment, axis=1)

# ตรวจสอบผลการจัดกลุ่มใหม่
print("=== สรุปหมวดหมู่ใหม่ที่สกัดจาก Description & Store ===")
print(df_joined['inferred_genre'].value_counts())
````

เลือกอย่างน้อยหนึ่งคำอธิบาย แล้ว **ทดสอบด้วยตัวเลขจริง** สรุปว่า insight ยังยืนอยู่ได้ หรือต้องปรับถ้อยคำให้แคบลง

````python
# ============================================================
# 2.6 REALITY CHECK: ตรวจสอบผลกระทบจาก Review Length
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# 1) ตรวจสอบจำนวนตัวอย่าง High Impact / General
# ------------------------------------------------------------

df_joined['is_high_impact'] = np.where(
    df_joined['helpful_vote'] >= 5,
    'High Impact (≥5)',
    'General (<5)'
)

n_general = (df_joined['helpful_vote'] < 5).sum()
n_high = (df_joined['helpful_vote'] >= 5).sum()
total_reviews = len(df_joined)

print("=== SAMPLE SIZE CHECK ===")
print(f"General Reviews (<5 votes): {n_general:,} "
      f"({n_general / total_reviews * 100:.1f}%)")

print(f"High Impact Reviews (≥5 votes): {n_high:,} "
      f"({n_high / total_reviews * 100:.1f}%)")

# ------------------------------------------------------------
# 2) แบ่งกลุ่มตามความยาวรีวิว
# ------------------------------------------------------------
# ใช้เกณฑ์เดิมของงาน เพื่อให้เทียบกับผลก่อนหน้าได้

df_joined['length_bin'] = pd.cut(
    df_joined['review_length'],
    bins=[0, 30, 100, np.inf],
    labels=[
        'Short (≤30 words)',
        'Medium (31–100 words)',
        'Long (>100 words)'
    ]
)

# ------------------------------------------------------------
# 3) คำนวณ Mention Rate + จำนวนรีวิวในแต่ละกลุ่ม
# ------------------------------------------------------------

group_cols = ['length_bin', 'is_high_impact']
feature_cols = [
    'mention_sound',
    'mention_nostalgia',
    'mention_value'
]

reality_check_df = (
    df_joined
    .groupby(group_cols, observed=False)
    .agg(
        n=('asin', 'size'),
        mention_sound=('mention_sound', 'mean'),
        mention_nostalgia=('mention_nostalgia', 'mean'),
        mention_value=('mention_value', 'mean')
    )
    .reset_index()
)

# เปลี่ยนค่า mean ของ binary เป็น %
for c in feature_cols:
    reality_check_df[c] = reality_check_df[c] * 100

print("\n=== REALITY CHECK SUMMARY ===")
display(
    reality_check_df.round({
        'mention_sound': 2,
        'mention_nostalgia': 2,
        'mention_value': 2
    })
)
````

````python
# ============================================================
# 4) คำนวณความแตกต่าง High Impact - General
# ============================================================

comparison_rows = []

for length in reality_check_df['length_bin'].dropna().unique():

    sub = reality_check_df[
        reality_check_df['length_bin'] == length
    ]

    general = sub[
        sub['is_high_impact'] == 'General (<5)'
    ].iloc[0]

    high = sub[
        sub['is_high_impact'] == 'High Impact (≥5)'
    ].iloc[0]

    comparison_rows.append({
        'Review Length': length,

        'n General': int(general['n']),
        'n High Impact': int(high['n']),

        'Sound General %': general['mention_sound'],
        'Sound High Impact %': high['mention_sound'],
        'Sound Difference (pp)': (
            high['mention_sound'] - general['mention_sound']
        ),

        'Nostalgia General %': general['mention_nostalgia'],
        'Nostalgia High Impact %': high['mention_nostalgia'],
        'Nostalgia Difference (pp)': (
            high['mention_nostalgia'] - general['mention_nostalgia']
        ),

        'Value General %': general['mention_value'],
        'Value High Impact %': high['mention_value'],
        'Value Difference (pp)': (
            high['mention_value'] - general['mention_value']
        )
    })

reality_comparison = pd.DataFrame(comparison_rows)

print("=== CONTROLLED DIFFERENCE: HIGH IMPACT - GENERAL ===")

display(
    reality_comparison.round(2)
)
````

````python
# ============================================================
# 5) VISUALIZATION: REALITY CHECK
# ============================================================

fig, axes = plt.subplots(
    1, 2,
    figsize=(15, 5.5),
    dpi=120
)

# ------------------------------------------------------------
# Sound
# ------------------------------------------------------------

sns.barplot(
    data=reality_check_df,
    x='length_bin',
    y='mention_sound',
    hue='is_high_impact',
    ax=axes[0]
)

axes[0].set_title(
    'Sound Quality Mention Rate by Review Length',
    fontweight='bold'
)

axes[0].set_xlabel('Review Length')
axes[0].set_ylabel('Mention Rate (%)')

axes[0].legend(
    title='Review Group'
)

axes[0].grid(axis='y', alpha=0.3)

# ------------------------------------------------------------
# Nostalgia
# ------------------------------------------------------------

sns.barplot(
    data=reality_check_df,
    x='length_bin',
    y='mention_nostalgia',
    hue='is_high_impact',
    ax=axes[1]
)

axes[1].set_title(
    'Nostalgia Mention Rate by Review Length',
    fontweight='bold'
)

axes[1].set_xlabel('Review Length')
axes[1].set_ylabel('Mention Rate (%)')

axes[1].legend(
    title='Review Group'
)

axes[1].grid(axis='y', alpha=0.3)

plt.suptitle(
    '2.6 Reality Check: Review Length and Aspect Mention',
    fontsize=14,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
````

Reality Check: เมื่อแบ่งรีวิวตามความยาว พบว่า Mention Rate เพิ่มขึ้นตามความยาวของรีวิวทั้งในกลุ่ม General และ High Impact เช่น Sound Quality เพิ่มจาก 2.5% เป็น 5.9% และ 13.5% ในกลุ่ม General และจาก 5.1% เป็น 9.5% และ 20.5% ในกลุ่ม High Impact ตามลำดับ Short → Medium → Long รีวิว ดังนั้นความยาวของข้อความเป็นปัจจัยที่ต้องคำนึงถึงเมื่อเปรียบเทียบ Mention Rate ระหว่างกลุ่ม อย่างไรก็ตาม หลังแบ่งตามความยาวแล้ว ความแตกต่างระหว่าง High Impact และ General ยังไม่หายไปทั้งหมด โดยเฉพาะ Sound Quality ในกลุ่ม Long ที่ต่างกันประมาณ 7.0 percentage points จึงยังสรุปไม่ได้ว่า keyword เป็นสาเหตุของ Helpful Vote และควรใช้ผลนี้เป็นหลักฐานเชิงความสัมพันธ์เท่านั้น

````python
# ============================================================
# 2.7 — INSIGHT CARD 1
# REALITY CHECK: REVIEW LENGTH & ASPECT MENTION
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# STEP 1: เตรียมกลุ่ม Review
# ------------------------------------------------------------

df_card1 = df_joined.copy()

# แบ่ง High Impact / General
df_card1['review_group'] = np.where(
    df_card1['helpful_vote'] >= 5,
    'High Impact (≥5 votes)',
    'General (<5 votes)'
)

# ------------------------------------------------------------
# STEP 2: แบ่งตามความยาวรีวิว
# ------------------------------------------------------------

df_card1['length_group'] = pd.cut(
    df_card1['review_length'],
    bins=[0, 30, 100, np.inf],
    labels=[
        'Short\n≤30 words',
        'Medium\n31–100 words',
        'Long\n>100 words'
    ],
    include_lowest=True
)

# ------------------------------------------------------------
# STEP 3: สร้าง Summary Table
# ------------------------------------------------------------

card1_summary = (
    df_card1
    .groupby(
        ['length_group', 'review_group'],
        observed=False
    )
    .agg(
        n=('asin', 'size'),
        sound_rate=('mention_sound', 'mean'),
        nostalgia_rate=('mention_nostalgia', 'mean'),
        value_rate=('mention_value', 'mean')
    )
    .reset_index()
)

# แปลงจาก proportion → %
for col in [
    'sound_rate',
    'nostalgia_rate',
    'value_rate'
]:
    card1_summary[col] *= 100

print("========== CARD 1 EVIDENCE TABLE ==========")

display(
    card1_summary.round(2)
)
````

````python
# ============================================================
# CARD 1 — MAIN EVIDENCE GRAPH
# ALL ASPECTS × REVIEW LENGTH
# ============================================================

# เตรียมข้อมูลสำหรับ plot
card1_long = card1_summary.melt(
    id_vars=[
        'length_group',
        'review_group',
        'n'
    ],
    value_vars=[
        'sound_rate',
        'nostalgia_rate',
        'value_rate'
    ],
    var_name='aspect',
    value_name='mention_rate'
)

card1_long['aspect'] = card1_long['aspect'].map({
    'sound_rate': 'Sound Quality',
    'nostalgia_rate': 'Nostalgia',
    'value_rate': 'Value for Money'
})

# ------------------------------------------------------------
# Plot
# ------------------------------------------------------------

g = sns.catplot(
    data=card1_long,
    x='length_group',
    y='mention_rate',
    hue='review_group',
    col='aspect',
    kind='bar',
    height=5,
    aspect=0.85,
    sharey=False
)

g.set_axis_labels(
    'Review Length',
    'Mention Rate (%)'
)

g.set_titles(
    '{col_name}'
)

g.fig.suptitle(
    'Reality Check: Aspect Mention Rate by Review Length',
    fontsize=16,
    fontweight='bold',
    y=1.05
)

# ใส่ตัวเลขบนแท่ง
for ax in g.axes.flat:
    for container in ax.containers:
        ax.bar_label(
            container,
            fmt='%.1f%%',
            padding=2,
            fontsize=8
        )

    ax.grid(
        axis='y',
        alpha=0.25
    )

plt.tight_layout()
plt.show()
````

Evidence: Mention Rate เพิ่มขึ้นตามความยาวรีวิวในทั้ง General และ High Impact โดย Sound Quality เพิ่มจาก 2.5% → 5.9% → 13.5% ใน General และ 5.1% → 9.5% → 20.5% ใน High Impact จาก Short → Medium → Long

````python
# ===== ตัวอย่างที่ 1: Hacker News API (JSON, ไม่ต้องใช้ key) =====
# ได้หัวข้อข่าวพร้อม score และจำนวนคอมเมนต์ เหมาะกับการดูว่าเรื่องแบบไหนคนสนใจ
import urllib.request, json, time
import pandas as pd

def hn_get(url, timeout=30):
    with urllib.request.urlopen(url, timeout=timeout) as r:
        return json.loads(r.read().decode("utf-8"))

def fetch_hackernews(n=200, sleep=0.1):
    base = "https://hacker-news.firebaseio.com/v0"
    # ตัวอย่างนี้เก็บสูงสุด 200 stories ก็เพียงพอสำหรับ Part 3 เวอร์ชันลด workload
    ids = hn_get(f"{base}/topstories.json")[:n]
    rows = []
    for i, item_id in enumerate(ids):
        try:
            it = hn_get(f"{base}/item/{item_id}.json")
        except Exception as e:
            print("ข้าม", item_id, e)
            continue
        if not it or it.get("type") != "story" or not it.get("title"):
            continue
        rows.append({
            "doc_id": str(it["id"]),
            "text": it["title"] + (" " + it.get("text", "") if it.get("text") else ""),
            "score": it.get("score", 0),
            "n_comments": it.get("descendants", 0),
            "author": it.get("by"),
            "published_at": pd.to_datetime(it.get("time", 0), unit="s").isoformat(),
            "url": f"https://news.ycombinator.com/item?id={it['id']}",
            "source": "hackernews",
        })
        time.sleep(sleep)
        if (i + 1) % 100 == 0:
            print(f"ดึงแล้ว {i+1} รายการ")
    return rows

# rows = fetch_hackernews(50)
# print(len(rows)); rows[0]
````

### ตัวอย่างที่ 4: Article Collection Robot — อธิบายทีละขั้น

ตัวอย่างนี้ใช้เว็บไซต์ข่าวเพื่อให้เห็นแนวคิดของ Web Robot ชัดเจน

**เป้าหมาย:** ต้องการสร้างชุดข้อมูลที่

> **1 row = 1 บทความข่าว**

ดังนั้น Robot จะไม่ crawl ทุก link ในเว็บไซต์ เพราะจะได้หน้าเมนู หน้า category หน้า search หรือ URL ซ้ำ ๆ ปะปนมา

Robot จะทำงานเป็น 2 ช่วง:

`หน้ารวมข่าว → หา URL ของบทความ → เปิดบทความแต่ละ URL → เก็บ title/text/metadata → 1 ข่าว = 1 row`

ตัวอย่างด้านล่างใช้รูปแบบ URL ของ Thai PBS (`/news/content/<id>`) เพื่อสาธิตหลักการเท่านั้น  
ก่อนใช้จริงต้องตรวจ `robots.txt`, Terms of Service และปรับ URL pattern / selector ให้ตรงกับเว็บไซต์ที่ทีมเลือก

> **ข้อสังเกต:** ตัวอย่างตั้งจำนวนข่าวไว้ไม่มากเพื่อทดลองก่อน เมื่อโค้ดถูกแล้วจึงค่อยเพิ่มจำนวนหรือทำ pagination/archive


#### Step 1 — กำหนดเว็บไซต์และสิ่งที่ต้องการเก็บ

เริ่มจากกำหนดให้ชัดก่อนว่าเราจะเก็บอะไร

ในตัวอย่างนี้
* `LISTING_URL` = หน้ารวมข่าว ใช้สำหรับ **หา link**
* `ARTICLE_PATH` = รูปแบบ URL ที่ถือว่าเป็น **บทความจริง**
* `MAX_ARTICLES` = จำนวนข่าวที่ทดลองเก็บ
* `DELAY` = เวลาพักระหว่างการเรียกแต่ละหน้า เพื่อไม่ยิง request ถี่เกินไป

จุดสำคัญคือ **หน้ารวมข่าวไม่ใช่ record ที่จะเอาไปวิเคราะห์** แต่เป็นเพียงทางเข้าไปหา article URLs


````python
# ===== Step 1: ตั้งค่า =====
# หากยังไม่มี package:
# !pip install -q requests beautifulsoup4

import re
import time
import requests
import pandas as pd

from bs4 import BeautifulSoup
from urllib.parse import urljoin, urlparse, urldefrag
from urllib.robotparser import RobotFileParser

BASE_URL = "https://www.thaipbs.or.th"
LISTING_URL = "https://www.thaipbs.or.th/news"

# Thai PBS ใช้ URL ข่าวลักษณะ /news/content/509281
ARTICLE_PATH_PATTERN = r"^/news/content/\d+$"

MAX_ARTICLES = 20     # เริ่มน้อย ๆ เพื่อทดสอบก่อน
DELAY = 1.5           # พัก 1.5 วินาทีระหว่างหน้า
BOT_USER_AGENT = "KKU-TextAnalytics-StudentBot/1.0"

session = requests.Session()
session.headers.update({"User-Agent": BOT_USER_AGENT})

print("Listing page:", LISTING_URL)
print("เป้าหมาย: 1 row = 1 news article")
````

#### Step 2 — ทำความสะอาด URL และตรวจ `robots.txt`

URL เดียวกันอาจเขียนได้หลายแบบ เช่น

`https://www.thaipbs.or.th/news`  
`https://www.thaipbs.or.th/news#main-content`

ส่วนหลัง `#...` เรียกว่า **fragment** และโดยทั่วไปไม่ได้หมายถึงหน้าใหม่ จึงควรตัดออกก่อนตรวจ URL ซ้ำ

จากนั้นตรวจ `robots.txt` ก่อนเข้าแต่ละส่วนของเว็บ ถ้าไม่อนุญาตให้ bot เข้า path นั้น ให้ **เปลี่ยนแหล่งข้อมูล** ไม่ใช่หาวิธีหลบ


````python
# ===== Step 2: normalize URL + robots.txt =====

_robots_cache = {}

def normalize_url(url):
    """ตัด #fragment ออกจาก URL เพื่อไม่ให้หน้าเดียวกันถูกนับซ้ำ"""
    clean_url, _ = urldefrag(url)
    return clean_url.rstrip("/")


def robots_allowed(url, user_agent=BOT_USER_AGENT):
    """คืน True เมื่อ robots.txt อนุญาตให้ bot เข้าถึง URL"""
    p = urlparse(url)
    root = f"{p.scheme}://{p.netloc}"

    if root not in _robots_cache:
        rp = RobotFileParser()
        rp.set_url(urljoin(root, "/robots.txt"))
        try:
            rp.read()
            _robots_cache[root] = rp
        except Exception as e:
            print("อ่าน robots.txt ไม่ได้:", e)
            return False

    return _robots_cache[root].can_fetch(user_agent, url)


print("robots.txt อนุญาตหน้ารวมข่าวหรือไม่:", robots_allowed(LISTING_URL))
````

#### Step 3 — เข้า “หน้ารวมข่าว” เพื่อหาเฉพาะ Article URLs

ขั้นนี้ **ยังไม่เก็บเนื้อหาข่าว**

Robot เปิดหน้ารวมข่าว แล้วดู `<a href="...">` ทุก link จากนั้นกรองให้เหลือเฉพาะ URL ที่ตรงกับรูปแบบบทความ เช่น

`/news/content/509281`

การกรองนี้คือสิ่งที่ทำให้ Robot ต่างจาก generic crawler ที่วิ่งไปทุกหน้าใน domain


````python
# ===== Step 3: หา link ของ "บทความจริง" =====

def find_article_links(listing_url):
    if not robots_allowed(listing_url):
        raise PermissionError("robots.txt ไม่อนุญาตให้เข้าหน้ารวมข่าวนี้")

    r = session.get(listing_url, timeout=20)
    r.raise_for_status()

    soup = BeautifulSoup(r.text, "html.parser")
    article_urls = set()

    for a in soup.find_all("a", href=True):
        url = urljoin(listing_url, a["href"])
        url = normalize_url(url)

        p = urlparse(url)

        # 1) เอาเฉพาะ domain เดียวกับเว็บไซต์หลัก
        same_domain = (p.netloc == urlparse(BASE_URL).netloc)

        # 2) เอาเฉพาะ path ที่มีรูปแบบเป็นบทความข่าว
        is_article = bool(re.match(ARTICLE_PATH_PATTERN, p.path))

        if same_domain and is_article:
            article_urls.add(url)

    return sorted(article_urls)


article_urls = find_article_links(LISTING_URL)

print("พบ article URLs =", len(article_urls))
print("ตัวอย่าง 5 URLs:")
for u in article_urls[:5]:
    print(" -", u)
````

#### Step 4 — เปิดบทความทีละหน้าแล้วดึง “เนื้อหาที่ต้องการ”

ตอนนี้เรามี URL ของบทความแล้ว จึงเปิดทีละข่าวและแยก field ที่จะใช้วิเคราะห์ เช่น

* `url`
* `title`
* `published_at`
* `text`
* `word_count`

หลักสำคัญคือ **พยายามดึงเฉพาะตัวบทความ** ไม่ใช่ menu/footer/ข่าวแนะนำทั้งหน้า

ตัวอย่างใช้ `<article>` ก่อน ถ้าเว็บไซต์ไม่มี tag นี้จึง fallback ไปที่ `<main>` และนำข้อความจาก `<p>` มาต่อกัน  
เว็บไซต์อื่นอาจต้องเปลี่ยน selector เช่น `.article-body`, `.post-content` หรือ selector เฉพาะของเว็บนั้น


````python
# ===== Step 4: ดึงข้อมูลจากบทความ 1 หน้า =====

def extract_article(url):
    if not robots_allowed(url):
        print("robots.txt ไม่อนุญาต ข้าม:", url)
        return None

    r = session.get(url, timeout=20)
    r.raise_for_status()

    soup = BeautifulSoup(r.text, "html.parser")

    # ---- title ----
    h1 = soup.find("h1")
    title = h1.get_text(" ", strip=True) if h1 else ""

    # ---- published date: พยายามอ่านจาก meta หรือ time ----
    published_at = ""

    meta_time = soup.find("meta", attrs={"property": "article:published_time"})
    if meta_time and meta_time.get("content"):
        published_at = meta_time["content"]
    else:
        time_tag = soup.find("time")
        if time_tag:
            published_at = time_tag.get("datetime") or time_tag.get_text(" ", strip=True)

    # ---- article text ----
    container = soup.find("article") or soup.find("main")

    if container is None:
        return None

    paragraphs = []
    for p in container.find_all("p"):
        txt = p.get_text(" ", strip=True)

        # ตัดบรรทัดสั้น ๆ ที่มักไม่ใช่เนื้อหาหลัก
        if len(txt) >= 30:
            paragraphs.append(txt)

    text = " ".join(paragraphs).strip()

    # ถ้าดึงมาแล้วสั้นมาก มีโอกาสว่า selector ไม่ถูก
    if len(text) < 200:
        return None

    return {
        "doc_id": normalize_url(url),
        "url": normalize_url(url),
        "title": title,
        "published_at": published_at,
        "text": text,
        "word_count": len(text.split()),
        "source_domain": urlparse(url).netloc,
    }
````

#### Step 5 — ให้ Robot ทำซ้ำกับหลายบทความ

นี่คือส่วนที่ทำให้มันเป็น “Robot”

แทนที่จะเปิดข่าวด้วยมือทีละหน้า โปรแกรมจะวน URL ที่หาได้ใน Step 3 แล้วเรียก `extract_article()` ซ้ำ ๆ

มี `time.sleep(DELAY)` ทุกครั้งเพื่อไม่ส่ง request ถี่เกินไป

ในช่วงพัฒนาให้เริ่มจาก 5–20 ข่าวก่อน เมื่อดูตัวอย่างแล้วแน่ใจว่า `text` เป็นเนื้อหาข่าวจริง ค่อยเพิ่มจำนวน


````python
# ===== Step 5: เก็บหลายบทความอัตโนมัติ =====

def collect_articles(article_urls, max_articles=20, delay=1.5):
    rows = []

    for i, url in enumerate(article_urls[:max_articles], start=1):
        try:
            item = extract_article(url)

            if item is not None:
                rows.append(item)
                print(f"[{i}] OK   :", item["title"][:70])
            else:
                print(f"[{i}] SKIP :", url)

        except Exception as e:
            print(f"[{i}] ERROR:", url, e)

        time.sleep(delay)

    return rows


rows = collect_articles(
    article_urls,
    max_articles=MAX_ARTICLES,
    delay=DELAY
)

print("\nเก็บบทความที่ใช้ได้ =", len(rows))
````

#### Step 6 — ตรวจว่า “1 row = 1 ข่าว” จริงหรือไม่

อย่าเชื่อเพียงว่าโค้ดรันผ่าน ต้องเปิดดูตัวอย่างข้อมูลด้วย

สิ่งที่ต้องตรวจ:
1. URL แต่ละแถวเป็น article URL จริงหรือไม่
2. `title` ตรงกับข่าวหรือไม่
3. `text` เป็นเนื้อหาข่าว ไม่ใช่ menu/footer
4. มีข่าวซ้ำหรือไม่
5. ข้อความสั้นผิดปกติหรือไม่

หลังผ่านขั้นนี้จึงค่อยบันทึกเป็น corpus และนำไป Text Analytics


````python
# ===== Step 6: Quality Check =====

df_robot = pd.DataFrame(rows)

if len(df_robot) > 0:
    print("จำนวนก่อน clean =", len(df_robot))
    print("URL ซ้ำ =", df_robot["url"].duplicated().sum())
    print("ข้อความซ้ำ =", df_robot["text"].duplicated().sum())

    # ตัด URL ซ้ำและข้อความซ้ำ
    df_robot = (
        df_robot
        .drop_duplicates(subset=["url"])
        .drop_duplicates(subset=["text"])
        .copy()
    )

    # ตัวอย่างเกณฑ์ง่าย ๆ: ข่าวต้องมีอย่างน้อย 80 คำ
    df_robot = df_robot[df_robot["word_count"] >= 80].copy()

    print("จำนวนหลัง clean =", len(df_robot))

    display(
        df_robot[
            ["title", "published_at", "word_count", "url"]
        ].head(10)
    )

    # อ่านเนื้อหาจริง 1 ข่าวเพื่อตรวจด้วยตา
    print("\nตัวอย่าง text:\n")
    print(df_robot.iloc[0]["text"][:1500])
else:
    print("ยังไม่มีข้อมูล ตรวจ article pattern / selector / robots.txt")
````
