### 3.1 เลือกแหล่งข้อมูลและวิธีเก็บ

เลือกแหล่งข้อมูลที่สัมพันธ์กับคำถามของทีม แล้วใช้ **เพียง 1 วิธีเก็บ**

ตัวอย่าง:
* Hacker News API — post + score + comments
* RSS — ข่าว/บทความ
* Traffy Fondue Open Data — เรื่องร้องเรียน

ด้านล่างมี
* **ตัวอย่างที่ 1:** API แบบสั้น
* **ตัวอย่างที่ 2–3:** ทางเลือกอื่นแบบสรุป

````python
# ===== ตัวอย่างที่ 1: Hacker News API (JSON, ไม่ต้องใช้ key) =====
# ได้หัวข้อข่าวพร้อม score และจำนวนคอมเมนต์ เหมาะกับการดูว่าเรื่องแบบไหนคนสนใจ
import urllib.request, json, time
import pandas as pd

def hn_get(url, timeout=30):
    with urllib.request.urlopen(url, timeout=timeout) as r:
        return json.loads(r.read().decode("utf-8"))

def fetch_hackernews(n=200, sleep=0.1):
    base = "https://hacker-news.firebaseio.com/v0"
    # ตัวอย่างนี้เก็บสูงสุด 200 stories ก็เพียงพอสำหรับ Part 3 เวอร์ชันลด workload
    ids = hn_get(f"{base}/topstories.json")[:n]
    rows = []
    for i, item_id in enumerate(ids):
        try:
            it = hn_get(f"{base}/item/{item_id}.json")
        except Exception as e:
            print("ข้าม", item_id, e)
            continue
        if not it or it.get("type") != "story" or not it.get("title"):
            continue
        rows.append({
            "doc_id": str(it["id"]),
            "text": it["title"] + (" " + it.get("text", "") if it.get("text") else ""),
            "score": it.get("score", 0),
            "n_comments": it.get("descendants", 0),
            "author": it.get("by"),
            "published_at": pd.to_datetime(it.get("time", 0), unit="s").isoformat(),
            "url": f"https://news.ycombinator.com/item?id={it['id']}",
            "source": "hackernews",
        })
        time.sleep(sleep)
        if (i + 1) % 100 == 0:
            print(f"ดึงแล้ว {i+1} รายการ")
    return rows

# rows = fetch_hackernews(50)
# print(len(rows)); rows[0]
````

#### Step 2 — ทำความสะอาด URL และตรวจ `robots.txt`

URL เดียวกันอาจเขียนได้หลายแบบ เช่น

`https://www.thaipbs.or.th/news`  
`https://www.thaipbs.or.th/news#main-content`

ส่วนหลัง `#...` เรียกว่า **fragment** และโดยทั่วไปไม่ได้หมายถึงหน้าใหม่ จึงควรตัดออกก่อนตรวจ URL ซ้ำ

จากนั้นตรวจ `robots.txt` ก่อนเข้าแต่ละส่วนของเว็บ ถ้าไม่อนุญาตให้ bot เข้า path นั้น ให้ **เปลี่ยนแหล่งข้อมูล** ไม่ใช่หาวิธีหลบ


````python
# ===== Step 2: normalize URL + robots.txt =====

_robots_cache = {}

def normalize_url(url):
    """ตัด #fragment ออกจาก URL เพื่อไม่ให้หน้าเดียวกันถูกนับซ้ำ"""
    clean_url, _ = urldefrag(url)
    return clean_url.rstrip("/")


def robots_allowed(url, user_agent=BOT_USER_AGENT):
    """คืน True เมื่อ robots.txt อนุญาตให้ bot เข้าถึง URL"""
    p = urlparse(url)
    root = f"{p.scheme}://{p.netloc}"

    if root not in _robots_cache:
        rp = RobotFileParser()
        rp.set_url(urljoin(root, "/robots.txt"))
        try:
            rp.read()
            _robots_cache[root] = rp
        except Exception as e:
            print("อ่าน robots.txt ไม่ได้:", e)
            return False

    return _robots_cache[root].can_fetch(user_agent, url)


print("robots.txt อนุญาตหน้ารวมข่าวหรือไม่:", robots_allowed(LISTING_URL))
````

#### Step 3 — เข้า “หน้ารวมข่าว” เพื่อหาเฉพาะ Article URLs

ขั้นนี้ **ยังไม่เก็บเนื้อหาข่าว**

Robot เปิดหน้ารวมข่าว แล้วดู `<a href="...">` ทุก link จากนั้นกรองให้เหลือเฉพาะ URL ที่ตรงกับรูปแบบบทความ เช่น

`/news/content/509281`

การกรองนี้คือสิ่งที่ทำให้ Robot ต่างจาก generic crawler ที่วิ่งไปทุกหน้าใน domain


````python
# ===== Step 3: หา link ของ "บทความจริง" =====

def find_article_links(listing_url):
    if not robots_allowed(listing_url):
        raise PermissionError("robots.txt ไม่อนุญาตให้เข้าหน้ารวมข่าวนี้")

    r = session.get(listing_url, timeout=20)
    r.raise_for_status()

    soup = BeautifulSoup(r.text, "html.parser")
    article_urls = set()

    for a in soup.find_all("a", href=True):
        url = urljoin(listing_url, a["href"])
        url = normalize_url(url)

        p = urlparse(url)

        # 1) เอาเฉพาะ domain เดียวกับเว็บไซต์หลัก
        same_domain = (p.netloc == urlparse(BASE_URL).netloc)

        # 2) เอาเฉพาะ path ที่มีรูปแบบเป็นบทความข่าว
        is_article = bool(re.match(ARTICLE_PATH_PATTERN, p.path))

        if same_domain and is_article:
            article_urls.add(url)

    return sorted(article_urls)


article_urls = find_article_links(LISTING_URL)

print("พบ article URLs =", len(article_urls))
print("ตัวอย่าง 5 URLs:")
for u in article_urls[:5]:
    print(" -", u)
````

#### Step 4 — เปิดบทความทีละหน้าแล้วดึง “เนื้อหาที่ต้องการ”

ตอนนี้เรามี URL ของบทความแล้ว จึงเปิดทีละข่าวและแยก field ที่จะใช้วิเคราะห์ เช่น

* `url`
* `title`
* `published_at`
* `text`
* `word_count`

หลักสำคัญคือ **พยายามดึงเฉพาะตัวบทความ** ไม่ใช่ menu/footer/ข่าวแนะนำทั้งหน้า

ตัวอย่างใช้ `<article>` ก่อน ถ้าเว็บไซต์ไม่มี tag นี้จึง fallback ไปที่ `<main>` และนำข้อความจาก `<p>` มาต่อกัน  
เว็บไซต์อื่นอาจต้องเปลี่ยน selector เช่น `.article-body`, `.post-content` หรือ selector เฉพาะของเว็บนั้น


````python
# ===== Step 4: ดึงข้อมูลจากบทความ 1 หน้า =====

def extract_article(url):
    if not robots_allowed(url):
        print("robots.txt ไม่อนุญาต ข้าม:", url)
        return None

    r = session.get(url, timeout=20)
    r.raise_for_status()

    soup = BeautifulSoup(r.text, "html.parser")

    # ---- title ----
    h1 = soup.find("h1")
    title = h1.get_text(" ", strip=True) if h1 else ""

    # ---- published date: พยายามอ่านจาก meta หรือ time ----
    published_at = ""

    meta_time = soup.find("meta", attrs={"property": "article:published_time"})
    if meta_time and meta_time.get("content"):
        published_at = meta_time["content"]
    else:
        time_tag = soup.find("time")
        if time_tag:
            published_at = time_tag.get("datetime") or time_tag.get_text(" ", strip=True)

    # ---- article text ----
    container = soup.find("article") or soup.find("main")

    if container is None:
        return None

    paragraphs = []
    for p in container.find_all("p"):
        txt = p.get_text(" ", strip=True)

        # ตัดบรรทัดสั้น ๆ ที่มักไม่ใช่เนื้อหาหลัก
        if len(txt) >= 30:
            paragraphs.append(txt)

    text = " ".join(paragraphs).strip()

    # ถ้าดึงมาแล้วสั้นมาก มีโอกาสว่า selector ไม่ถูก
    if len(text) < 200:
        return None

    return {
        "doc_id": normalize_url(url),
        "url": normalize_url(url),
        "title": title,
        "published_at": published_at,
        "text": text,
        "word_count": len(text.split()),
        "source_domain": urlparse(url).netloc,
    }
````

#### Step 5 — ให้ Robot ทำซ้ำกับหลายบทความ

นี่คือส่วนที่ทำให้มันเป็น “Robot”

แทนที่จะเปิดข่าวด้วยมือทีละหน้า โปรแกรมจะวน URL ที่หาได้ใน Step 3 แล้วเรียก `extract_article()` ซ้ำ ๆ

มี `time.sleep(DELAY)` ทุกครั้งเพื่อไม่ส่ง request ถี่เกินไป

ในช่วงพัฒนาให้เริ่มจาก 5–20 ข่าวก่อน เมื่อดูตัวอย่างแล้วแน่ใจว่า `text` เป็นเนื้อหาข่าวจริง ค่อยเพิ่มจำนวน


````python
# ===== Step 5: เก็บหลายบทความอัตโนมัติ =====

def collect_articles(article_urls, max_articles=20, delay=1.5):
    rows = []

    for i, url in enumerate(article_urls[:max_articles], start=1):
        try:
            item = extract_article(url)

            if item is not None:
                rows.append(item)
                print(f"[{i}] OK   :", item["title"][:70])
            else:
                print(f"[{i}] SKIP :", url)

        except Exception as e:
            print(f"[{i}] ERROR:", url, e)

        time.sleep(delay)

    return rows


rows = collect_articles(
    article_urls,
    max_articles=MAX_ARTICLES,
    delay=DELAY
)

print("\nเก็บบทความที่ใช้ได้ =", len(rows))
````

### 3.3 หา Insight

ใช้ **1–2 เทคนิค** ที่เหมาะกับคำถาม เช่น

สร้างกราฟหรือตารางหลักอย่างน้อย **1 ชิ้น** แล้วตอบว่า

> “จากข้อมูลนี้ ผู้ใช้ผลวิเคราะห์ควรรู้อะไรหรือทำอะไรต่อ?”


````python
# ----------------- Your code here -----------------
import re
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

# 1. โหลดข้อมูล Final Corpus จากไฟล์ corpus_cleaned.jsonl
corp_analysis = pd.read_json("corpus_cleaned.jsonl", lines=True)
corp_analysis['published_at'] = pd.to_datetime(corp_analysis['published_at'])

# 2. ทำการสกัดคำศัพท์หรือคีย์เวิร์ดเด่น (Keyword Frequency)
# เนื่องจากเป็นภาษาไทย เราใช้การจับกลุ่มคีย์เวิร์ดที่เป็นประเด็นข่าวหลักเพื่อหาว่าข่าวพูดถึงอะไรมากที่สุด
keywords = {
    'ศาลรัฐธรรมนูญ / กฎหมาย': r'ศาล|ศาลรัฐธรรมนูญ|รัฐธรรมนูญ|คำวินิจฉัย|คดี',
    'การเลือกตั้ง / สว. / อบจ.': r'เลือกตั้ง|สว|กกต|อบจ|คูหา|ฮั้ว',
    'งบประมาณ / ดิจิทัลวอลเล็ต': r'งบ|งบประมาณ|ดิจิทัลวอลเล็ต|แจกเงิน',
    'ความขัดแย้ง / ชายแดนใต้': r'ชายแดนใต้|ความรุนแรง|ยะลา|ปัตตานี|นราธิวาส'
}

# นับจำนวนข่าวที่กล่าวถึงแต่ละประเด็น
keyword_counts = {}
for category, pattern in keywords.items():
    count = corp_analysis['text'].str.contains(pattern, flags=re.IGNORECASE, regex=True).sum()
    keyword_counts[category] = count

df_keywords = pd.DataFrame(list(keyword_counts.items()), columns=['ประเด็นข่าว', 'จำนวนข่าว']).sort_values(by='จำนวนข่าว', ascending=True)
````

````python
# ----------------- Your code here -----------------
# กราฟสำหรับนำเสนอ
# 3. สร้างกราฟแสดงความสนใจของสื่อ (Media Attention)
plt.style.use('default')
fig, ax = plt.subplots(figsize=(10, 5), facecolor='#FFFFFF')
ax.set_facecolor('#FFFFFF')

colors = ['#CCCCCC', '#94A3B8', '#1DB954', '#1E3A8A'] # เน้นประเด็นสำคัญสีเขียวเข้มและน้ำเงิน

bars = ax.barh(df_keywords['ประเด็นข่าว'], df_keywords['จำนวนข่าว'], color=colors[:len(df_keywords)], edgecolor='#0F172A', height=0.55)

# ใส่ตัวเลขบนแท่งกราฟ
for bar in bars:
    width = bar.get_width()
    ax.annotate(f'{int(width)} ข่าว ({width/len(corp_analysis)*100:.1f}%)',
                (width, bar.get_y() + bar.get_height() / 2.0),
                ha='left', va='center', xytext=(5, 0),
                textcoords='offset points', fontsize=10, fontweight='bold', color='#0F172A')

ax.set_title('การกระจายตัวของประเด็นข่าวในคลังข้อมูล (Media Attention Distribution)', fontsize=14, fontweight='bold', color='#0F172A', pad=15)
ax.set_xlabel('จำนวนบทความข่าว (ชิ้น)', fontsize=11, color='#0F172A')
ax.set_xlim(0, len(corp_analysis) * 1.2)

# ซ่อนเส้นขอบ
for spine in ['top', 'right', 'bottom']:
    ax.spines[spine].set_visible(False)
ax.spines['left'].set_color('#CCCCCC')
ax.grid(axis='x', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()

# 4. วิเคราะห์สรุปเชิงธุรกิจ
print("\n=== จากข้อมูลนี้ ผู้ใช้ผลวิเคราะห์ควรรู้อะไรหรือทำอะไรต่อ? ===")
print("1. สื่อให้ความพึงพอใจและโฟกัสหลักไปที่การตรวจสอบกลไกของรัฐและการเมืองส่วนกลาง (Political & Legal Scrutiny):")
print(f"   - ประเด็นข่าวเรื่อง 'การเลือกตั้ง / สว. / อบจ.' และ 'ศาลรัฐธรรมนูญ / กฎหมาย' ครองพื้นที่ข่าวกว่า 70% ของ Corpus ข้อมูล")
print("   - บ่งชี้ว่าผู้บริโภคสื่อในช่องทางนี้กำลังตื่นตัวและเสพข่าวเชิงโครงสร้างทางกฎหมายและการเมืองอย่างใกล้ชิด")
print("\n2. สิ่งที่แบรนด์หรือผู้ใช้ผลวิเคราะห์ควรทำต่อ (Strategic Actions):")
print("   - หลีกเลี่ยงการทำแคมเปญการตลาดเชิงอารมณ์ขันหรือเรื่องเบาสมองในช่วงเวลาที่กระแสข่าวเชิงตรวจสอบกำลังตึงเครียด")
print("   - ปรับการสื่อสารของแบรนด์ให้อยู่บนพื้นฐานของ 'ความโปร่งใส ความถูกต้อง และตรวจสอบได้' (Transparency & Integrity) สอดคล้องกับโทนกระแสสังคม")
print("   - สำหรับแบรนด์ที่ทำคอนเทนต์การเมือง/นโยบายสาธารณะ ควรขยายผลจากการสกัดประเด็นเหล่านี้ไปสู่การสร้าง 'สรุปประเด็นเข้าใจง่าย (Explainer Content)' เพื่อดึงยอด Engagement")
````

>
> **So what:** บ่งชี้ว่าบรรยากาศและความสนใจของสังคมส่วนใหญ่กำลังจดจ่ออยู่กับความโปร่งใส ความถูกต้อง และกฎกติกา แบรนด์หรือองค์กรที่ทำแคมเปญการสื่อสารในช่วงนี้ควรหลีกเลี่ยงข้อความที่ตลกโขกฮากหรือเบาสมองเกินไป เพราะอาจขัดต่ออารมณ์ร่วม (Social Sentiment) ของผู้เสพสื่อ
>
> **Action:** ทีม PR และ Marketing ของแบรนด์ควรปรับโทนการสื่อสาร (Brand Voice) มาสู่ความจริงใจและโปร่งใส (Transparency) และสำหรับเพจที่เน้นทำคอนเทนต์ ควรทำสรุปประเด็นซับซ้อนเชิงกฎหมาย/การเลือกตั้งให้สั้นกระชับและเข้าใจง่าย (Explainer Content) เพื่อดึง Engagement ระดับสูงจากผู้บริโภคที่กำลังมองหาคำชี้แจงที่ชัดเจน
>
> **ความมั่นใจ:** **กลาง (Medium)** — แม้จะเห็นทิศทางความสนใจของสื่อที่สอดคล้องกับสถานการณ์ปัจจุบันได้อย่างชัดเจน แต่ขนาดของชุดข้อมูลที่สุ่มมาวิเคราะห์เชิงสาธิตยังมีขนาดเล็ก (17 valid records หลังคลีน)
